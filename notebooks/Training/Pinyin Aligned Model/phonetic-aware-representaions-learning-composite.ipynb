Uses a checked shared experiment bundle; this notebook does not prepare or filter data.

```python
from pathlib import Path
import sys
def find_project_root(start=Path.cwd()):
    for candidate in (start.resolve(), *start.resolve().parents):
        if (candidate/'datasets').is_dir() and (candidate/'src').is_dir(): return candidate
    raise RuntimeError('project root not found')
PROJECT_ROOT=find_project_root(); sys.path.insert(0,str(PROJECT_ROOT/'src'))
from phonetics import load_experiment, train_experiment
```

```python
EXPERIMENT_ID='REPLACE_WITH_EXPLICIT_ID'
SEED=42
CONDITION='combined'
LAMBDA_MSE=1.0  # Use 100.0 only as a separate explicit run.
RUN_REAL_TRAINING=False
experiment_path=PROJECT_ROOT/'datasets'/'experiments'/EXPERIMENT_ID
experiment=load_experiment(experiment_path) if EXPERIMENT_ID!='REPLACE_WITH_EXPLICIT_ID' else None
```

```python
assert RUN_REAL_TRAINING, 'Set RUN_REAL_TRAINING=True only after explicit run approval'
result=train_experiment(experiment_path,condition=CONDITION,seed=SEED,lambda_mse=LAMBDA_MSE)
result['test_metrics']
```
