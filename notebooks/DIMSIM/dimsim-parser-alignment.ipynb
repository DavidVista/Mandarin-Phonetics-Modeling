# DIMSIM parser alignment

Compare original DIMSIM decomposition (A) with strict pypinyin components (B) on one pinned evaluation inventory. Raw components and parser errors are retained for every syllable, including zero initials. Numerical eligibility requires a nonempty initial under B.

All eligible unordered pairs are checked in both directions for symmetry and exact-path reproduction of A within the configured tolerance. Detailed distance exports cover the union of pairs touching any discrepant syllable. Category scopes can overlap; the overall row is recomputed from unique pairs. Comparative estimates use the same jointly valid pairs, with both directions equally weighted when asymmetry is detected.

95% percentile intervals describe sensitivity to inventory composition. Aggregate replicates resample syllables within fixed categories, including matching syllables; pair weights are products of endpoint multiplicities. Per-syllable replicates hold the focal syllable fixed and resample its partners. A/B and directions share draws. These intervals do not quantify uncertainty in the deterministic metric, training seeds, or speakers. Undefined estimates and invalid replicates are reported explicitly.


```python
from pathlib import Path
import sys

def find_project_root(start=None):
    current = (Path.cwd() if start is None else Path(start)).resolve()
    for parent in (current, *current.parents):
        for candidate in (parent, parent / "project"):
            if (candidate / "src" / "phonetics").is_dir():
                return candidate
    raise FileNotFoundError("Could not find project root")

ROOT = find_project_root()
sys.path.insert(0, str(ROOT / "src"))
from phonetics.evaluation.dimsim_alignment import run_alignment
```

```python
CONFIG = {
    "run_id": "duyu-chinese-lessons-zh-cn-v1",
    "input_path": ROOT / "datasets" / "evaluations" / "duyu-chinese-lessons-zh-cn-v1",
    "input_manifest_sha256": "1fbd485b38931369221e38dd30be5ea186de883b6c1fe7ace7000855645f43a2",
    "output_root": ROOT / "outputs" / "dimsim-parser-alignment",
    "seed": 42,
    "bootstrap_replicates": 1000,
    "confidence": 0.95,
    "atol": 1e-12,
    "rtol": 1e-10,
}
```

```python
output = run_alignment(CONFIG)
print(output)
```
