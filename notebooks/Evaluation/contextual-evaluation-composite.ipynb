```python
from pathlib import Path

def _reject_placeholders(value, name='CONFIG'):
    if isinstance(value, dict):
        for key, item in value.items():
            _reject_placeholders(item, f'{name}.{key}')
    elif isinstance(value, (list, tuple)):
        for index, item in enumerate(value):
            _reject_placeholders(item, f'{name}[{index}]')
    elif isinstance(value, (str, Path)) and 'EDIT_ME' in str(value).upper():
        raise ValueError(f'Replace placeholder {name} in the parameter cell')
_reject_placeholders(CONFIG)
_reject_placeholders(CANONICAL_SOURCE_ROOT, 'CANONICAL_SOURCE_ROOT')
if not CANONICAL_SOURCE_ROOT.is_dir():
    raise FileNotFoundError('Set CANONICAL_SOURCE_ROOT to the verified src/phonetics directory')
```
