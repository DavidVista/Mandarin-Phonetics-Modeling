### Model identifiers


```python
"""Objective identifiers and arbitrary finite acoustic loss weights."""
import math
import re
from numbers import Real
BASE_MODEL_IDS = ('syllable', 'components', 'acoustic')
COMBINED_PREFIX = 'components-acoustic-lambda-'

def combined_model_id(weight: Real) -> str:
    """Encode a finite coefficient without truncating its float precision."""
    if isinstance(weight, bool) or not isinstance(weight, Real):
        raise ValueError('lambda must be a finite real number')
    try:
        value = float(weight)
    except (OverflowError, ValueError) as exc:
        raise ValueError('lambda must be a finite real number') from exc
    if not math.isfinite(value):
        raise ValueError('lambda must be a finite real number')
    token = str(int(value)) if value.is_integer() and abs(value) < 1e+16 else repr(value).replace('e+', 'e')
    return COMBINED_PREFIX + token

def parse_model_id(model_id: str) -> tuple[str, float | None]:
    """Return the objective family and the combined acoustic coefficient."""
    if isinstance(model_id, str) and model_id in BASE_MODEL_IDS:
        return (model_id, None)
    if not isinstance(model_id, str) or not model_id.startswith(COMBINED_PREFIX):
        raise ValueError('unknown objective model_id')
    token = model_id[len(COMBINED_PREFIX):]
    if not re.fullmatch('[+-]?(?:\\d+(?:\\.\\d*)?|\\.\\d+)(?:[eE][+-]?\\d+)?', token):
        raise ValueError('combined model_id must contain a finite numeric lambda')
    weight = float(token)
    if not math.isfinite(weight):
        raise ValueError('lambda must be a finite real number')
    return ('components-acoustic', weight)

def canonical_model_id(model_id: str) -> str:
    family, weight = parse_model_id(model_id)
    return combined_model_id(weight) if family == 'components-acoustic' else family
```

### Pronunciation labels


```python
"""Project TONE3 input contract backed by pinned pypinyin inventories."""
from dataclasses import dataclass
from hashlib import sha256
from importlib.metadata import version
import re
from pypinyin.constants import PINYIN_DICT
from pypinyin.contrib.tone_convert import to_finals, to_initials, to_normal
PINYIN_CONTRACT_VERSION = 'pypinyin-tone3-1'

def pinyin_require_pypinyin_version():
    installed = version('pypinyin')
    if installed != '0.55.0':
        raise RuntimeError(f'pypinyin 0.55.0 required, found {installed}')
    return installed
pinyin_PYPINYIN_VERSION = pinyin_require_pypinyin_version()
VALID_SYLLABLES = frozenset((to_normal(reading, v_to_u=False).lower() for readings in PINYIN_DICT.values() for reading in readings.split(',')))
pinyin_INPUT_SYLLABLES = tuple(sorted((base for base in VALID_SYLLABLES if re.fullmatch('[a-z]+', base))))
pinyin_STRICT_COMPONENTS = tuple(((to_initials(base, strict=True), to_finals(base, strict=True, v_to_u=False)) for base in pinyin_INPUT_SYLLABLES))
INITIALS = sorted({initial for initial, final in pinyin_STRICT_COMPONENTS if final})
FINALS = sorted({final for _, final in pinyin_STRICT_COMPONENTS if final})
TONES = [str(tone) for tone in range(1, 6)]
PARSER_PROVENANCE = {'library': 'pypinyin', 'version': pinyin_PYPINYIN_VERSION, 'api': 'pypinyin.contrib.tone_convert', 'strict': True, 'input_format': 'lowercase-ascii-tone3', 'umlaut_storage': 'v', 'input_alias': 'u:', 'missing_tone': '5', 'empty_final': 'reject', 'inventory_source': 'pypinyin.constants.PINYIN_DICT', 'inventory_sha256': sha256('\n'.join(sorted(VALID_SYLLABLES)).encode('utf-8')).hexdigest(), 'vocabulary_sha256': sha256(repr((INITIALS, FINALS, TONES)).encode('utf-8')).hexdigest()}

@dataclass(frozen=True)
class PinyinComponents:
    initial: str
    final: str
    tone: str

def canonical_label(token: str, *, tone_convention='neutral_unmarked') -> str:
    """Accept one lowercase TONE3 syllable, allowing u: and unmarked neutral."""
    if tone_convention not in {'neutral_unmarked', 'explicit'}:
        raise ValueError(f'unsupported tone convention: {tone_convention!r}')
    if not isinstance(token, str) or not token:
        raise ValueError(f'Pinyin must be a non-empty string: {token!r}')
    if re.fullmatch('(?:[a-z]|u:)+[1-5]?', token) is None:
        raise ValueError(f'invalid Pinyin format (lowercase ASCII TONE3 required): {token!r}')
    canonical = token.replace('u:', 'v')
    if canonical[-1] not in TONES:
        if tone_convention == 'explicit':
            raise ValueError(f'explicit tone required: {token!r}')
        canonical += '5'
    if re.fullmatch('[a-z]+[1-5]', canonical) is None:
        raise ValueError(f'invalid canonical Pinyin format: {token!r}')
    base = canonical[:-1]
    if base not in VALID_SYLLABLES:
        raise ValueError(f'syllable absent from pypinyin dictionary: {token!r}')
    initial = to_initials(base, strict=True)
    final = to_finals(base, strict=True, v_to_u=False)
    if not final or initial not in INITIALS or final not in FINALS:
        raise ValueError(f'unsupported strict Pinyin components (non-empty final required): {token!r}')
    return canonical

def validate_canonical_label(token: str) -> str:
    """Reject noncanonical stored labels instead of silently repairing them."""
    if canonical_label(token, tone_convention='explicit') != token:
        raise ValueError(f'stored Pinyin label is not canonical: {token!r}')
    return token

def parse_components(token: str, *, tone_convention='neutral_unmarked') -> PinyinComponents:
    canonical = canonical_label(token, tone_convention=tone_convention)
    base = canonical[:-1]
    return PinyinComponents(to_initials(base, strict=True), to_finals(base, strict=True, v_to_u=False), canonical[-1])

def encode_components(components: PinyinComponents) -> tuple[int, int, int]:
    try:
        return (INITIALS.index(components.initial), FINALS.index(components.final), TONES.index(components.tone))
    except ValueError as exc:
        raise ValueError(f'components outside Pinyin vocabulary: {components!r}') from exc

def canonical_labels(tokens, *, tone_convention='neutral_unmarked'):
    return [canonical_label(token, tone_convention=tone_convention) for token in tokens]

def decompose(token: str, *, tone_convention='neutral_unmarked') -> tuple[int, int, int]:
    return encode_components(parse_components(token, tone_convention=tone_convention))
```

### Artifact files


```python
"""Small, deterministic artifact helpers for prepared datasets."""
import hashlib
import json
from pathlib import Path

def sha256_file(path: str | Path) -> str:
    digest = hashlib.sha256()
    with Path(path).open('rb') as handle:
        for block in iter(lambda: handle.read(1024 * 1024), b''):
            digest.update(block)
    return digest.hexdigest()

def write_json(path: str | Path, value) -> None:
    Path(path).write_text(json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2) + '\n', encoding='utf-8')

def write_jsonl(path: str | Path, rows) -> None:
    Path(path).write_text(''.join((json.dumps(row, ensure_ascii=False, sort_keys=True) + '\n' for row in rows)), encoding='utf-8')

def read_jsonl(path: str | Path) -> list[dict]:
    result = []
    for number, line in enumerate(Path(path).read_text(encoding='utf-8').splitlines(), 1):
        try:
            value = json.loads(line)
        except json.JSONDecodeError as exc:
            raise ValueError(f'invalid JSONL at line {number}') from exc
        if not isinstance(value, dict):
            raise ValueError(f'JSONL row {number} is not an object')
        result.append(value)
    return result

def safe_artifact(root: str | Path, relative: str) -> Path:
    root = Path(root).resolve()
    rel = Path(relative)
    if rel.is_absolute() or '..' in rel.parts:
        raise ValueError(f'unsafe artifact path: {relative}')
    path = (root / rel).resolve()
    if path != root and root not in path.parents:
        raise ValueError(f'unsafe artifact path: {relative}')
    cursor = root
    for part in rel.parts:
        cursor = cursor / part
        if cursor.exists() and cursor.is_symlink():
            raise ValueError(f'symlink artifact path: {relative}')
    return path
```

### IDS vocabulary


```python
"""Pinned, deterministic IDS parsing and token mapping."""
import hashlib
import json
from pathlib import Path
IDS_SCHEMA_VERSION = 1
IDS_RULES_VERSION = 'ids-prefix-1'
PAD = '[PAD]'
UNKNOWN = '？'
OPERATORS = {chr(code) for code in range(12272, 12284)}
TERNARY_OPERATORS = {'⿲', '⿳'}
UNSUPPORTED_OPERATORS = {chr(code) for code in range(12284, 12288)} | {'\u31ef'}
EXPECTED_SNAPSHOT = '618dbab8a8ddefb958763c8b4afbaa741a4460de'
EXPECTED_LICENSE = 'LGPL-3.0-or-later'
EXPECTED_ATTRIBUTION = 'Make Me a Hanzi; dictionary.txt derived from Unihan and CJKlib'

def parse_expression(expression: str) -> tuple[str, ...]:
    if not isinstance(expression, str) or not expression:
        raise ValueError('IDS decomposition must be a non-empty string')
    if any((char in UNSUPPORTED_OPERATORS for char in expression)):
        raise ValueError('unsupported IDS operator')
    cursor = 0

    def parse_node() -> None:
        nonlocal cursor
        if cursor >= len(expression):
            raise ValueError('IDS expression has too few operands')
        symbol = expression[cursor]
        cursor += 1
        if symbol in OPERATORS:
            arity = 3 if symbol in TERNARY_OPERATORS else 2
            for _ in range(arity):
                parse_node()
    parse_node()
    if cursor != len(expression):
        raise ValueError('IDS expression has trailing operands')
    return tuple(expression)

def read_dictionary(path: str | Path) -> dict[str, dict]:
    path = Path(path)
    if not path.is_file():
        raise FileNotFoundError(path)
    mapping: dict[str, dict] = {}
    with path.open('r', encoding='utf-8', newline='') as handle:
        for line_number, line in enumerate(handle, 1):
            try:
                row = json.loads(line)
            except json.JSONDecodeError as exc:
                raise ValueError(f'invalid IDS JSONL at line {line_number}') from exc
            if not isinstance(row, dict):
                raise ValueError(f'IDS line {line_number} must be an object')
            character, decomposition = (row.get('character'), row.get('decomposition'))
            if not isinstance(character, str) or len(character) != 1:
                raise ValueError(f'IDS line {line_number} character must be one code point')
            if character in mapping:
                raise ValueError(f'duplicate IDS character: {character}')
            if not isinstance(decomposition, str) or not decomposition:
                raise ValueError(f'empty or invalid IDS decomposition for {character}')
            parsed = parse_expression(decomposition)
            if UNKNOWN in decomposition:
                tokens, mode = ((character,), 'character_fallback')
            else:
                tokens, mode = (parsed, 'decomposition')
            mapping[character] = {'decomposition': decomposition, 'tokens': list(tokens), 'mode': mode}
    if not mapping:
        raise ValueError('IDS dictionary is empty')
    return dict(sorted(mapping.items(), key=lambda item: tuple(map(ord, item[0]))))

def validate_pin(spec: dict) -> tuple[Path, str]:
    expected = {'path', 'sha256', 'snapshot_commit', 'license', 'attribution'}
    if not isinstance(spec, dict) or set(spec) != expected:
        raise ValueError('IDS pin must contain path, sha256, snapshot_commit, license, attribution')
    path = Path(spec['path']).resolve()
    if not path.is_file():
        raise FileNotFoundError(path)
    digest = sha256_file(path)
    if digest != spec['sha256']:
        raise ValueError('IDS dictionary checksum mismatch')
    if spec['snapshot_commit'] != EXPECTED_SNAPSHOT:
        raise ValueError('IDS snapshot commit mismatch')
    if spec['license'] != EXPECTED_LICENSE or not isinstance(spec['attribution'], str) or (not spec['attribution'].strip()):
        raise ValueError('IDS license or attribution mismatch')
    return (path, digest)

def build_mapping(spec: dict, *, dictionary_path: str | Path | None=None) -> tuple[dict, dict, dict]:
    if dictionary_path is None:
        path, digest = validate_pin(spec)
        records_path = path
    else:
        expected = {'path', 'sha256', 'snapshot_commit', 'license', 'attribution'}
        if not isinstance(spec, dict) or set(spec) != expected:
            raise ValueError('IDS pin must contain path, sha256, snapshot_commit, license, attribution')
        records_path = Path(dictionary_path).resolve()
        if not records_path.is_file() or sha256_file(records_path) != spec['sha256']:
            raise ValueError('bundled IDS dictionary checksum mismatch')
        if spec['snapshot_commit'] != EXPECTED_SNAPSHOT or spec['license'] != EXPECTED_LICENSE or (not str(spec['attribution']).strip()):
            raise ValueError('IDS source provenance mismatch')
        path, digest = (spec['path'], spec['sha256'])
    records = read_dictionary(records_path)
    tokens = sorted({token for record in records.values() for token in record['tokens']}, key=ord)
    vocabulary = [PAD, *tokens]
    stoi = {token: index for index, token in enumerate(vocabulary)}
    encoded = {character: {**record, 'ids': [stoi[token] for token in record['tokens']]} for character, record in records.items()}
    provenance = {'path': str(path), 'sha256': digest, 'snapshot_commit': spec['snapshot_commit'], 'license': spec['license'], 'attribution': spec['attribution']}
    mapping_artifact = {'schema_version': IDS_SCHEMA_VERSION, 'rules_version': IDS_RULES_VERSION, 'provenance': provenance, 'characters': encoded}
    vocabulary_artifact = {'schema_version': IDS_SCHEMA_VERSION, 'rules_version': IDS_RULES_VERSION, 'tokens': vocabulary, 'stoi': stoi}
    return (mapping_artifact, vocabulary_artifact, provenance)

def encode_text(text: str, mapping: dict, vocabulary: dict) -> tuple[list[list[int]], list[int], str | None]:
    sequences, lengths = ([], [])
    for character in text:
        if character not in mapping['characters']:
            raise ValueError(f'character is absent from IDS mapping: {character}')
        record = mapping['characters'][character]
        if len(record['ids']) != len(record['tokens']) or any((type(i) is not int or i <= 0 or i >= len(vocabulary['tokens']) or (vocabulary['tokens'][i] != token) for token, i in zip(record['tokens'], record['ids']))):
            raise ValueError(f'invalid IDS token ID in mapping: {character}')
        sequences.append(list(record['ids']))
        lengths.append(len(record['ids']))
    return (sequences, lengths, None)

def content_hash(value: dict) -> str:
    raw = json.dumps(value, ensure_ascii=False, sort_keys=True, separators=(',', ':')).encode('utf-8')
    return hashlib.sha256(raw).hexdigest()
```

```python
"""Portable dataset-only experiment bundles with no ML runtime dependency."""
import copy
import json
import math
import os
import pickle
import random
import shutil
import tempfile
from pathlib import Path, PureWindowsPath
import numpy as np
PREPARED_SHA256 = 'c562722d2a866f4f1904c746c86c5863a310a66ee3875d86fc36f254fd2d7c10'
experiment_SCHEMA_VERSION = 3
DEFAULT_MODEL_IDS = ['syllable', 'components', 'acoustic', 'components-acoustic-lambda-1', 'components-acoustic-lambda-10']
REQUIRED_FILES = {'samples.jsonl', 'ids/mapping.json', 'ids/vocabulary.json', 'ids/dictionary.jsonl', 'ids/COPYING', 'ids/LGPL', 'acoustic_targets.npy', 'vocab.json', 'splits.json', 'config.json', 'provenance/prepared_manifest.json', 'provenance/processed_manifest.json', 'provenance/processed_config.json', 'provenance/acoustic_manifest.json', 'provenance/sound_inventory.jsonl', 'provenance/prepared_corpus.jsonl', 'provenance/prepared_acoustic_targets.pkl'}
experiment_ARCH = {'embedding_dim': 64, 'ids_bigru_hidden_per_direction': 32, 'ids_bigru_layers': 1, 'sentence_bilstm_hidden_per_direction': 64, 'sentence_bilstm_layers': 1, 'context_dim': 128, 'classification_hidden_dim': 64, 'classification_dropout': 0.5}
experiment_TRAIN = {'optimizer': 'Adam', 'learning_rate': 0.001, 'weight_decay': 0.0, 'batch_size': 32, 'epochs': 20, 'scheduler': None, 'early_stopping': False, 'gradient_clipping': None, 'cosine_epsilon': 1e-08}

def default_experiment_config(acoustic_dim=1024):
    """Return a fresh config with the agreed protocol and current input pin."""
    return {'schema_version': experiment_SCHEMA_VERSION, 'prepared_manifest_sha256': PREPARED_SHA256, 'split_ratios': [0.8, 0.1, 0.1], 'split_seed': 0, 'training_seeds': [0, 1, 2, 3, 4], 'architecture': {**experiment_ARCH, 'acoustic_dim': acoustic_dim}, 'training': copy.deepcopy(experiment_TRAIN), 'model_ids': list(DEFAULT_MODEL_IDS)}

def experiment_validate_config(value, acoustic_dim):
    expected = default_experiment_config(acoustic_dim)
    if not isinstance(value, dict) or set(value) != set(expected):
        raise ValueError('invalid experiment config keys')
    cfg = copy.deepcopy(value)

    def exact_type_value(actual, wanted):
        if type(actual) is not type(wanted):
            return False
        if isinstance(wanted, dict):
            return set(actual) == set(wanted) and all((exact_type_value(actual[k], wanted[k]) for k in wanted))
        if isinstance(wanted, list):
            return len(actual) == len(wanted) and all((exact_type_value(a, b) for a, b in zip(actual, wanted)))
        return actual == wanted
    if type(cfg['schema_version']) is not int or cfg['schema_version'] != experiment_SCHEMA_VERSION or (not exact_type_value(cfg['architecture'], expected['architecture'])) or (not exact_type_value(cfg['training'], experiment_TRAIN)):
        raise ValueError('experiment config conflicts with the training protocol')
    model_ids = cfg['model_ids']
    if not isinstance(model_ids, list) or not model_ids:
        raise ValueError('model_ids must be a nonempty list')
    canonical_ids = [canonical_model_id(model_id) for model_id in model_ids]
    if len(set(canonical_ids)) != len(canonical_ids):
        raise ValueError('model_ids must identify distinct objectives')
    pin = cfg['prepared_manifest_sha256']
    if not isinstance(pin, str) or len(pin) != 64 or any((c not in '0123456789abcdef' for c in pin)):
        raise ValueError('prepared_manifest_sha256 must be a lowercase SHA-256')
    ratios = cfg['split_ratios']
    if not isinstance(ratios, list) or len(ratios) != 3 or any((type(x) not in (int, float) or not math.isfinite(x) or x <= 0 for x in ratios)) or (not math.isclose(sum(ratios), 1.0, rel_tol=0.0, abs_tol=1e-09)):
        raise ValueError('split_ratios must contain three finite positive values summing to one')
    if type(cfg['split_seed']) is not int or cfg['split_seed'] != 0:
        raise ValueError('split_seed must be the fixed protocol value 0')
    seeds = cfg['training_seeds']
    if not isinstance(seeds, list) or len(seeds) != 5 or any((type(s) is not int for s in seeds)) or (seeds != [0, 1, 2, 3, 4]):
        raise ValueError('training_seeds must be the fixed protocol values [0, 1, 2, 3, 4]')
    return cfg

def experiment_split_ids(ids, seed, ratios):
    values = sorted(ids)
    random.Random(seed).shuffle(values)
    n = len(values)
    train_end = int(n * ratios[0])
    validation_end = train_end + int(n * ratios[1])
    result = {'train': values[:train_end], 'validation': values[train_end:validation_end], 'test': values[validation_end:]}
    if any((not group for group in result.values())):
        raise ValueError('train, validation, and test splits must all be nonempty')
    return result

def experiment_safe_id(value):
    return isinstance(value, str) and bool(value) and (value not in {'.', '..'}) and ('/' not in value) and ('\\' not in value) and (Path(value).name == value) and (PureWindowsPath(value).name == value)

def experiment_validate_semantics(root, manifest, config, rows, splits, vocab, matrix):
    if manifest.get('pinyin_contract') != PINYIN_CONTRACT_VERSION or manifest.get('parser_provenance') != PARSER_PROVENANCE or manifest.get('ids_rules') != IDS_RULES_VERSION:
        raise ValueError('unsupported experiment provenance contract')
    pin = manifest.get('prepared_manifest_sha256')
    if sha256_file(root / 'provenance/prepared_manifest.json') != pin or pin != config['prepared_manifest_sha256']:
        raise ValueError('prepared provenance pin mismatch')
    prepared = json.loads((root / 'provenance/prepared_manifest.json').read_text(encoding='utf-8'))
    if prepared.get('dataset_id') != manifest.get('prepared_dataset_id'):
        raise ValueError('prepared dataset identity mismatch')
    processed_path = root / 'provenance/processed_manifest.json'
    processed = json.loads(processed_path.read_text(encoding='utf-8'))
    acoustic_path = root / 'provenance/acoustic_manifest.json'
    acoustic_manifest = json.loads(acoustic_path.read_text(encoding='utf-8'))
    inventory_path = root / 'provenance/sound_inventory.jsonl'
    inventory = read_jsonl(inventory_path)
    inventory_hash = sha256_file(inventory_path)
    corpus_path = root / 'provenance/prepared_corpus.jsonl'
    acoustic_targets_path = root / 'provenance/prepared_acoustic_targets.pkl'
    if prepared.get('artifacts', {}).get('corpus.jsonl') != sha256_file(corpus_path) or prepared.get('artifacts', {}).get('acoustic/targets.pkl') != sha256_file(acoustic_targets_path) or prepared.get('processed_corpus_manifest_sha256') != sha256_file(processed_path) or (prepared.get('acoustic_bundle_manifest_sha256') != sha256_file(acoustic_path)) or (processed.get('sources', {}).get('sound_inventory_sha256') != inventory_hash) or (acoustic_manifest.get('inventory_sha256') != inventory_hash) or (processed.get('artifacts', {}).get('config.json') != sha256_file(root / 'provenance/processed_config.json')):
        raise ValueError('copied upstream provenance does not agree')
    if prepared.get('artifacts', {}).get('ids/mapping.json') != sha256_file(root / 'ids/mapping.json'):
        raise ValueError('prepared IDS mapping provenance mismatch')
    if prepared.get('artifacts', {}).get('ids/vocabulary.json') != sha256_file(root / 'ids/vocabulary.json'):
        raise ValueError('prepared IDS vocabulary provenance mismatch')
    if prepared.get('artifacts', {}).get('ids/dictionary.jsonl') != sha256_file(root / 'ids/dictionary.jsonl'):
        raise ValueError('prepared IDS dictionary provenance mismatch')
    for name, digest in manifest['provenance'].items():
        if sha256_file(safe_artifact(root, name)) != digest:
            raise ValueError(f'provenance checksum mismatch: {name}')
    if set(manifest['provenance']) != {n for n in manifest['artifacts'] if n.startswith('provenance/')}:
        raise ValueError('provenance file set mismatch')
    original_vocab = json.loads((root / 'ids/vocabulary.json').read_text(encoding='utf-8'))
    original_mapping = json.loads((root / 'ids/mapping.json').read_text(encoding='utf-8'))
    bundled_vocab = vocab['ids']
    if bundled_vocab != original_vocab or vocab['ids_mapping'] != original_mapping:
        raise ValueError('duplicated IDS mapping or vocabulary mismatch')
    if original_vocab.get('tokens', [None])[0] != '[PAD]' or original_vocab.get('stoi', {}).get('[PAD]') != 0:
        raise ValueError('IDS PAD token must have ID zero')
    ids_cfg = prepared.get('ids_provenance')
    rebuilt_mapping, rebuilt_vocab, _ = build_mapping(ids_cfg, dictionary_path=root / 'ids/dictionary.jsonl')
    if rebuilt_mapping != original_mapping or rebuilt_vocab != original_vocab:
        raise ValueError('bundled IDS mapping or vocabulary fails reconstruction')
    tokens = vocab.get('acoustic_tokens')
    if not isinstance(tokens, list) or not tokens or tokens != sorted(tokens) or (len(tokens) != len(set(tokens))) or (vocab.get('syllables') != tokens) or (vocab.get('labels') != {'initials': INITIALS, 'finals': FINALS, 'tones': TONES}):
        raise ValueError('invalid acoustic or syllable vocabulary')
    inventory_labels = [row.get('label') for row in inventory]
    if sorted(inventory_labels) != tokens or len(inventory_labels) != len(set(inventory_labels)):
        raise ValueError('acoustic inventory does not match full syllable vocabulary')
    key_ids = {label: i for i, label in enumerate(tokens)}
    for label in tokens:
        validate_canonical_label(label)
    if matrix.shape != (len(tokens), config['architecture']['acoustic_dim']) or not np.isfinite(matrix).all() or np.any(np.linalg.norm(matrix.astype(np.float64), axis=1) <= 1e-12):
        raise ValueError('invalid acoustic target matrix')
    with acoustic_targets_path.open('rb') as handle:
        pinned_acoustic = pickle.load(handle)
    pinned_keys = sorted(pinned_acoustic)
    pinned_matrix = np.stack([np.asarray(pinned_acoustic[k], dtype=np.float32) for k in pinned_keys])
    if pinned_keys != tokens or pinned_matrix.shape != matrix.shape or (not np.array_equal(pinned_matrix, matrix)):
        raise ValueError('acoustic matrix does not reproduce pinned prepared targets')
    mapping_chars = original_mapping.get('characters', {})
    id_count = len(original_vocab['tokens'])
    seen_ids = []
    seen_sources = set()
    for row in rows:
        required = {'component_lengths', 'ids_sequences', 'labels', 'positions', 'sample_id', 'sentence_length', 'source_ids', 'text', 'component_labels', 'syllable_ids', 'acoustic_ids'}
        if set(row) != required:
            raise ValueError('sample keys do not match experiment schema')
        text, labels = (row['text'], row['labels'])
        n = len(text) if isinstance(text, str) else -1
        if n <= 0 or type(row['sentence_length']) is not int or row['sentence_length'] != n:
            raise ValueError('invalid sample text or sentence length')
        if row['sample_id'] != sample_id(text, labels):
            raise ValueError('sample ID does not match text and labels')
        if not isinstance(row['positions'], list) or any((type(pos) is not int for pos in row['positions'])) or row['positions'] != list(range(n)) or (not isinstance(labels, list)) or (len(labels) != n) or (not isinstance(row['ids_sequences'], list)) or (len(row['ids_sequences']) != n) or (not isinstance(row['component_lengths'], list)) or (len(row['component_lengths']) != n):
            raise ValueError('sample alignment mismatch')
        if not isinstance(row['source_ids'], list) or not row['source_ids'] or any((not isinstance(s, str) or not s for s in row['source_ids'])) or (len(set(row['source_ids'])) != len(row['source_ids'])):
            raise ValueError('invalid sample source IDs')
        if seen_sources.intersection(row['source_ids']):
            raise ValueError('source IDs occur in more than one sample')
        seen_sources.update(row['source_ids'])
        expected_components, expected_syllables, expected_sequences = ([], [], [])
        for char, label, sequence, length in zip(text, labels, row['ids_sequences'], row['component_lengths']):
            canonical = validate_canonical_label(label)
            expected_components.append(list(decompose(canonical)))
            if canonical not in key_ids:
                raise ValueError(f'missing acoustic target for label: {canonical}')
            expected_syllables.append(key_ids[canonical])
            if not isinstance(sequence, list) or not sequence or type(length) is not int or (length != len(sequence)):
                raise ValueError('invalid IDS sequence or component length')
            if any((type(token) is not int or token <= 0 or token >= id_count for token in sequence)):
                raise ValueError('IDS token ID outside vocabulary or padding')
            char_entry = mapping_chars.get(char)
            if char_entry is None or sequence != char_entry.get('ids'):
                raise ValueError('IDS sequence does not match character mapping')
            expected_sequences.append(sequence)
        if not isinstance(row['component_labels'], list) or any((not isinstance(x, list) or any((type(i) is not int for i in x)) for x in row['component_labels'])) or row['component_labels'] != expected_components:
            raise ValueError('component targets mismatch canonical Pinyin')
        if any((type(i) is not int for i in row['syllable_ids'] + row['acoustic_ids'])):
            raise ValueError('syllable/acoustic IDs must be integers')
        if row['syllable_ids'] != expected_syllables or row['acoustic_ids'] != expected_syllables:
            raise ValueError('syllable/acoustic targets mismatch canonical Pinyin')
        seen_ids.append(row['sample_id'])
    if seen_ids != sorted(seen_ids) or len(seen_ids) != len(set(seen_ids)):
        raise ValueError('sample ID order or uniqueness invalid')
    source_rows = read_jsonl(corpus_path)
    base_fields = {'component_lengths', 'ids_sequences', 'labels', 'positions', 'sample_id', 'sentence_length', 'source_ids', 'text'}
    projected_rows = [{key: row[key] for key in base_fields} for row in rows]
    if projected_rows != source_rows:
        raise ValueError('experiment samples do not match pinned prepared corpus')
    if set(splits) != {'train', 'validation', 'test'}:
        raise ValueError('split names are invalid')
    split_ids = [sid for name in ('train', 'validation', 'test') for sid in splits[name]]
    if any((not splits[k] for k in splits)) or len(split_ids) != len(set(split_ids)) or set(split_ids) != set(seen_ids):
        raise ValueError('split membership is incomplete or duplicated')
    if splits != experiment_split_ids(seen_ids, config['split_seed'], config['split_ratios']):
        raise ValueError('splits do not match deterministic config')
    counts = manifest.get('counts')
    expected_counts = {'samples': len(rows), 'positions': sum((row['sentence_length'] for row in rows)), 'acoustic_targets': len(tokens), 'acoustic_dim': matrix.shape[1]}
    if counts != expected_counts:
        raise ValueError('experiment counts mismatch')

def validate_experiment(path):
    root = Path(path).resolve()
    manifest = json.loads((root / 'manifest.json').read_text(encoding='utf-8'))
    if set(manifest) != {'bundle_type', 'schema_version', 'experiment_id', 'pinyin_contract', 'parser_provenance', 'ids_rules', 'prepared_manifest_sha256', 'prepared_dataset_id', 'provenance', 'counts', 'artifacts'}:
        raise ValueError('invalid experiment manifest keys')
    if manifest.get('bundle_type') != 'experiment_dataset' or manifest.get('schema_version') != experiment_SCHEMA_VERSION or (not experiment_safe_id(manifest.get('experiment_id'))):
        raise ValueError('unsupported experiment bundle')
    artifacts = manifest['artifacts']
    if set(artifacts) != REQUIRED_FILES:
        raise ValueError('experiment required artifact set mismatch')
    actual = {str(p.relative_to(root)).replace('\\', '/') for p in root.rglob('*') if p.is_file() and p.name != 'manifest.json'}
    if actual != set(artifacts):
        raise ValueError('experiment contains untracked or missing artifacts')
    for relative, digest in artifacts.items():
        p = safe_artifact(root, relative)
        if not p.is_file() or not isinstance(digest, str) or sha256_file(p) != digest:
            raise ValueError(f'artifact checksum mismatch: {relative}')
    config = json.loads((root / 'config.json').read_text(encoding='utf-8'))
    matrix = np.load(root / 'acoustic_targets.npy', allow_pickle=False)
    config = experiment_validate_config(config, matrix.shape[1] if matrix.ndim == 2 else -1)
    if manifest.get('prepared_manifest_sha256') != config['prepared_manifest_sha256']:
        raise ValueError('manifest and config input pins disagree')
    vocabulary = json.loads((root / 'vocab.json').read_text(encoding='utf-8'))
    splits = json.loads((root / 'splits.json').read_text(encoding='utf-8'))
    rows = read_jsonl(root / 'samples.jsonl')
    experiment_validate_semantics(root, manifest, config, rows, splits, vocabulary, matrix)
    return manifest

def load_experiment(path):
    """Load a portable bundle; samples are keyed by sample_id."""
    manifest = validate_experiment(path)
    root = Path(path).resolve()
    return {'root': root, 'manifest': manifest, 'config': json.loads((root / 'config.json').read_text(encoding='utf-8')), 'vocab': json.loads((root / 'vocab.json').read_text(encoding='utf-8')), 'splits': json.loads((root / 'splits.json').read_text(encoding='utf-8')), 'samples': {row['sample_id']: row for row in read_jsonl(root / 'samples.jsonl')}, 'acoustic_targets': np.load(root / 'acoustic_targets.npy', allow_pickle=False)}
```

### Encoder random streams


```python
"""Deterministic training seeds and serializable RNG snapshots."""
import random
import numpy as np
import torch

def seed_everything(seed: int, deterministic: bool=True) -> None:
    if type(seed) is not int or seed < 0:
        raise ValueError('seed must be a nonnegative integer')
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    if deterministic:
        torch.use_deterministic_algorithms(True, warn_only=True)
        if hasattr(torch.backends, 'cudnn'):
            torch.backends.cudnn.benchmark = False

def make_generator(seed: int, stream: int=0) -> torch.Generator:
    generator = torch.Generator()
    generator.manual_seed(int(seed) + 1000003 * int(stream))
    return generator

def rng_snapshot(loader_generator: torch.Generator | None=None) -> dict:
    numpy_state = np.random.get_state()
    state = {'python': list(random.getstate()), 'numpy': {'algorithm': numpy_state[0], 'keys': numpy_state[1].tolist(), 'position': int(numpy_state[2]), 'has_gauss': int(numpy_state[3]), 'cached_gaussian': float(numpy_state[4])}, 'torch_cpu': torch.get_rng_state(), 'torch_cuda': torch.cuda.get_rng_state_all() if torch.cuda.is_available() else []}
    if loader_generator is not None:
        state['dataloader'] = loader_generator.get_state()
    return state
```

### Sentence batches


```python
"""Torch dataset and collation for portable schema-3 experiment bundles."""
import torch
from torch.utils.data import DataLoader, Dataset

class ExperimentDataset(Dataset):

    def __init__(self, sample_ids, samples):
        self.sample_ids = list(sample_ids)
        self.samples = samples

    def __len__(self):
        return len(self.sample_ids)

    def __getitem__(self, index):
        return self.samples[self.sample_ids[index]]

def collate_experiment(samples):
    if not samples:
        raise ValueError('cannot collate an empty batch')
    batch_size = len(samples)
    max_sentence = max((sample['sentence_length'] for sample in samples))
    max_component = max((max(sample['component_lengths']) for sample in samples))
    ids_sequences = torch.zeros((batch_size, max_sentence, max_component), dtype=torch.long)
    component_lengths = torch.zeros((batch_size, max_sentence), dtype=torch.long)
    component_labels = torch.full((batch_size, max_sentence, 3), -100, dtype=torch.long)
    syllable_ids = torch.full((batch_size, max_sentence), -100, dtype=torch.long)
    acoustic_ids = torch.full((batch_size, max_sentence), -1, dtype=torch.long)
    sentence_lengths = torch.empty(batch_size, dtype=torch.long)
    for row_index, sample in enumerate(samples):
        n = sample['sentence_length']
        sentence_lengths[row_index] = n
        for position, (sequence, length) in enumerate(zip(sample['ids_sequences'], sample['component_lengths'])):
            if length != len(sequence) or length <= 0:
                raise ValueError('sample contains invalid IDS component length')
            ids_sequences[row_index, position, :length] = torch.as_tensor(sequence, dtype=torch.long)
            component_lengths[row_index, position] = length
        component_labels[row_index, :n] = torch.as_tensor(sample['component_labels'], dtype=torch.long)
        syllable_ids[row_index, :n] = torch.as_tensor(sample['syllable_ids'], dtype=torch.long)
        acoustic_ids[row_index, :n] = torch.as_tensor(sample['acoustic_ids'], dtype=torch.long)
    char_mask = component_lengths > 0
    return {'ids_sequences': ids_sequences, 'component_lengths': component_lengths, 'sentence_lengths': sentence_lengths, 'char_mask': char_mask, 'component_labels': component_labels, 'syllable_ids': syllable_ids, 'acoustic_ids': acoustic_ids, 'sample_ids': [sample['sample_id'] for sample in samples], 'texts': [sample['text'] for sample in samples], 'positions': [sample['positions'] for sample in samples]}

def create_loaders(experiment: dict, seed: int, batch_size: int | None=None, num_workers: int=0):
    manifest = experiment.get('manifest', {})
    if manifest.get('bundle_type') != 'experiment_dataset' or manifest.get('schema_version') != 3:
        raise ValueError('training loaders require a validated schema-3 experiment dataset')
    if type(seed) is not int or seed < 0:
        raise ValueError('loader seed must be a nonnegative integer')
    batch = batch_size or experiment['config'].get('training', {}).get('batch_size', 32)
    if type(batch) is not int or batch <= 0 or type(num_workers) is not int or (num_workers < 0):
        raise ValueError('batch_size must be positive and num_workers nonnegative')
    result = {}
    for stream, split in enumerate(('train', 'validation', 'test'), start=1):
        dataset = ExperimentDataset(experiment['splits'][split], experiment['samples'])
        result[split] = DataLoader(dataset, batch_size=batch, shuffle=split == 'train', generator=make_generator(seed, stream), num_workers=num_workers, collate_fn=collate_experiment)
    return result
```

### Contextual encoder


```python
"""Shared IDS and sentence encoders with condition-specific readout heads."""
import torch
from torch import nn
from torch.nn.utils.rnn import pack_padded_sequence, pad_packed_sequence

class IDSCharacterEncoder(nn.Module):

    def __init__(self, vocab_size: int):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, 64, padding_idx=0)
        self.bigru = nn.GRU(64, 32, num_layers=1, bidirectional=True, batch_first=True)

    def forward(self, token_ids: torch.Tensor, lengths: torch.Tensor) -> torch.Tensor:
        valid = lengths.reshape(-1) > 0
        if not bool(valid.any()):
            return self.embedding.weight.new_zeros((lengths.numel(), 64))
        selected_ids = token_ids.reshape(-1, token_ids.shape[-1])[valid]
        selected_lengths = lengths.reshape(-1)[valid].to(device='cpu', dtype=torch.long)
        embedded = self.embedding(selected_ids)
        packed = pack_padded_sequence(embedded, selected_lengths, batch_first=True, enforce_sorted=False)
        _, hidden = self.bigru(packed)
        character_vectors = torch.cat((hidden[-2], hidden[-1]), dim=-1)
        result = self.embedding.weight.new_zeros((lengths.numel(), 64))
        return result.index_copy(0, valid.nonzero(as_tuple=False).flatten(), character_vectors).reshape(*lengths.shape, 64)

class SharedEncoder(nn.Module):
    """Encode variable-length IDS for each character, then full sentence context."""
    context_dim = 128

    def __init__(self, ids_vocab_size: int):
        super().__init__()
        self.ids = IDSCharacterEncoder(ids_vocab_size)
        self.sentence = nn.LSTM(64, 64, num_layers=1, bidirectional=True, batch_first=True)

    def encode(self, batch: dict) -> tuple[torch.Tensor, torch.Tensor]:
        token_ids = batch['ids_sequences']
        component_lengths = batch['component_lengths']
        sentence_lengths = batch['sentence_lengths'].to(dtype=torch.long)
        mask = batch['char_mask'].to(dtype=torch.bool)
        if token_ids.ndim != 3 or component_lengths.shape != token_ids.shape[:2] or mask.shape != token_ids.shape[:2]:
            raise ValueError('invalid padded IDS batch shapes')
        if not torch.equal(mask, component_lengths > 0):
            raise ValueError('character mask and IDS lengths disagree')
        if not bool((sentence_lengths > 0).all()) or not torch.equal(sentence_lengths.cpu(), mask.sum(1).cpu()):
            raise ValueError('sentence lengths disagree with character mask')
        char_vectors = self.ids(token_ids, component_lengths)
        packed = pack_padded_sequence(char_vectors, sentence_lengths.cpu(), batch_first=True, enforce_sorted=False)
        packed_output, _ = self.sentence(packed)
        contextual, _ = pad_packed_sequence(packed_output, batch_first=True, total_length=token_ids.shape[1])
        return (contextual, mask)

    def forward(self, batch: dict) -> tuple[torch.Tensor, torch.Tensor]:
        return self.encode(batch)

class ClassifierHead(nn.Module):

    def __init__(self, output_dim: int):
        super().__init__()
        self.layers = nn.Sequential(nn.Linear(128, 64), nn.ReLU(), nn.Dropout(0.5), nn.Linear(64, output_dim))

    def forward(self, hidden: torch.Tensor) -> torch.Tensor:
        return self.layers(hidden)

class TrainingModel(nn.Module):

    def __init__(self, encoder: SharedEncoder, model_id: str, vocab: dict, acoustic_dim: int):
        super().__init__()
        model_id = canonical_model_id(model_id)
        family, weight = parse_model_id(model_id)
        self.encoder = encoder
        self.model_id = model_id
        self.lambda_cosine = weight
        self.context_dim = 128
        labels = vocab['labels']
        self.component_heads = nn.ModuleDict()
        if family in {'components', 'components-acoustic'}:
            for name in ('initials', 'finals', 'tones'):
                self.component_heads[name] = ClassifierHead(len(labels[name]))
        self.syllable_head = ClassifierHead(len(vocab['syllables'])) if family == 'syllable' else None
        self.acoustic_head = nn.Linear(128, int(acoustic_dim)) if family in {'acoustic', 'components-acoustic'} else None

    def encode(self, batch: dict) -> tuple[torch.Tensor, torch.Tensor]:
        return self.encoder.encode(batch)

    def forward(self, batch: dict) -> dict:
        hidden, mask = self.encoder.encode(batch)
        result = {'hidden': hidden, 'mask': mask}
        if self.component_heads:
            result['component_logits'] = {name: head(hidden) for name, head in self.component_heads.items()}
        if self.syllable_head is not None:
            result['syllable_logits'] = self.syllable_head(hidden)
        if self.acoustic_head is not None:
            result['acoustic_prediction'] = self.acoustic_head(hidden)
        return result

def model_construct_with_streams(config: dict, vocab: dict, model_id: str, seed: int) -> TrainingModel:
    """Initialize shared weights and heads from independent seed streams."""
    model_id = canonical_model_id(model_id)
    devices = []
    with torch.random.fork_rng(devices=devices):
        torch.manual_seed(int(seed))
        encoder = SharedEncoder(len(vocab['ids']['tokens']))
    with torch.random.fork_rng(devices=devices):
        torch.manual_seed(int(seed) + 1000003)
        model = TrainingModel(encoder, model_id, vocab, int(config['architecture']['acoustic_dim']))
    return model

def build_encoder(config: dict, vocab: dict, model_id: str | None=None, seed: int=0) -> TrainingModel:
    """Build a full model; the returned `.encode(batch)` is the shared export API."""
    architecture = config.get('architecture', {})
    if architecture.get('embedding_dim') != 64 or architecture.get('context_dim') != 128 or architecture.get('acoustic_dim', 0) <= 0:
        raise ValueError('experiment architecture violates encoder dimension contract')
    selected = model_id or config.get('model_id')
    if selected is None:
        raise ValueError('model_id is required to construct objective heads')
    return model_construct_with_streams(config, vocab, selected, seed)

def shared_encoder_sha256(model: TrainingModel) -> str:
    import hashlib
    digest = hashlib.sha256()
    for name, tensor in sorted(model.encoder.state_dict().items()):
        value = tensor.detach().cpu().contiguous()
        digest.update(name.encode('utf-8'))
        digest.update(str(value.dtype).encode('ascii'))
        digest.update(repr(tuple(value.shape)).encode('ascii'))
        digest.update(value.numpy().tobytes())
    return digest.hexdigest()
```

### Saved sample identities


```python
"""Deterministic sampling of unique unordered pairs without quadratic materialization."""
import random
from collections.abc import Iterator, Sequence

def pair_count(item_count: int) -> int:
    """Return the number of unique unordered pairs of ``item_count`` objects."""
    if item_count < 0:
        raise ValueError('item_count must be non-negative')
    return item_count * (item_count - 1) // 2

def pair_from_rank(item_count: int, rank: int) -> tuple[int, int]:
    """Map a rank in lexicographic upper-triangle order to ``(left, right)``."""
    total = pair_count(item_count)
    if rank < 0 or rank >= total:
        raise IndexError(f'pair rank {rank} outside [0, {total})')
    low, high = (0, item_count - 1)
    while low + 1 < high:
        middle = (low + high) // 2
        before = middle * (2 * item_count - middle - 1) // 2
        if before <= rank:
            low = middle
        else:
            high = middle
    left = low
    before = left * (2 * item_count - left - 1) // 2
    right = left + 1 + (rank - before)
    return (left, right)

def iter_all_pairs(item_count: int, *, max_pairs: int) -> Iterator[tuple[int, int]]:
    """Yield all unique pairs, rejecting an unexpectedly large expansion."""
    total = pair_count(item_count)
    if max_pairs < 0:
        raise ValueError('max_pairs must be non-negative')
    if total > max_pairs:
        raise ValueError(f'all-pairs plan has {total} pairs, above cap {max_pairs}')
    for left in range(item_count):
        for right in range(left + 1, item_count):
            yield (left, right)

def sample_pair_indices(item_count: int, sample_count: int, seed: int) -> list[tuple[int, int]]:
    """Uniformly sample unique unordered pairs without constructing all pairs."""
    total = pair_count(item_count)
    if sample_count < 0:
        raise ValueError('sample_count must be non-negative')
    if sample_count > total:
        raise ValueError(f'requested {sample_count} pairs from population of {total}')
    ranks = random.Random(seed).sample(range(total), sample_count)
    return [pair_from_rank(item_count, rank) for rank in sorted(ranks)]

def make_pair_plan(item_ids: Sequence[str], *, sample_count: int | None, seed: int, max_all_pairs: int) -> list[tuple[str, str]]:
    """Return canonical ID pairs from an already stably ordered collection."""
    if len(item_ids) != len(set(item_ids)):
        raise ValueError('item_ids must be unique')
    if list(item_ids) != sorted(item_ids):
        raise ValueError('item_ids must be stably sorted')
    if sample_count is None:
        indices = iter_all_pairs(len(item_ids), max_pairs=max_all_pairs)
    else:
        indices = sample_pair_indices(len(item_ids), sample_count, seed)
    return [(item_ids[left], item_ids[right]) for left, right in indices]
```

### Anchor transitions and coverage


```python
"""Versioned hierarchical anchor-Q triplet sampling."""
import hashlib
import json
import numpy as np
anchor_sampling_SAMPLER_VERSION = 1
anchor_sampling_TRIPLET_SCHEMA_VERSION = 1
SEED = 20261026
EDGES = (('initial', '000', '100'), ('initial', '001', '101'), ('initial', '010', '110'), ('initial', '011', '111'), ('final', '000', '010'), ('final', '001', '011'), ('final', '100', '110'), ('final', '101', '111'), ('tone', '000', '001'), ('tone', '010', '011'), ('tone', '100', '101'), ('tone', '110', '111'))

def anchor_sampling_stable_id(prefix: str, values: list[str]) -> str:
    payload = json.dumps(values, ensure_ascii=False, separators=(',', ':')).encode('utf-8')
    return prefix + hashlib.sha256(payload).hexdigest()

def sample_anchor_q(occurrences: list[dict], syllables: list[dict], *, draws_per_edge: int=1000, seed: int=SEED) -> tuple[list[dict], dict]:
    """Generate fixed draws; diagnostics are returned even for coverage failures."""
    if isinstance(draws_per_edge, bool) or not isinstance(draws_per_edge, int) or draws_per_edge <= 0:
        raise ValueError('draws_per_edge must be a positive integer')
    if isinstance(seed, bool) or not isinstance(seed, int) or seed < 0:
        raise ValueError('anchor-Q seed must be a non-negative integer')
    by_syllable = {row['syllable_id']: row for row in syllables}
    occurrence_map = {row['occurrence_id']: row for row in occurrences}
    members = {key: sorted(row['occurrence_ids']) for key, row in by_syllable.items()}

    def overlap(anchor_id: str, candidate_id: str) -> str:
        return ''.join((str(int(by_syllable[anchor_id][key] == by_syllable[candidate_id][key])) for key in ('initial', 'final', 'tone')))
    diagnostics = {'sampler_version': anchor_sampling_SAMPLER_VERSION, 'seed': seed, 'rng': 'numpy.random.PCG64', 'draws_per_edge': draws_per_edge, 'edges': {}}
    supports = {}
    candidate_supports = {}
    for component, low_mask, high_mask in EDGES:
        edge_id = f'{component}:{low_mask}->{high_mask}'
        eligible, excluded = ([], {})
        for sid in sorted(by_syllable):
            ids = members[sid]
            if not ids:
                excluded[sid] = 'no_occurrences'
                continue
            if high_mask == '111' and len(ids) < 2:
                excluded[sid] = 'high_111_requires_two_occurrences'
                continue
            high = [x for x in sorted(by_syllable) if overlap(sid, x) == high_mask and (x != sid or high_mask == '111')]
            low = [x for x in sorted(by_syllable) if overlap(sid, x) == low_mask and x != sid]
            if not high or not low:
                excluded[sid] = 'missing_high_or_low_syllable'
                continue
            eligible.append(sid)
            candidate_supports[component, low_mask, high_mask, sid] = (high, low)
        supports[edge_id] = (eligible, excluded)
        diagnostics['edges'][edge_id] = {'eligible_anchor_syllables': len(eligible), 'excluded_anchor_syllables': len(excluded), 'exclusion_reasons': excluded}
    missing = [edge for edge, (eligible, _) in supports.items() if not eligible]
    if missing:
        diagnostics['missing_edges'] = missing
        raise ValueError('anchor-Q coverage failure; diagnostics=' + json.dumps(diagnostics, ensure_ascii=False, sort_keys=True))
    rng = np.random.Generator(np.random.PCG64(seed))
    output = []
    for component, low_mask, high_mask in EDGES:
        edge_id = f'{component}:{low_mask}->{high_mask}'
        anchors = supports[edge_id][0]
        for draw_index in range(draws_per_edge):
            anchor_sid = anchors[int(rng.integers(len(anchors)))]
            anchor_occ = members[anchor_sid][int(rng.integers(len(members[anchor_sid])))]
            eligible_high, eligible_low = candidate_supports[component, low_mask, high_mask, anchor_sid]
            high_sid = eligible_high[int(rng.integers(len(eligible_high)))]
            low_sid = eligible_low[int(rng.integers(len(eligible_low)))]
            high_occ_choices = [x for x in members[high_sid] if x != anchor_occ]
            high_occ = high_occ_choices[int(rng.integers(len(high_occ_choices)))]
            low_occ_choices = members[low_sid]
            low_occ = low_occ_choices[int(rng.integers(len(low_occ_choices)))]
            if len({anchor_occ, high_occ, low_occ}) != 3:
                raise ValueError(f'anchor-Q sampler produced repeated occurrence IDs for {edge_id}')
            triple = [anchor_occ, high_occ, low_occ]
            triplet_id = anchor_sampling_stable_id('triplet:', [edge_id, *triple])
            output.append({'schema_version': anchor_sampling_TRIPLET_SCHEMA_VERSION, 'sampler_version': anchor_sampling_SAMPLER_VERSION, 'edge_id': edge_id, 'component': component, 'low_mask': low_mask, 'high_mask': high_mask, 'draw_index': draw_index, 'draw_id': f'anchor-q:v1:{edge_id}:{draw_index:06d}', 'triplet_id': triplet_id, 'anchor_syllable_id': anchor_sid, 'high_syllable_id': high_sid, 'low_syllable_id': low_sid, 'anchor_id': anchor_occ, 'high_id': high_occ, 'low_id': low_occ})
    diagnostics['draw_count'] = len(output)
    diagnostics['unique_triplets_by_edge'] = {edge: len({r['triplet_id'] for r in output if r['edge_id'] == edge}) for edge in supports}
    diagnostics['unique_anchor_occurrences_by_edge'] = {edge: len({r['anchor_id'] for r in output if r['edge_id'] == edge}) for edge in supports}
    diagnostics['unique_anchor_syllables_by_edge'] = {edge: len({r['anchor_syllable_id'] for r in output if r['edge_id'] == edge}) for edge in supports}
    diagnostics['unique_occurrences_by_edge'] = {edge: len({r[k] for r in output if r['edge_id'] == edge for k in ('anchor_id', 'high_id', 'low_id')}) for edge in supports}
    for edge in supports:
        diagnostics['edges'][edge].update({'draw_count': sum((row['edge_id'] == edge for row in output)), 'unique_triplets': diagnostics['unique_triplets_by_edge'][edge], 'unique_anchor_occurrences': diagnostics['unique_anchor_occurrences_by_edge'][edge], 'unique_anchor_syllables': diagnostics['unique_anchor_syllables_by_edge'][edge], 'unique_occurrences': diagnostics['unique_occurrences_by_edge'][edge]})
    return (output, diagnostics)
```

### DIMSIM pair identities


```python
"""Occurrence-only DIMSIM sampling plans for schema-3 bundles."""
import hashlib
import json
import numpy as np
dimsim_sampling_SAMPLER_VERSION = 1
dimsim_sampling_PAIR_SCHEMA_VERSION = 1

def pair_identity(left: str, right: str) -> str:
    endpoints = sorted((left, right))
    raw = json.dumps(endpoints, ensure_ascii=False, separators=(',', ':')).encode('utf-8')
    return 'pair:' + hashlib.sha256(raw).hexdigest()

def sample_natural(occurrences: list[dict], count: int=12000, seed: int=42) -> list[tuple[str, str, str, int]]:
    ids = sorted((row['occurrence_id'] for row in occurrences if row['initial']))
    population = len(ids) * (len(ids) - 1) // 2
    if isinstance(count, bool) or not isinstance(count, int) or count < 0 or (count > population):
        raise ValueError(f'natural DIMSIM sample_count {count} exceeds eligible unique pair population {population}')
    selected = sample_pair_indices(len(ids), count, seed)
    return [(ids[i], ids[j], pair_identity(ids[i], ids[j]), index) for index, (i, j) in enumerate(selected)]

def sample_balanced(occurrences: list[dict], syllables: list[dict], count: int=12000, seed: int=20261026) -> list[tuple[str, str, str, int, str, str]]:
    eligible = {row['syllable_id']: sorted(row['occurrence_ids']) for row in syllables if row['initial']}
    ids = sorted(eligible)
    if not ids:
        raise ValueError('balanced DIMSIM has no nonzero-initial syllables')
    rng = np.random.Generator(np.random.PCG64(seed))
    result = []
    for index in range(count):
        first_sid = ids[int(rng.integers(len(ids)))]
        seconds = [sid for sid in ids if sid != first_sid or len(eligible[sid]) >= 2]
        if not seconds:
            raise ValueError('balanced DIMSIM coverage failure: no eligible second syllable')
        second_sid = seconds[int(rng.integers(len(seconds)))]
        first_id = eligible[first_sid][int(rng.integers(len(eligible[first_sid])))]
        second_choices = [x for x in eligible[second_sid] if x != first_id]
        second_id = second_choices[int(rng.integers(len(second_choices)))]
        result.append((first_id, second_id, pair_identity(first_id, second_id), index, first_sid, second_sid))
    return result
```

### Immutable evaluation bundle


```python
"""Schema-3 immutable contextual-only evaluation bundle implementation."""
import hashlib
import importlib.metadata
import json
import math
import shutil
from pathlib import Path
from typing import Any
contextual_bundle_SCHEMA_VERSION = 3
Q_SEED = 20261026
anchor_sampling_SAMPLER_VERSION = 1
anchor_sampling_TRIPLET_SCHEMA_VERSION = 1
dimsim_sampling_SAMPLER_VERSION = 1
dimsim_sampling_PAIR_SCHEMA_VERSION = 1
FILES = {'config.json', 'manifest.json', 'coverage.json', 'sentences.jsonl', 'occurrences.jsonl', 'syllables.jsonl', 'pairs/dimsim-natural.jsonl', 'pairs/dimsim-balanced.jsonl', 'triplets/anchor-q.jsonl'}
SOURCE_HASH_NAMES = {'manifest.json', 'config.json', 'vocab.json', 'samples.jsonl', 'splits.json'}

def contextual_bundle_sha(path: Path) -> str:
    return hashlib.sha256(path.read_bytes()).hexdigest()

def contextual_bundle_read(path: Path) -> Any:
    return json.loads(path.read_text(encoding='utf-8'))

def contextual_bundle_readl(path: Path) -> list[dict]:
    return [json.loads(line) for line in path.read_text(encoding='utf-8').splitlines() if line]

def contextual_bundle_stable(prefix: str, values: list[str]) -> str:
    return prefix + hashlib.sha256(json.dumps(values, ensure_ascii=False, separators=(',', ':')).encode('utf-8')).hexdigest()

def contextual_bundle_syllables(occurrences: list[dict]) -> list[dict]:
    members: dict[str, list[str]] = {}
    metadata = {}
    for row in occurrences:
        sid = contextual_bundle_stable('syllable:', [row['pinyin']])
        if row.get('syllable_id') != sid:
            raise ValueError(f"occurrence syllable_id disagrees with canonical Pinyin: {row.get('occurrence_id')}")
        metadata[sid] = {'syllable_id': sid, 'pinyin': row['pinyin'], 'initial': row['initial'], 'final': row['final'], 'tone': row['tone']}
        members.setdefault(sid, []).append(row['occurrence_id'])
    return [{**metadata[sid], 'occurrence_ids': sorted(members[sid]), 'occurrence_count': len(members[sid])} for sid in sorted(metadata)]

def contextual_bundle_anchor_q_support(syllables: list[dict]) -> dict[str, tuple[list[str], dict[str, str]]]:
    """Independently derive eligible anchors and exclusions from gold metadata."""
    by_id = {row['syllable_id']: row for row in syllables}
    ordered = sorted(by_id)

    def overlap(anchor_id: str, candidate_id: str) -> str:
        a, b = (by_id[anchor_id], by_id[candidate_id])
        return ''.join((str(int(a[key] == b[key])) for key in ('initial', 'final', 'tone')))
    support = {}
    for component, low_mask, high_mask in EDGES:
        eligible, excluded = ([], {})
        for anchor_id in ordered:
            members = by_id[anchor_id]['occurrence_ids']
            if high_mask == '111' and len(members) < 2:
                excluded[anchor_id] = 'high_111_requires_two_occurrences'
                continue
            high = [candidate for candidate in ordered if overlap(anchor_id, candidate) == high_mask and (candidate != anchor_id or high_mask == '111')]
            low = [candidate for candidate in ordered if overlap(anchor_id, candidate) == low_mask and candidate != anchor_id]
            if high and low:
                eligible.append(anchor_id)
            else:
                excluded[anchor_id] = 'missing_high_or_low_syllable'
        support[f'{component}:{low_mask}->{high_mask}'] = (eligible, excluded)
    return support

def contextual_bundle_anchor_q_coverage(syllables: list[dict], triplets: list[dict], *, seed: int, draws_per_edge: int) -> dict:
    support = contextual_bundle_anchor_q_support(syllables)
    by_edge = {edge: [row for row in triplets if row['edge_id'] == edge] for edge in support}
    result = {'sampler_version': anchor_sampling_SAMPLER_VERSION, 'seed': seed, 'rng': 'numpy.random.PCG64', 'draws_per_edge': draws_per_edge, 'edges': {}}
    unique_triples, unique_anchor_occurrences, unique_anchor_syllables, unique_occurrences = ({}, {}, {}, {})
    for edge, (eligible, excluded) in support.items():
        rows = by_edge[edge]
        unique_triples[edge] = len({row['triplet_id'] for row in rows})
        unique_anchor_occurrences[edge] = len({row['anchor_id'] for row in rows})
        unique_anchor_syllables[edge] = len({row['anchor_syllable_id'] for row in rows})
        unique_occurrences[edge] = len({row[field] for row in rows for field in ('anchor_id', 'high_id', 'low_id')})
        result['edges'][edge] = {'eligible_anchor_syllables': len(eligible), 'excluded_anchor_syllables': len(excluded), 'exclusion_reasons': excluded, 'draw_count': len(rows), 'unique_triplets': unique_triples[edge], 'unique_anchor_occurrences': unique_anchor_occurrences[edge], 'unique_anchor_syllables': unique_anchor_syllables[edge], 'unique_occurrences': unique_occurrences[edge]}
    result.update({'draw_count': len(triplets), 'unique_triplets_by_edge': unique_triples, 'unique_anchor_occurrences_by_edge': unique_anchor_occurrences, 'unique_anchor_syllables_by_edge': unique_anchor_syllables, 'unique_occurrences_by_edge': unique_occurrences})
    return result

def contextual_bundle_validate_provenance(code_provenance):
    if not isinstance(code_provenance, dict) or not code_provenance or any((not isinstance(k, str) or not isinstance(v, str) or len(v) != 64 or any((c not in '0123456789abcdef' for c in v)) for k, v in code_provenance.items())):
        raise ValueError('explicit preparation code_provenance hashes are required')

def contextual_bundle_validate_source_provenance(source_experiment):
    if not isinstance(source_experiment, dict) or set(source_experiment) != {'path', 'experiment_id', 'hashes'}:
        raise ValueError('source_experiment provenance must contain path, experiment_id, and hashes')
    for field in ('path', 'experiment_id'):
        value = source_experiment.get(field)
        if not isinstance(value, str) or not value.strip():
            raise ValueError(f'source_experiment {field} must be a non-empty string')
    hashes = source_experiment.get('hashes')
    if not isinstance(hashes, dict) or set(hashes) != SOURCE_HASH_NAMES or any((not isinstance(value, str) or len(value) != 64 or any((ch not in '0123456789abcdef' for ch in value)) for value in hashes.values())):
        raise ValueError('source_experiment hashes must contain exactly five valid SHA-256 values')

def contextual_bundle_typed_equal(actual, expected) -> bool:
    """Compare JSON contract values without Python bool/int equality aliases."""
    if type(actual) is not type(expected):
        return False
    if isinstance(expected, dict):
        return set(actual) == set(expected) and all((contextual_bundle_typed_equal(actual[key], value) for key, value in expected.items()))
    if isinstance(expected, list):
        return len(actual) == len(expected) and all((contextual_bundle_typed_equal(left, right) for left, right in zip(actual, expected)))
    return actual == expected

def load_contextual_bundle(root: Path, manifest: dict) -> dict[str, Any]:
    if type(manifest.get('schema_version')) is not int or manifest.get('schema_version') != contextual_bundle_SCHEMA_VERSION:
        raise ValueError('unsupported evaluation bundle schema')
    if manifest.get('pinyin_contract') != PINYIN_CONTRACT_VERSION or manifest.get('parser_provenance') != PARSER_PROVENANCE:
        raise ValueError('unsupported evaluation Pinyin contract')
    contextual_bundle_validate_provenance(manifest.get('code_provenance'))
    contextual_bundle_validate_source_provenance(manifest.get('source_experiment'))
    config = contextual_bundle_read(root / 'config.json')
    if config.get('schema_version') != contextual_bundle_SCHEMA_VERSION:
        raise ValueError('unsupported evaluation config schema')
    if contextual_bundle_validate_id(config.get('evaluation_id')) != contextual_bundle_validate_id(manifest.get('evaluation_id')):
        raise ValueError('schema-3 evaluation IDs disagree')
    contextual_bundle_validate_config(config, manifest.get('metric'))
    artifacts = manifest.get('artifacts')
    if not isinstance(artifacts, dict) or set(artifacts) != FILES - {'manifest.json'}:
        raise ValueError('schema-3 manifest artifact inventory is invalid')
    actual = {p.relative_to(root).as_posix() for p in root.rglob('*') if p.is_file() and p.name != 'manifest.json'}
    if actual != FILES - {'manifest.json'}:
        raise ValueError('schema-3 bundle file inventory is invalid')
    for rel, expected in artifacts.items():
        path = (root / rel).resolve()
        if root not in path.parents or not path.is_file() or contextual_bundle_sha(path) != expected:
            raise ValueError(f'evaluation artifact checksum mismatch: {rel}')
    sentences, occurrences, syllables = (contextual_bundle_readl(root / name) for name in ('sentences.jsonl', 'occurrences.jsonl', 'syllables.jsonl'))
    sentence_map = {r.get('sentence_id'): r for r in sentences}
    occurrence_map = {r.get('occurrence_id'): r for r in occurrences}
    syllable_map = {r.get('syllable_id'): r for r in syllables}
    if len(sentence_map) != len(sentences) or len(occurrence_map) != len(occurrences) or len(syllable_map) != len(syllables):
        raise ValueError('duplicate or invalid sentence, occurrence, or syllable IDs')
    positions_by_sentence: dict[str, list[int]] = {sid: [] for sid in sentence_map}
    expected_syllables = contextual_bundle_syllables(occurrences)
    if expected_syllables != syllables:
        raise ValueError('syllable metadata or membership disagrees with occurrences')
    for row in occurrences:
        sid = row.get('sample_id')
        pos = row.get('character_position')
        if sid not in sentence_map or isinstance(pos, bool) or (not isinstance(pos, int)) or (pos < 0):
            raise ValueError('occurrence sentence alignment is invalid')
        if row.get('occurrence_id') != f'{sid}:{pos}':
            raise ValueError('occurrence ID does not match sentence position')
        text = sentence_map[sid]['text']
        if pos >= len(text) or text[pos] != row.get('character'):
            raise ValueError(f"occurrence character alignment mismatch: {row.get('occurrence_id')}")
        parsed = parse_components(validate_canonical_label(row.get('pinyin')))
        if tuple((row.get(k) for k in ('initial', 'final', 'tone'))) != (parsed.initial, parsed.final, parsed.tone):
            raise ValueError(f"occurrence components disagree with Pinyin: {row.get('occurrence_id')}")
        positions_by_sentence[sid].append(pos)
    if any((sorted(pos) != list(range(len(sentence_map[sid]['text']))) for sid, pos in positions_by_sentence.items())):
        raise ValueError('sentence occurrence positions do not fully cover sentence text')
    occurrence_syllable = {oid: row['syllable_id'] for row in syllables for oid in row['occurrence_ids']}
    anchor_support = contextual_bundle_anchor_q_support(syllables)
    pairs = {'dimsim-natural': contextual_bundle_readl(root / 'pairs/dimsim-natural.jsonl'), 'dimsim-balanced': contextual_bundle_readl(root / 'pairs/dimsim-balanced.jsonl')}
    for mode, rows in pairs.items():
        if any((r.get('mode') != ('natural' if mode.endswith('natural') else 'balanced') for r in rows)):
            raise ValueError(f'{mode} mode labels are invalid')
        draw_ids = [r.get('draw_id') for r in rows]
        config_mode = 'dimsim_' + mode.removeprefix('dimsim-')
        if len(draw_ids) != len(set(draw_ids)) or len(rows) != config['sampling'][config_mode]['count']:
            raise ValueError(f'{mode} draw count or IDs are invalid')
        if any((type(r.get('draw_index')) is not int or r.get('draw_index') != i for i, r in enumerate(rows))):
            raise ValueError(f'{mode} draw order is invalid')
        if mode.endswith('natural') and len({r.get('pair_id') for r in rows}) != len(rows):
            raise ValueError('natural DIMSIM pairs must be unique')
        for row in rows:
            if type(row.get('schema_version')) is not int or row.get('schema_version') != dimsim_sampling_PAIR_SCHEMA_VERSION or type(row.get('sampler_version')) is not int or (row.get('sampler_version') != dimsim_sampling_SAMPLER_VERSION):
                raise ValueError(f'{mode} unsupported row or sampler version')
            a, b = (row.get('left_id'), row.get('right_id'))
            if a not in occurrence_map or b not in occurrence_map or a == b or (a >= b):
                raise ValueError(f'{mode} pair has invalid endpoints')
            left, right = (occurrence_map[a], occurrence_map[b])
            if not left['initial'] or not right['initial']:
                raise ValueError(f'{mode} violates zero-initial exclusion')
            if row.get('pair_id') != pair_identity(a, b):
                raise ValueError(f'{mode} pair identity mismatch')
            if row.get('left_pinyin') != left['pinyin'] or row.get('right_pinyin') != right['pinyin']:
                raise ValueError(f'{mode} Pinyin labels disagree with endpoint metadata')
            expected_draw = f"dimsim-{('natural' if mode.endswith('natural') else 'balanced')}:v1:{row['draw_index']:06d}"
            if row.get('draw_id') != expected_draw:
                raise ValueError(f'{mode} draw ID mismatch')
            mask = ''.join((str(int(left[k] == right[k])) for k in ('initial', 'final', 'tone')))
            if row.get('overlap_mask') != mask:
                raise ValueError(f'{mode} overlap mask mismatch')
            if any((row.get('equal_' + key) is not (mask[i] == '1') for i, key in enumerate(('initial', 'final', 'tone')))):
                raise ValueError(f'{mode} equality flags mismatch')
            if row.get('left_zero_initial') is not False or row.get('right_zero_initial') is not False or row.get('no_zero_initial_diagnostic') is not True:
                raise ValueError(f'{mode} zero-initial flags mismatch')
            if mode.endswith('balanced'):
                first_id, second_id = (row.get('first_id'), row.get('second_id'))
                if {first_id, second_id} != {a, b} or first_id == second_id:
                    raise ValueError('balanced DIMSIM orientation does not match pair endpoints')
                if row.get('first_syllable_id') != occurrence_syllable.get(first_id) or row.get('second_syllable_id') != occurrence_syllable.get(second_id):
                    raise ValueError('balanced DIMSIM syllable orientation is invalid')
            value, status = (row.get('dimsim_distance'), row.get('distance_status'))
            if status == 'ok':
                if isinstance(value, bool) or not isinstance(value, (int, float)) or (not math.isfinite(value)) or (value < 0):
                    raise ValueError(f'{mode} ok distance must be finite and non-negative')
            elif value is not None or not isinstance(status, str) or (not (status == 'not_computed' or status.startswith('invalid:'))):
                raise ValueError(f'{mode} distance status/value mismatch')
    triplets = contextual_bundle_readl(root / 'triplets/anchor-q.jsonl')
    sampling = config.get('sampling', {}).get('anchor_q', {})
    per_edge = sampling.get('draws_per_edge')
    expected_edges = [f'{c}:{l}->{h}' for c, l, h in EDGES]
    if sampling.get('edge_order') != expected_edges or len(triplets) != len(EDGES) * per_edge:
        raise ValueError('anchor-Q edge order or draw count is invalid')
    edge_counts = {edge: 0 for edge in expected_edges}
    draw_ids = set()
    for global_index, row in enumerate(triplets):
        edge = row.get('edge_id')
        expected_global_edge = expected_edges[global_index // per_edge] if per_edge else None
        if edge != expected_global_edge:
            raise ValueError('anchor-Q rows do not follow configured global edge order')
        if edge not in edge_counts or type(row.get('schema_version')) is not int or row.get('schema_version') != anchor_sampling_TRIPLET_SCHEMA_VERSION or (type(row.get('sampler_version')) is not int) or (row.get('sampler_version') != anchor_sampling_SAMPLER_VERSION):
            raise ValueError('anchor-Q row schema or edge is invalid')
        ids = [row.get(k) for k in ('anchor_id', 'high_id', 'low_id')]
        if any((x not in occurrence_map for x in ids)) or len(set(ids)) != 3:
            raise ValueError('anchor-Q occurrence references must be known and distinct')
        comp, low, high = (edge.split(':')[0], edge.split(':')[1].split('->')[0], edge.split('->')[1])
        a, h, l = (occurrence_map[x] for x in ids)
        if occurrence_syllable[ids[0]] not in anchor_support[edge][0]:
            raise ValueError('anchor-Q draw uses an ineligible anchor syllable')

        def mask(x):
            return ''.join((str(int(a[k] == x[k])) for k in ('initial', 'final', 'tone')))
        if mask(h) != high or mask(l) != low:
            raise ValueError('anchor-Q gold masks disagree with edge')
        expected_triplet = contextual_bundle_stable('triplet:', [edge, *ids])
        if row.get('triplet_id') != expected_triplet or row.get('component') != comp or row.get('low_mask') != low or (row.get('high_mask') != high):
            raise ValueError('anchor-Q triplet identity or edge metadata mismatch')
        expected_draw_id = f"anchor-q:v1:{edge}:{row.get('draw_index', -1):06d}"
        if row.get('draw_id') != expected_draw_id:
            raise ValueError('anchor-Q draw identity mismatch')
        if high == '111' and (row.get('anchor_syllable_id') != row.get('high_syllable_id') or row.get('anchor_id') == row.get('high_id')):
            raise ValueError('anchor-Q high-111 identity rule violated')
        for name, oid in (('anchor', ids[0]), ('high', ids[1]), ('low', ids[2])):
            if row.get(name + '_syllable_id') != occurrence_syllable.get(oid):
                raise ValueError('anchor-Q syllable reference mismatch')
        if type(row.get('draw_index')) is not int or row.get('draw_index') != edge_counts[edge]:
            raise ValueError('anchor-Q draw index/order mismatch')
        if row.get('draw_id') in draw_ids:
            raise ValueError('duplicate anchor-Q draw ID')
        draw_ids.add(row['draw_id'])
        edge_counts[edge] += 1
    if any((count != per_edge for count in edge_counts.values())):
        raise ValueError('anchor-Q edge coverage counts are invalid')
    coverage = contextual_bundle_read(root / 'coverage.json')
    expected_q_coverage = contextual_bundle_anchor_q_coverage(syllables, triplets, seed=sampling['seed'], draws_per_edge=per_edge)
    counts = manifest.get('counts', {})
    expected_counts = {'sentences': len(sentences), 'occurrences': len(occurrences), 'syllables': len(syllables), 'dimsim_natural': len(pairs['dimsim-natural']), 'dimsim_balanced': len(pairs['dimsim-balanced']), 'anchor_q': len(triplets)}
    expected_edge_counts = {edge: per_edge for edge in expected_edges}
    if not contextual_bundle_typed_equal(counts, expected_counts) or coverage.get('sentence_count') != len(sentences) or coverage.get('occurrence_count') != len(occurrences) or (coverage.get('anchor_q') != expected_q_coverage) or (coverage.get('anchor_q', {}).get('edges') is None) or (set(coverage.get('anchor_q', {}).get('edges', {})) != set(expected_edges)) or any((coverage['anchor_q']['edges'][edge].get('draw_count') != n for edge, n in expected_edge_counts.items())):
        raise ValueError('schema-3 coverage or manifest counts mismatch')
    expected_coverage = {'sentence_count': len(sentences), 'occurrence_count': len(occurrences), 'syllable_count': len(syllables), 'zero_initial_occurrences_excluded_from_dimsim': sum((not r['initial'] for r in occurrences)), 'dimsim_natural': {'draw_count': len(pairs['dimsim-natural']), 'unique_pairs': len({r['pair_id'] for r in pairs['dimsim-natural']}), 'unique_occurrences': len({r[k] for r in pairs['dimsim-natural'] for k in ('left_id', 'right_id')}), 'unique_syllables': len({occurrence_syllable[x] for r in pairs['dimsim-natural'] for x in (r['left_id'], r['right_id'])})}, 'dimsim_balanced': {'draw_count': len(pairs['dimsim-balanced']), 'unique_pairs': len({r['pair_id'] for r in pairs['dimsim-balanced']}), 'unique_occurrences': len({r[k] for r in pairs['dimsim-balanced'] for k in ('left_id', 'right_id')}), 'unique_syllables': len({r[k] for r in pairs['dimsim-balanced'] for k in ('first_syllable_id', 'second_syllable_id')}), 'same_pronunciation_draws': sum((r['overlap_mask'] == '111' for r in pairs['dimsim-balanced']))}, 'anchor_q': expected_q_coverage}
    if not contextual_bundle_typed_equal(coverage, expected_coverage):
        raise ValueError('schema-3 coverage diagnostics disagree with saved samples')
    return {'root': root, 'manifest': manifest, 'config': config, 'coverage': coverage, 'sentences': sentences, 'occurrences': occurrences, 'syllables': syllables, 'pairs': pairs, 'triplets': triplets}

def contextual_bundle_validate_id(value):
    if not isinstance(value, str) or not value or '/' in value or ('\\' in value) or ('..' in value):
        raise ValueError('invalid schema-3 evaluation_id')
    return value

def contextual_bundle_validate_config(config, manifest_metric):
    if type(config.get('schema_version')) is not int or config['schema_version'] != contextual_bundle_SCHEMA_VERSION:
        raise ValueError('schema-3 config version must be the supported integer version')
    expected_population = {'unit': 'contextual_occurrence', 'zero_initial_excluded_from_dimsim': True, 'syllable_identity': 'canonical_toned_pinyin', 'aggregates': False}
    population = config.get('population')
    if not isinstance(population, dict) or set(population) != set(expected_population) or type(population.get('zero_initial_excluded_from_dimsim')) is not bool or (type(population.get('aggregates')) is not bool) or (not contextual_bundle_typed_equal(population, expected_population)):
        raise ValueError('schema-3 population contract mismatch')
    sampling = config.get('sampling')
    if not isinstance(sampling, dict) or set(sampling) != {'dimsim_natural', 'dimsim_balanced', 'anchor_q'}:
        raise ValueError('schema-3 sampling contract is invalid')
    expected_keys = {'dimsim_natural': {'count', 'without_replacement', 'rng', 'seed', 'sampler_version', 'pair_schema_version'}, 'dimsim_balanced': {'count', 'with_replacement', 'rng', 'seed', 'choice_order', 'sampler_version', 'pair_schema_version'}, 'anchor_q': {'draws_per_edge', 'edge_order', 'with_replacement', 'rng', 'seed', 'sampler_version', 'triplet_schema_version'}}
    specs = (('dimsim_natural', 'random.Random', False, dimsim_sampling_SAMPLER_VERSION, dimsim_sampling_PAIR_SCHEMA_VERSION), ('dimsim_balanced', 'numpy.random.PCG64', True, dimsim_sampling_SAMPLER_VERSION, dimsim_sampling_PAIR_SCHEMA_VERSION), ('anchor_q', 'numpy.random.PCG64', True, anchor_sampling_SAMPLER_VERSION, anchor_sampling_TRIPLET_SCHEMA_VERSION))
    for name, rng, replacement, sampler_version, schema_version in specs:
        item = sampling[name]
        if set(item) != expected_keys[name]:
            raise ValueError(f'schema-3 {name} config fields are invalid')
        count_key = 'draws_per_edge' if name == 'anchor_q' else 'count'
        count = item.get(count_key)
        seed = item.get('seed')
        if isinstance(count, bool) or not isinstance(count, int) or count <= 0 or isinstance(seed, bool) or (not isinstance(seed, int)) or (seed < 0) or (item.get('rng') != rng) or (type(item.get('sampler_version')) is not int) or (item['sampler_version'] != sampler_version) or (type(item.get('pair_schema_version' if name != 'anchor_q' else 'triplet_schema_version')) is not int) or (item['pair_schema_version' if name != 'anchor_q' else 'triplet_schema_version'] != schema_version) or (name != 'dimsim_natural' and item.get('with_replacement') is not replacement):
            raise ValueError(f'schema-3 {name} sampling contract is invalid')
    if sampling['dimsim_natural'].get('without_replacement') is not True:
        raise ValueError('natural DIMSIM must sample without replacement')
    if sampling['dimsim_balanced'].get('choice_order') != ['first_syllable', 'second_syllable', 'first_occurrence', 'second_occurrence']:
        raise ValueError('balanced DIMSIM choice order mismatch')
    score = config.get('scoring', {}).get('anchor_q', {})
    if score != {'cosine': 'stable_float64_scale_before_norm', 'tie_tolerance': 1e-12}:
        raise ValueError('anchor-Q scoring contract mismatch')
    if config.get('scoring', {}).get('dimsim') != manifest_metric:
        raise ValueError('DIMSIM scoring provenance mismatch')
```

```python
"""Storage, aggregation, and random baselines for aligned representation matrices."""
import hashlib
import json
from pathlib import Path
from typing import Any
import numpy as np
representations_SOURCE_HASH_NAMES = {'manifest.json', 'config.json', 'vocab.json', 'samples.jsonl', 'splits.json', 'acoustic_targets.npy'}
representations_PROVENANCE_HASH_NAMES = ('checkpoint_sha256', 'bundle_manifest_sha256', 'config_sha256', 'vocab_sha256', 'source_sha256')

def representations_is_sha256(value: Any) -> bool:
    return isinstance(value, str) and len(value) == 64 and all((c in '0123456789abcdef' for c in value))

def representations_validate_contextual_provenance(provenance: Any) -> dict[str, Any]:
    if not isinstance(provenance, dict):
        raise ValueError('contextual representation provenance is required')
    required = {'model_id', 'seed', *representations_PROVENANCE_HASH_NAMES, 'experiment_hashes', 'environment', 'evaluation_mode'}
    if not required.issubset(provenance):
        raise ValueError('contextual provenance is incomplete')
    try:
        model_id = canonical_model_id(provenance['model_id'])
    except (TypeError, ValueError) as error:
        raise ValueError('contextual provenance model_id is invalid') from error
    if model_id != provenance['model_id']:
        raise ValueError('contextual provenance model_id is not canonical')
    if not isinstance(provenance['seed'], int) or isinstance(provenance['seed'], bool):
        raise ValueError('contextual seed must be an integer')
    for name in representations_PROVENANCE_HASH_NAMES:
        if not representations_is_sha256(provenance[name]):
            raise ValueError(f'contextual provenance hash is invalid: {name}')
    hashes = provenance['experiment_hashes']
    if not isinstance(hashes, dict) or set(hashes) != representations_SOURCE_HASH_NAMES or any((not representations_is_sha256(v) for v in hashes.values())):
        raise ValueError('contextual experiment hash inventory is invalid')
    if hashes['config.json'] != provenance['config_sha256'] or hashes['vocab.json'] != provenance['vocab_sha256']:
        raise ValueError('contextual config/vocabulary hashes disagree with experiment hashes')
    environment = provenance['environment']
    env_fields = {'python', 'numpy', 'torch', 'platform', 'device'}
    if not isinstance(environment, dict) or set(environment) != env_fields or any((not isinstance(environment[key], str) or not environment[key].strip() for key in env_fields)):
        raise ValueError('contextual environment provenance is invalid')
    if provenance['evaluation_mode'] is not True:
        raise ValueError('contextual extraction must be verified in evaluation mode')
    return provenance

def validate_matrix(ids: list[str], matrix: np.ndarray, *, require_nonzero: bool=True) -> np.ndarray:
    values = np.asarray(matrix)
    if not isinstance(ids, list) or not ids or values.ndim != 2 or (values.shape[0] != len(ids)) or (values.shape[1] == 0) or any((not isinstance(x, str) or not x for x in ids)) or (len(ids) != len(set(ids))):
        raise ValueError('representation IDs and matrix shape/uniqueness disagree')
    if values.dtype.kind != 'f':
        raise ValueError('representation matrix must use a real floating-point dtype')
    if ids != sorted(ids):
        raise ValueError('representation IDs must be stably sorted')
    finite_rows = np.isfinite(values).all(axis=1)
    if not finite_rows.all():
        raise ValueError(f'representation matrix contains non-finite occurrences: {[ids[i] for i in np.flatnonzero(~finite_rows)]}')
    if require_nonzero:
        rows = values.astype(np.float64)
        scales = np.max(np.abs(rows), axis=1)
        bad = np.flatnonzero(scales == 0)
        if len(bad):
            raise ValueError(f'representation matrix contains zero-norm occurrences: {[ids[i] for i in bad]}')
    return values

def save_contextual_representations(output_dir: str | Path, *, occurrence_ids: list[str], hidden: np.ndarray, provenance: dict[str, Any]) -> Path:
    """Persist raw contextual h only; the output is immutable and hash pinned."""
    target = Path(output_dir).resolve()
    if target.exists():
        raise FileExistsError(f'representation output already exists: {target}')
    values = validate_matrix(occurrence_ids, hidden)
    provenance = representations_validate_contextual_provenance(provenance)
    target.mkdir(parents=True)
    (target / 'occurrence_ids.json').write_text(json.dumps(occurrence_ids, ensure_ascii=False) + '\n', encoding='utf-8')
    np.save(target / 'occurrence_hidden.npy', values, allow_pickle=False)
    artifacts = {name: hashlib.sha256((target / name).read_bytes()).hexdigest() for name in ('occurrence_ids.json', 'occurrence_hidden.npy')}
    manifest = {'schema_version': 1, 'representation': 'contextual_hidden_prehead', 'provenance': provenance, 'dimension': int(values.shape[1]), 'dtype': str(values.dtype), 'evaluation_mode': True, 'normalization': 'none', 'artifacts': artifacts}
    (target / 'manifest.json').write_text(json.dumps(manifest, ensure_ascii=False, sort_keys=True, indent=2) + '\n', encoding='utf-8')
    return target

def load_contextual_representations(path: str | Path) -> dict[str, Any]:
    """Load and verify an immutable raw contextual representation run."""
    root = Path(path)
    manifest = json.loads((root / 'manifest.json').read_text(encoding='utf-8'))
    if manifest.get('schema_version') != 1 or manifest.get('representation') != 'contextual_hidden_prehead':
        raise ValueError('unsupported contextual representation manifest')
    expected_artifacts = {'occurrence_ids.json', 'occurrence_hidden.npy'}
    artifact_inventory = manifest.get('artifacts')
    if not isinstance(artifact_inventory, dict) or set(artifact_inventory) != expected_artifacts:
        raise ValueError('contextual artifact inventory is invalid')
    for name, digest in artifact_inventory.items():
        if not representations_is_sha256(digest):
            raise ValueError(f'contextual artifact checksum is invalid: {name}')
        if representations_file_sha256(root / name) != digest:
            raise ValueError(f'contextual artifact checksum mismatch: {name}')
    provenance = manifest.get('provenance', {})
    representations_validate_contextual_provenance(provenance)
    if manifest.get('evaluation_mode') is not True or provenance.get('evaluation_mode') is not manifest.get('evaluation_mode') or manifest.get('normalization') != 'none':
        raise ValueError('contextual representation mode or normalization is invalid')
    ids = json.loads((root / 'occurrence_ids.json').read_text(encoding='utf-8'))
    hidden = np.load(root / 'occurrence_hidden.npy', allow_pickle=False)
    validate_matrix(ids, hidden)
    dimension = manifest.get('dimension')
    if not isinstance(dimension, int) or isinstance(dimension, bool) or dimension <= 0 or (hidden.shape[1] != dimension) or (str(hidden.dtype) != manifest.get('dtype')):
        raise ValueError('contextual artifact shape or dtype disagrees with manifest')
    return {'occurrence_ids': ids, 'hidden': hidden, 'manifest': manifest}

def representations_file_sha256(path: Path) -> str:
    return hashlib.sha256(path.read_bytes()).hexdigest()
```

### Contextual extraction


```python
"""Strict adapters that align encoder outputs to model-independent occurrences."""
import hashlib
import json
import platform
from pathlib import Path
import numpy as np
import torch

def encoders_sha256(path: Path) -> str:
    return hashlib.sha256(path.read_bytes()).hexdigest()

def encoders_checkpoint_hash_contract(source_hashes: dict, experiment_manifest: dict) -> dict:
    """Complete evaluation's five-file source pin with its validated acoustic artifact pin."""
    required = {'manifest.json', 'config.json', 'vocab.json', 'samples.jsonl', 'splits.json'}
    if not isinstance(source_hashes, dict) or set(source_hashes) != required:
        raise ValueError('evaluation source experiment hash set is invalid')
    acoustic_hash = experiment_manifest.get('artifacts', {}).get('acoustic_targets.npy')
    if not isinstance(acoustic_hash, str) or len(acoustic_hash) != 64 or any((c not in '0123456789abcdef' for c in acoustic_hash)):
        raise ValueError('validated experiment manifest lacks acoustic target hash')
    return {**source_hashes, 'acoustic_targets.npy': acoustic_hash}

def extract_bilstm(bundle_path: str | Path, *, experiment_path: str | Path, checkpoint_path: str | Path, output_dir: str | Path, device: str='cpu', batch_size: int=32, contextual_only: bool=False, expected_model_id: str | None=None, expected_seed: int | None=None, source_sha256: str | None=None) -> Path:
    """Extract full-sentence BiLSTM outputs and save exact occurrence alignment."""
    if not isinstance(batch_size, int) or batch_size <= 0 or isinstance(batch_size, bool):
        raise ValueError('batch_size must be a positive integer')
    if source_sha256 is not None and (not isinstance(source_sha256, str) or len(source_sha256) != 64 or any((c not in '0123456789abcdef' for c in source_sha256))):
        raise ValueError('source_sha256 must be a lowercase SHA-256 digest')
    bundle = load_evaluation_bundle(bundle_path)
    experiment_root, checkpoint_file = (Path(experiment_path).resolve(), Path(checkpoint_path).resolve())
    source_hashes = bundle['manifest']['source_experiment']['hashes']
    for name in ('manifest.json', 'config.json', 'vocab.json', 'samples.jsonl', 'splits.json'):
        path = experiment_root / name
        if not path.is_file() or encoders_sha256(path) != source_hashes.get(name):
            raise ValueError(f'experiment does not match evaluation source hash: {name}')
    experiment = load_experiment(experiment_root)
    expected_checkpoint_hashes = encoders_checkpoint_hash_contract(source_hashes, experiment['manifest'])
    config, vocab = (experiment['config'], experiment['vocab'])
    checkpoint = torch.load(checkpoint_file, map_location='cpu', weights_only=True)
    model_id = canonical_model_id(checkpoint.get('model_id'))
    seed = checkpoint.get('seed')
    if not isinstance(seed, int) or isinstance(seed, bool):
        raise ValueError('checkpoint seed must be an integer')
    if expected_model_id is not None and canonical_model_id(expected_model_id) != model_id:
        raise ValueError('checkpoint model_id does not match expected_model_id')
    if expected_seed is not None and (not isinstance(expected_seed, int) or isinstance(expected_seed, bool) or expected_seed != seed):
        raise ValueError('checkpoint seed does not match expected_seed')
    _, lambda_cosine = parse_model_id(model_id)
    if 'lambda_cosine' in checkpoint and checkpoint['lambda_cosine'] != lambda_cosine:
        raise ValueError('checkpoint lambda disagrees with model_id')
    if checkpoint.get('experiment_hashes') != expected_checkpoint_hashes:
        raise ValueError('checkpoint provenance does not match evaluation experiment')
    if checkpoint.get('config_sha256') != source_hashes['config.json'] or checkpoint.get('vocab_sha256') != source_hashes['vocab.json']:
        raise ValueError('checkpoint config or vocabulary provenance mismatch')
    model = build_encoder(config, vocab, model_id=model_id, seed=seed)
    model.load_state_dict(checkpoint['state_dict'], strict=True)
    model.to(torch.device(device)).eval()
    samples = experiment['samples']
    occurrence_by_sample: dict[str, list[dict]] = {}
    for occurrence in bundle['occurrences']:
        occurrence_by_sample.setdefault(occurrence['sample_id'], []).append(occurrence)
    sentence_key = 'sentence_id' if bundle['manifest'].get('schema_version') == 3 else 'sample_id'
    sentence_ids = [row[sentence_key] for row in bundle['sentences']]
    if set(sentence_ids) != set(occurrence_by_sample) or len(sentence_ids) != len(set(sentence_ids)):
        raise ValueError('bundle sentences and occurrence samples disagree')
    extracted: dict[str, np.ndarray] = {}
    with torch.no_grad():
        for start in range(0, len(sentence_ids), batch_size):
            batch_ids = sentence_ids[start:start + batch_size]
            selected = [samples[sample_id] for sample_id in batch_ids]
            batch = {key: value.to(device) if torch.is_tensor(value) else value for key, value in collate_experiment(selected).items()}
            outputs, output_mask = model.encode(batch)
            expected_mask = batch['char_mask']
            context_dimension = getattr(model, 'context_dim', None)
            if not torch.is_tensor(outputs) or not torch.is_tensor(output_mask) or (not torch.is_tensor(expected_mask)) or (expected_mask.dtype != torch.bool) or (outputs.ndim != 3) or (outputs.shape[:2] != expected_mask.shape) or (not isinstance(context_dimension, int)) or (outputs.shape[2] != context_dimension) or (output_mask.shape != expected_mask.shape) or (output_mask.dtype != torch.bool):
                raise ValueError('encoder output shape or mask shape/type is invalid')
            outputs = outputs.detach().cpu().numpy()
            if not torch.equal(output_mask, batch['char_mask']):
                raise ValueError('encoder output mask differs from source positions')
            for row_index, (sample_id, sample) in enumerate(zip(batch_ids, selected)):
                expected_positions = sorted(occurrence_by_sample[sample_id], key=lambda row: row['character_position'])
                if [row['character_position'] for row in expected_positions] != list(range(len(sample['text']))) or any((row['character'] != sample['text'][row['character_position']] for row in expected_positions)):
                    raise ValueError(f'bundle occurrences do not exactly cover source sentence {sample_id}')
                for occurrence in expected_positions:
                    extracted[occurrence['occurrence_id']] = outputs[row_index, occurrence['character_position']]
    occurrence_ids = sorted((row['occurrence_id'] for row in bundle['occurrences']))
    if set(extracted) != set(occurrence_ids):
        raise ValueError('encoder extraction did not produce every expected occurrence')
    hidden = np.stack([extracted[occurrence_id] for occurrence_id in occurrence_ids])
    validate_matrix(occurrence_ids, hidden)
    provenance = {'adapter': 'phonetics.evaluation.encoders.extract_bilstm', 'device': str(device), 'checkpoint_path': str(checkpoint_file), 'checkpoint_sha256': encoders_sha256(checkpoint_file), 'bundle_manifest_sha256': encoders_sha256(Path(bundle_path).resolve() / 'manifest.json'), 'experiment_hashes': expected_checkpoint_hashes, 'config_sha256': source_hashes['config.json'], 'vocab_sha256': source_hashes['vocab.json'], 'model_id': model_id, 'seed': seed, 'hidden_dimension': int(hidden.shape[1]), 'evaluation_mode': model.training is False, 'output_dtype': str(hidden.dtype), 'environment': {'python': platform.python_version(), 'numpy': np.__version__, 'torch': torch.__version__, 'platform': platform.platform(), 'device': str(device)}, 'source_sha256': source_sha256 or BUILD_METADATA['source_sha256']['evaluation/encoders.py']}
    if True:
        return save_contextual_representations(output_dir, occurrence_ids=occurrence_ids, hidden=hidden, provenance=provenance)
```

### Stable cosine similarity


```python
"""Numerically stable cosine calculations for raw contextual vectors."""
import numpy as np

def stable_cosine(left, right, *, left_id: str='left', right_id: str='right') -> float:
    """Compute float64 cosine after scaling each vector to avoid norm overflow."""
    a = np.asarray(left, dtype=np.float64)
    b = np.asarray(right, dtype=np.float64)
    if a.ndim != 1 or b.ndim != 1 or a.shape != b.shape:
        raise ValueError(f'cosine vectors must have the same one-dimensional shape: {left_id}, {right_id}')
    if not np.isfinite(a).all() or not np.isfinite(b).all():
        raise ValueError(f'cosine vectors contain non-finite values: {left_id}, {right_id}')
    scale_a = float(np.max(np.abs(a))) if a.size else 0.0
    scale_b = float(np.max(np.abs(b))) if b.size else 0.0
    if scale_a == 0.0 or scale_b == 0.0:
        bad = left_id if scale_a == 0.0 else right_id
        raise ValueError(f'cosine vector has zero norm: {bad}')
    a = a / scale_a
    b = b / scale_b
    norm_a, norm_b = (float(np.linalg.norm(a)), float(np.linalg.norm(b)))
    value = float(np.dot(a / norm_a, b / norm_b))
    if not np.isfinite(value):
        raise ValueError(f'cosine result is non-finite: {left_id}, {right_id}')
    return value
```

### Ranks and Spearman correlation


```python
"""RSA-style cosine analysis and strict paired comparison of encoder runs."""
import hashlib
import json
import math
from pathlib import Path
from typing import Any
import numpy as np

def distance_bin(distance: float | None) -> tuple[str, str]:
    """Return ``(status, label)`` with an explicit unbounded overflow bin."""
    if distance is None or not isinstance(distance, (int, float)) or (not math.isfinite(float(distance))):
        return ('invalid_nonfinite', 'invalid')
    value = float(distance)
    if value < 0:
        return ('invalid_negative', 'invalid')
    if value >= 120:
        return ('ok', '[120,+inf)')
    lower = int(value // 10) * 10
    return ('ok', f'[{lower},{lower + 10})')

def analysis_rankdata(values: np.ndarray) -> np.ndarray:
    order = np.argsort(values, kind='mergesort')
    ranks = np.empty(len(values), dtype=np.float64)
    position = 0
    while position < len(values):
        end = position + 1
        while end < len(values) and values[order[end]] == values[order[position]]:
            end += 1
        ranks[order[position:end]] = (position + end - 1) / 2.0 + 1.0
        position = end
    return ranks

def spearman_result(distances: list[float], dissimilarities: list[float]) -> dict[str, Any]:
    if len(distances) != len(dissimilarities):
        raise ValueError('Spearman inputs have different lengths')
    count = len(distances)
    if count < 2:
        return {'rho': None, 'valid_pair_count': count, 'undefined_reason': 'fewer_than_two_pairs'}
    left, right = (np.asarray(distances, dtype=np.float64), np.asarray(dissimilarities, dtype=np.float64))
    if not np.isfinite(left).all() or not np.isfinite(right).all():
        raise ValueError('Spearman inputs must be finite after validity filtering')
    left_rank, right_rank = (analysis_rankdata(left), analysis_rankdata(right))
    if np.ptp(left_rank) == 0 or np.ptp(right_rank) == 0:
        reason = 'constant_distance' if np.ptp(left_rank) == 0 else 'constant_hidden_dissimilarity'
        return {'rho': None, 'valid_pair_count': count, 'undefined_reason': reason}
    rho = float(np.corrcoef(left_rank, right_rank)[0, 1])
    return {'rho': rho, 'valid_pair_count': count, 'undefined_reason': None}
```

### Anchor-Q scoring


```python
"""Full precision scoring for fixed schema-3 contextual anchor-Q draws."""
import hashlib
import math
from pathlib import Path
from typing import Any
import numpy as np
DEFAULT_TIE_TOLERANCE = 1e-12

def anchor_q_validate_tolerance(value: float) -> float:
    if isinstance(value, bool) or not isinstance(value, (int, float)) or (not math.isfinite(value)) or (value < 0):
        raise ValueError('tie_tolerance must be a finite non-negative number')
    return float(value)

def anchor_q_score_anchor_q(bundle: dict[str, Any], occurrence_ids: list[str], hidden: np.ndarray, provenance: dict[str, Any], *, bundle_manifest_sha256: str, tie_tolerance: float=DEFAULT_TIE_TOLERANCE) -> dict[str, Any]:
    """Score all pre-sampled draws against raw contextual h.

    Inputs are a verified schema-3 bundle object, its representation occurrence ID
    inventory and raw matrix (rows may be in any order), validated representation
    provenance, and the SHA-256 digest of the already-verified bundle manifest.
    Returns JSON-compatible draw scores, edge/component/overall Q and
    MC standard errors, plus coverage and provenance. No sampling or h aggregation occurs.
    """
    tolerance = anchor_q_validate_tolerance(tie_tolerance)
    manifest = bundle['manifest']
    if manifest.get('schema_version') != 3:
        raise ValueError('anchor-Q scorer requires a schema-3 bundle')
    expected_hash = bundle_manifest_sha256
    if not isinstance(expected_hash, str) or len(expected_hash) != 64 or any((char not in '0123456789abcdef' for char in expected_hash)):
        raise ValueError('bundle manifest SHA-256 digest is invalid')
    if provenance.get('bundle_manifest_sha256') != expected_hash:
        raise ValueError('representation provenance bundle manifest hash mismatch')
    bundle_occurrences = [row['occurrence_id'] for row in bundle['occurrences']]
    if len(occurrence_ids) != len(set(occurrence_ids)) or set(occurrence_ids) != set(bundle_occurrences):
        raise ValueError('representation occurrence ID inventory does not match bundle')
    matrix = np.asarray(hidden)
    if matrix.ndim != 2 or matrix.shape[0] != len(occurrence_ids):
        raise ValueError('hidden matrix rows do not match occurrence ID inventory')
    if matrix.dtype.kind != 'f':
        raise ValueError('raw hidden matrix must have a floating-point dtype')
    positions = {oid: i for i, oid in enumerate(occurrence_ids)}
    sampling = bundle['config']['sampling']['anchor_q']
    contract_tol = bundle['config']['scoring']['anchor_q']['tie_tolerance']
    if tolerance != contract_tol:
        raise ValueError('tie_tolerance differs from immutable bundle scoring contract')
    expected_edges = [f'{c}:{lo}->{hi}' for c, lo, hi in EDGES]
    if sampling['edge_order'] != expected_edges:
        raise ValueError('bundle anchor-Q edge order mismatch')
    per_edge = sampling['draws_per_edge']
    rows = bundle['triplets']
    if len(rows) != len(expected_edges) * per_edge:
        raise ValueError('anchor-Q draw count does not match bundle contract')
    counts = {edge: 0 for edge in expected_edges}
    draw_ids = set()
    scored = []
    for global_index, row in enumerate(rows):
        draw_id, edge = (row['draw_id'], row['edge_id'])
        expected_edge = expected_edges[global_index // per_edge]
        if edge != expected_edge:
            raise ValueError(f'anchor-Q rows violate global edge block order at draw {draw_id}')
        if draw_id in draw_ids:
            raise ValueError(f'duplicate anchor-Q draw ID: {draw_id}')
        draw_ids.add(draw_id)
        if edge not in counts:
            raise ValueError(f'unexpected anchor-Q edge: {edge}')
        if len({row[k] for k in ('anchor_id', 'high_id', 'low_id')}) != 3:
            raise ValueError(f'triplet occurrence IDs must be distinct: {draw_id}')
        if type(row.get('draw_index')) is not int or row['draw_index'] != counts[edge]:
            raise ValueError(f'anchor-Q draw index/order mismatch: {draw_id}')
        expected_draw_id = f'anchor-q:v1:{edge}:{counts[edge]:06d}'
        if draw_id != expected_draw_id:
            raise ValueError(f'anchor-Q draw ID/order mismatch: {draw_id}')
        counts[edge] += 1
        if counts[edge] > per_edge:
            raise ValueError(f'anchor-Q edge draw count exceeds contract: {edge}')
        a, h, l = (row[k] for k in ('anchor_id', 'high_id', 'low_id'))
        ca = stable_cosine(matrix[positions[a]], matrix[positions[h]], left_id=a, right_id=h)
        cb = stable_cosine(matrix[positions[a]], matrix[positions[l]], left_id=a, right_id=l)
        difference = ca - cb
        score = 0.5 if abs(difference) <= tolerance else 1.0 if difference > 0 else 0.0
        scored.append({'draw_id': draw_id, 'triplet_id': row['triplet_id'], 'edge_id': edge, 'cosine_anchor_high': ca, 'cosine_anchor_low': cb, 'difference': difference, 'score': score})
    if any((counts[e] != per_edge for e in expected_edges)):
        raise ValueError('anchor-Q edge count mismatch; complete coverage required')
    edge_results = {}
    edge_ses = {}
    for edge in expected_edges:
        values = np.asarray([r['score'] for r in scored if r['edge_id'] == edge], dtype=np.float64)
        se = float(np.std(values, ddof=1) / math.sqrt(len(values))) if len(values) > 1 else None
        edge_ses[edge] = se
        edge_results[edge] = {'q': float(np.mean(values)), 'mc_se': se, 'mc_se_status': 'ok' if se is not None else 'not_available_single_draw', 'draw_count': int(len(values))}
    components = {}
    for component in ('initial', 'final', 'tone'):
        edges = [f'{c}:{lo}->{hi}' for c, lo, hi in EDGES if c == component]
        components[component] = {'q': float(np.mean([edge_results[e]['q'] for e in edges])), 'mc_se': float(math.sqrt(sum((edge_ses[e] ** 2 for e in edges))) / len(edges)) if all((edge_ses[e] is not None for e in edges)) else None, 'mc_se_status': 'ok' if all((edge_ses[e] is not None for e in edges)) else 'not_available_single_draw'}
    overall_se = float(math.sqrt(sum((edge_ses[e] ** 2 for e in expected_edges))) / len(expected_edges)) if all((edge_ses[e] is not None for e in expected_edges)) else None
    return {'schema_version': 1, 'model_id': provenance['model_id'], 'seed': provenance['seed'], 'draw_scores': scored, 'edges': edge_results, 'components': components, 'overall': {'q': float(np.mean([edge_results[e]['q'] for e in expected_edges])), 'mc_se': overall_se, 'mc_se_status': 'ok' if overall_se is not None else 'not_available_single_draw'}, 'coverage': bundle['coverage']['anchor_q'], 'provenance': provenance, 'bundle_manifest_sha256': expected_hash, 'tie_tolerance': tolerance}

def score_anchor_q_paths(bundle_path: str | Path, representation_path: str | Path) -> dict[str, Any]:
    """Load validated schema-3 bundle and contextual representation paths, then score."""
    bundle = load_evaluation_bundle(bundle_path)
    if bundle['manifest'].get('schema_version') != 3:
        raise ValueError('anchor-Q path scorer requires a schema-3 bundle')
    bundle_manifest_sha256 = hashlib.sha256((Path(bundle_path) / 'manifest.json').read_bytes()).hexdigest()
    representation = load_contextual_representations(representation_path)
    return anchor_q_score_anchor_q(bundle, representation['occurrence_ids'], representation['hidden'], representation['manifest']['provenance'], bundle_manifest_sha256=bundle_manifest_sha256, tie_tolerance=bundle['config']['scoring']['anchor_q']['tie_tolerance'])
```

### Five-seed discovery


```python
"""Explicit, integrity-checked discovery of seeds in a training run."""
import hashlib
import json
from pathlib import Path
from typing import Any

def seed_runs_sha256(path: Path) -> str:
    digest = hashlib.sha256()
    with path.open('rb') as stream:
        for block in iter(lambda: stream.read(1024 * 1024), b''):
            digest.update(block)
    return digest.hexdigest()

def discover_seed_runs(run_dir: str | Path, expected_model_id: str, expected_seeds: tuple[int, ...]=(0, 1, 2, 3, 4)) -> dict[str, Any]:
    """Return sorted seed/checkpoint records from exactly the selected run directory."""
    root = Path(run_dir).resolve(strict=True)
    if not root.is_dir():
        raise ValueError('training run path must be a directory')
    model_id = canonical_model_id(expected_model_id)
    if not isinstance(expected_seeds, (tuple, list)) or not expected_seeds:
        raise ValueError('expected_seeds must be a non-empty sequence of integers')
    if any((type(seed) is not int for seed in expected_seeds)) or len(set(expected_seeds)) != len(expected_seeds):
        raise ValueError('expected_seeds must contain distinct integers')
    expected_seeds = tuple(sorted(expected_seeds))
    if len(expected_seeds) != 5 or any((seed < 0 for seed in expected_seeds)):
        raise ValueError('expected_seeds must contain exactly five non-negative seed IDs')
    manifest_path = root / 'manifest.json'
    try:
        raw = manifest_path.read_bytes()
    except OSError as error:
        raise ValueError(f'training run manifest is missing or unreadable: {manifest_path}') from error
    manifest = json.loads(raw)
    if type(manifest.get('schema_version')) is not int or manifest.get('schema_version') != 1 or manifest.get('run_type') != 'training_run' or (manifest.get('status') != 'complete') or (manifest.get('run_id') != root.name) or (manifest.get('model_id') != model_id):
        raise ValueError('training run manifest identity or completion status is invalid')
    seeds = manifest.get('seeds')
    if not isinstance(seeds, list) or any((type(seed) is not int for seed in seeds)) or len(seeds) != len(set(seeds)) or (tuple(sorted(seeds)) != expected_seeds):
        raise ValueError('training run seed inventory differs from expected seeds')
    statuses = manifest.get('seed_status')
    if not isinstance(statuses, dict) or statuses != {str(seed): 'complete' for seed in seeds}:
        raise ValueError('training run does not have complete status for every seed')
    seed_dirs = {path.name for path in root.iterdir() if path.is_dir() and path.name.startswith('seed-')}
    if seed_dirs != {f'seed-{seed}' for seed in expected_seeds}:
        raise ValueError('training run seed directory inventory is not exact')
    artifact_hashes = manifest.get('artifacts')
    if not isinstance(artifact_hashes, dict):
        raise ValueError('training run artifact inventory is missing')
    records = []
    for seed in expected_seeds:
        relative = Path(f'seed-{seed}') / 'best.pt'
        try:
            checkpoint = (root / relative).resolve(strict=True)
        except OSError as error:
            raise ValueError(f'missing checkpoint for model {model_id}, seed {seed}: {root / relative}') from error
        if root not in checkpoint.parents or not checkpoint.is_file():
            raise ValueError('checkpoint path escapes training run or is not a file')
        expected_hash = artifact_hashes.get(relative.as_posix())
        actual_hash = seed_runs_sha256(checkpoint)
        if not isinstance(expected_hash, str) or expected_hash != actual_hash:
            raise ValueError(f'training checkpoint checksum mismatch: seed {seed}')
        records.append({'seed': seed, 'checkpoint_path': str(checkpoint), 'checkpoint_sha256': actual_hash})
    experiment_hashes = manifest.get('experiment_hashes')
    required = {'manifest.json', 'config.json', 'vocab.json', 'samples.jsonl', 'splits.json', 'acoustic_targets.npy'}
    if not isinstance(experiment_hashes, dict) or set(experiment_hashes) != required or any((not isinstance(value, str) or len(value) != 64 or any((char not in '0123456789abcdef' for char in value)) for value in experiment_hashes.values())):
        raise ValueError('training source experiment hash inventory is invalid')
    return {'run_dir': str(root), 'run_id': manifest['run_id'], 'model_id': model_id, 'manifest_sha256': hashlib.sha256(raw).hexdigest(), 'experiment_hashes': experiment_hashes, 'seeds': records}
```

### Anchor-Q across models and seeds


```python
"""Batch scoring for explicitly selected complete training runs."""
import hashlib
from pathlib import Path
from typing import Any

def score_anchor_q_batch(bundle_path: str | Path, run_dirs: dict[str, str | Path], representation_dirs: dict[str, dict[int, str | Path]]) -> dict[str, Any]:
    """Score all five explicit seeds per model against one verified schema-3 bundle."""
    if not isinstance(run_dirs, dict) or not run_dirs or (not isinstance(representation_dirs, dict)):
        raise ValueError('explicit run and representation mappings are required')
    canonical_runs = {canonical_model_id(key): value for key, value in run_dirs.items()}
    canonical_reps = {canonical_model_id(key): value for key, value in representation_dirs.items()}
    if len(canonical_runs) != len(run_dirs) or len(canonical_reps) != len(representation_dirs) or set(canonical_runs) != set(canonical_reps):
        raise ValueError('run and representation model inventories must match uniquely')
    bundle = load_evaluation_bundle(bundle_path)
    if bundle['manifest'].get('schema_version') != 3:
        raise ValueError('batch anchor-Q requires a schema-3 bundle')
    bundle_root = Path(bundle_path).resolve()
    bundle_sha = hashlib.sha256((bundle_root / 'manifest.json').read_bytes()).hexdigest()
    source_hashes = bundle['manifest'].get('source_experiment', {}).get('hashes')
    if not isinstance(source_hashes, dict) or set(source_hashes) != {'manifest.json', 'config.json', 'vocab.json', 'samples.jsonl', 'splits.json'}:
        raise ValueError('schema-3 bundle source experiment hashes are missing')
    scored = []
    for model_id in sorted(canonical_runs):
        try:
            run = discover_seed_runs(canonical_runs[model_id], model_id)
        except (OSError, ValueError) as error:
            raise ValueError(f'training run validation failed for model {model_id}: {error}') from error
        rep_seeds = canonical_reps[model_id]
        if not isinstance(rep_seeds, dict) or any((type(seed) is not int for seed in rep_seeds)) or set(rep_seeds) != {0, 1, 2, 3, 4}:
            raise ValueError(f'representations must explicitly provide seeds 0 through 4 for {model_id}')
        for record in run['seeds']:
            seed = record['seed']
            try:
                loaded = load_contextual_representations(rep_seeds[seed])
            except (OSError, ValueError) as error:
                raise ValueError(f'representation validation failed for model {model_id}, seed {seed}, path {rep_seeds[seed]}: {error}') from error
            provenance = loaded['manifest']['provenance']
            if provenance['model_id'] != model_id or provenance['seed'] != seed:
                raise ValueError(f'representation model/seed mismatch: {model_id} seed {seed}')
            if provenance['checkpoint_sha256'] != record['checkpoint_sha256']:
                raise ValueError(f'representation checkpoint mismatch: {model_id} seed {seed}')
            if provenance['experiment_hashes'] != run['experiment_hashes']:
                raise ValueError(f'representation training source hashes mismatch: {model_id} seed {seed}')
            if any((run['experiment_hashes'].get(name) != digest for name, digest in source_hashes.items())):
                raise ValueError('evaluation bundle and training run source experiment hashes differ')
            result = anchor_q_score_anchor_q(bundle, loaded['occurrence_ids'], loaded['hidden'], provenance, bundle_manifest_sha256=bundle_sha, tie_tolerance=bundle['config']['scoring']['anchor_q']['tie_tolerance'])
            scored.append({'model_id': model_id, 'seed': seed, 'run_id': run['run_id'], 'run_manifest_sha256': run['manifest_sha256'], 'checkpoint_path': record['checkpoint_path'], 'checkpoint_sha256': record['checkpoint_sha256'], 'experiment_hashes': run['experiment_hashes'], 'score': result})
    return {'schema_version': 1, 'bundle_manifest_sha256': bundle_sha, 'results': scored}
```

### Anchor-Q summaries and paired differences


```python
"""Pure, full-precision tables and plot data for complete anchor-Q batches.

The report preserves model/seed identity. MC SE describes fixed-draw Monte Carlo
variation; sample SD describes variation across the five training seeds. They are
reported separately and are never combined into confidence intervals.
"""
import math
from numbers import Real
from typing import Any
import numpy as np
anchor_q_reporting_SEEDS = (0, 1, 2, 3, 4)
COMPONENTS = ('initial', 'final', 'tone')
METRICS = (*COMPONENTS, 'overall')
EDGE_ORDER = tuple((f'{component}:{low}->{high}' for component, low, high in EDGES))

def anchor_q_reporting_number(value: Any) -> bool:
    return isinstance(value, Real) and (not isinstance(value, bool)) and math.isfinite(float(value))

def anchor_q_reporting_sample_sd(values: list[float]) -> float:
    return float(np.std(np.asarray(values, dtype=np.float64), ddof=1))

def anchor_q_reporting_mean_se(draws: list[np.ndarray]) -> tuple[float, float | None]:
    """MC SE of the seed-mean estimator, retaining corresponding draw pairing."""
    pooled = np.mean(np.stack(draws, axis=0), axis=0)
    if pooled.size < 2:
        return (float(np.mean(pooled)), None)
    return (float(np.mean(pooled)), float(np.std(pooled, ddof=1) / math.sqrt(pooled.size)))

def anchor_q_reporting_edge_ids(draws: list[dict[str, Any]], edges: list[str]) -> tuple[list[tuple[str, str, str]], dict[str, list[float]]]:
    identities = []
    values = {edge: [] for edge in edges}
    for row in draws:
        if not isinstance(row, dict):
            raise ValueError('draw score rows must be objects')
        draw_id, edge_id, triplet_id = (row.get(key) for key in ('draw_id', 'edge_id', 'triplet_id'))
        if not all((isinstance(x, str) and x for x in (draw_id, edge_id, triplet_id))):
            raise ValueError('draw identities must be non-empty strings')
        if edge_id not in values:
            raise ValueError(f'unexpected edge identity: {edge_id}')
        score = row.get('score')
        if not anchor_q_reporting_number(score) or float(score) not in (0.0, 0.5, 1.0):
            raise ValueError('draw score must be numeric 0, 0.5, or 1')
        identities.append((draw_id, edge_id, triplet_id))
        values[edge_id].append(float(score))
    if len({item[0] for item in identities}) != len(identities):
        raise ValueError('draw IDs must be unique within each seed')
    expected_edge_sequence = [edge for edge in edges for _ in values[edge]]
    if [item[1] for item in identities] != expected_edge_sequence:
        raise ValueError('draw rows must follow canonical edge blocks')
    return (identities, values)

def build_anchor_q_report(batch_result: dict[str, Any], reference_model: str='syllable') -> dict[str, Any]:
    """Validate paired five-seed batch scores and return tidy JSON-compatible data.

    ``tables`` contains per-model/per-seed edges and metrics, per-model seed
    summaries, per-model global summaries, paired deltas, and normalized shares.
    ``plot_data`` contains edge means with their five seed points and the two share
    series. All values retain full numeric precision. Missing one-draw MC SE is
    represented by ``None`` and an explicit status.
    """
    if not isinstance(batch_result, dict) or type(batch_result.get('schema_version')) is not int or batch_result['schema_version'] != 1:
        raise ValueError('batch schema_version must be integer 1')
    bundle_hash = batch_result.get('bundle_manifest_sha256')
    if not isinstance(bundle_hash, str) or len(bundle_hash) != 64 or any((c not in '0123456789abcdef' for c in bundle_hash)):
        raise ValueError('invalid batch bundle manifest SHA-256')
    try:
        reference = canonical_model_id(reference_model)
    except (TypeError, ValueError) as error:
        raise ValueError('invalid reference model') from error
    records = batch_result.get('results')
    if not isinstance(records, list) or not records:
        raise ValueError('batch results are required')
    models = set()
    for record in records:
        if not isinstance(record, dict) or not isinstance(record.get('model_id'), str):
            raise ValueError('each batch record requires a model ID')
        try:
            model = canonical_model_id(record['model_id'])
        except (TypeError, ValueError) as error:
            raise ValueError('invalid model ID') from error
        if model != record['model_id']:
            raise ValueError('model IDs must be canonical')
        models.add(model)
    if reference not in models:
        raise ValueError('reference model is absent from batch')
    model_order = sorted(models)
    grouped: dict[str, dict[int, dict[str, Any]]] = {model: {} for model in model_order}
    for record in records:
        model, seed = (record['model_id'], record.get('seed'))
        if type(seed) is not int or seed not in anchor_q_reporting_SEEDS or seed in grouped[model]:
            raise ValueError(f'invalid or duplicate seed for {model}')
        score = record.get('score')
        if not isinstance(score, dict):
            raise ValueError('score must be an object')
        if score.get('model_id') != model or type(score.get('seed')) is not int or score['seed'] != seed:
            raise ValueError('internal model/seed metadata mismatch')
        if score.get('bundle_manifest_sha256') != bundle_hash:
            raise ValueError('bundle hash mismatch')
        provenance = score.get('provenance')
        if not isinstance(provenance, dict) or provenance.get('model_id') != model or type(provenance.get('seed')) is not int or (provenance['seed'] != seed) or (provenance.get('bundle_manifest_sha256') != bundle_hash):
            raise ValueError('score provenance model/seed/bundle metadata mismatch')
        tolerance = score.get('tie_tolerance')
        if not anchor_q_reporting_number(tolerance) or tolerance < 0:
            raise ValueError('invalid tie tolerance')
        grouped[model][seed] = score
    if any((set(scores) != set(anchor_q_reporting_SEEDS) for scores in grouped.values())):
        raise ValueError('every model requires seeds 0 through 4 exactly once')
    first_score = grouped[model_order[0]][anchor_q_reporting_SEEDS[0]]
    tolerance = float(first_score['tie_tolerance'])
    edge_rows: list[dict[str, Any]] = []
    metric_rows: list[dict[str, Any]] = []
    all_draw_values: dict[str, dict[int, dict[str, np.ndarray]]] = {model: {} for model in model_order}
    identity_reference = None
    draw_count = None
    coverage_reference = None
    for model in model_order:
        for seed in anchor_q_reporting_SEEDS:
            score = grouped[model][seed]
            if float(score['tie_tolerance']) != tolerance:
                raise ValueError('tie tolerance mismatch')
            edge_summary = score.get('edges')
            if not isinstance(edge_summary, dict) or set(edge_summary) != set(EDGE_ORDER):
                raise ValueError('score edges must contain the canonical twelve-edge inventory')
            draws = score.get('draw_scores')
            if not isinstance(draws, list) or not draws:
                raise ValueError('draw_scores must be a non-empty list')
            identities, edge_values = anchor_q_reporting_edge_ids(draws, list(EDGE_ORDER))
            expected_draw_ids = [f'anchor-q:v1:{edge}:{index:06d}' for edge in EDGE_ORDER for index in range(len(edge_values[edge]))]
            if [identity[0] for identity in identities] != expected_draw_ids:
                raise ValueError('draw IDs do not match canonical edge/index order')
            if identity_reference is None:
                identity_reference = identities
            elif identities != identity_reference:
                raise ValueError('draw identities differ across models or seeds')
            sizes = {edge: len(edge_values[edge]) for edge in EDGE_ORDER}
            if len(set(sizes.values())) != 1:
                raise ValueError('all edge draw counts must match')
            if draw_count is None:
                draw_count = next(iter(sizes.values()))
            if any((count != draw_count for count in sizes.values())):
                raise ValueError('edge draw counts differ across models or seeds')
            coverage = score.get('coverage')
            if not isinstance(coverage, dict) or not isinstance(coverage.get('edges'), dict) or set(coverage['edges']) != set(EDGE_ORDER):
                raise ValueError('per-edge coverage diagnostics are required')
            if coverage_reference is None:
                coverage_reference = coverage
            elif coverage != coverage_reference:
                raise ValueError('coverage diagnostics differ across models or seeds')
            arrays = {edge: np.asarray(edge_values[edge], dtype=np.float64) for edge in EDGE_ORDER}
            all_draw_values[model][seed] = arrays
            q = {}
            mcse = {}
            for edge in EDGE_ORDER:
                vals = arrays[edge]
                q[edge] = float(np.mean(vals))
                mcse[edge] = float(np.std(vals, ddof=1) / math.sqrt(vals.size)) if vals.size > 1 else None
                declared = edge_summary[edge]
                if not isinstance(declared, dict) or type(declared.get('draw_count')) is not int or declared['draw_count'] != vals.size:
                    raise ValueError('declared edge draw_count differs from actual draws')
                cov = coverage['edges'][edge]
                edge_rows.append({'model_id': model, 'seed': seed, 'edge_id': edge, 'q': q[edge], 'mc_se': mcse[edge], 'mc_se_status': 'ok' if mcse[edge] is not None else 'not_available_single_draw', 'draw_count': int(vals.size), 'coverage': cov})
            for component in COMPONENTS:
                edges = [edge for edge in EDGE_ORDER if edge.startswith(component + ':')]
                metric_rows.append({'model_id': model, 'seed': seed, 'metric': component, 'q': float(np.mean([q[e] for e in edges])), 'mc_se': float(math.sqrt(sum((mcse[e] ** 2 for e in edges))) / len(edges)) if all((mcse[e] is not None for e in edges)) else None, 'mc_se_status': 'ok' if all((mcse[e] is not None for e in edges)) else 'not_available_single_draw'})
            metric_rows.append({'model_id': model, 'seed': seed, 'metric': 'overall', 'q': float(np.mean([q[e] for e in EDGE_ORDER])), 'mc_se': float(math.sqrt(sum((mcse[e] ** 2 for e in EDGE_ORDER))) / len(EDGE_ORDER)) if all((mcse[e] is not None for e in EDGE_ORDER)) else None, 'mc_se_status': 'ok' if all((mcse[e] is not None for e in EDGE_ORDER)) else 'not_available_single_draw'})
    edge_summary_rows = []
    for model in model_order:
        for edge in EDGE_ORDER:
            seed_values = [float(np.mean(all_draw_values[model][seed][edge])) for seed in anchor_q_reporting_SEEDS]
            mean_q, mean_mcse = anchor_q_reporting_mean_se([all_draw_values[model][seed][edge] for seed in anchor_q_reporting_SEEDS])
            edge_summary_rows.append({'model_id': model, 'edge_id': edge, 'mean': mean_q, 'sample_sd': anchor_q_reporting_sample_sd(seed_values), 'mc_se': mean_mcse, 'mc_se_status': 'ok' if mean_mcse is not None else 'not_available_single_draw', 'seed_values': [{'seed': seed, 'q': val} for seed, val in zip(anchor_q_reporting_SEEDS, seed_values)]})
    metric_summary_rows = []
    global_summary_rows = []
    for model in model_order:
        for metric in METRICS:
            edge_subset = list(EDGE_ORDER) if metric == 'overall' else [e for e in EDGE_ORDER if e.startswith(metric + ':')]
            seed_values = []
            edge_mean_ses = []
            for seed in anchor_q_reporting_SEEDS:
                edge_means = [float(np.mean(all_draw_values[model][seed][edge])) for edge in edge_subset]
                seed_values.append(float(np.mean(edge_means)))
            mean_q = float(np.mean(seed_values))
            for edge in edge_subset:
                _, edge_mean_se = anchor_q_reporting_mean_se([all_draw_values[model][seed][edge] for seed in anchor_q_reporting_SEEDS])
                edge_mean_ses.append(edge_mean_se)
            mean_mcse = float(math.sqrt(sum((value ** 2 for value in edge_mean_ses))) / len(edge_subset)) if all((value is not None for value in edge_mean_ses)) else None
            row = {'model_id': model, 'metric': metric, 'mean': mean_q, 'sample_sd': anchor_q_reporting_sample_sd(seed_values), 'mc_se': mean_mcse, 'mc_se_status': 'ok' if mean_mcse is not None else 'not_available_single_draw', 'seed_values': [{'seed': seed, 'q': val} for seed, val in zip(anchor_q_reporting_SEEDS, seed_values)]}
            metric_summary_rows.append(row)
            global_summary_rows.append(dict(row))
    paired_rows = []
    for model in model_order:
        if model == reference:
            continue
        for metric in METRICS:
            edge_subset = list(EDGE_ORDER) if metric == 'overall' else [e for e in EDGE_ORDER if e.startswith(metric + ':')]
            seed_deltas = []
            mean_draw_differences = []
            for seed in anchor_q_reporting_SEEDS:
                edge_diff_means = []
                for edge in edge_subset:
                    diffs = all_draw_values[model][seed][edge] - all_draw_values[reference][seed][edge]
                    edge_diff_means.append(float(np.mean(diffs)))
                seed_deltas.append(float(np.mean(edge_diff_means)))
            for edge in edge_subset:
                across_seeds = np.mean(np.stack([all_draw_values[model][seed][edge] - all_draw_values[reference][seed][edge] for seed in anchor_q_reporting_SEEDS]), axis=0)
                mean_draw_differences.append(across_seeds)
            edge_mcse = [float(np.std(values, ddof=1) / math.sqrt(values.size)) if values.size > 1 else None for values in mean_draw_differences]
            paired_mcse = float(math.sqrt(sum((value ** 2 for value in edge_mcse)) / len(edge_mcse) ** 2)) if all((value is not None for value in edge_mcse)) else None
            paired_rows.append({'model_id': model, 'reference_model': reference, 'metric': metric, 'mean_delta_q': float(np.mean(seed_deltas)), 'sample_sd_seed_delta': anchor_q_reporting_sample_sd(seed_deltas), 'mc_se': paired_mcse, 'mc_se_status': 'ok' if paired_mcse is not None else 'not_available_single_draw', 'seed_deltas': [{'seed': seed, 'delta_q': value} for seed, value in zip(anchor_q_reporting_SEEDS, seed_deltas)]})
    share_rows = []
    component_share_rows = []
    for model in model_order:
        means = {row['edge_id']: row['mean'] for row in edge_summary_rows if row['model_id'] == model}
        total = sum(means.values())
        for edge in EDGE_ORDER:
            share_rows.append({'model_id': model, 'edge_id': edge, 'share': means[edge] / total if total else None, 'status': 'ok' if total else 'undefined_zero_total'})
        comp_means = {component: float(np.mean([means[e] for e in EDGE_ORDER if e.startswith(component + ':')])) for component in COMPONENTS}
        comp_total = sum(comp_means.values())
        for component in COMPONENTS:
            component_share_rows.append({'model_id': model, 'component': component, 'share': comp_means[component] / comp_total if comp_total else None, 'status': 'ok' if comp_total else 'undefined_zero_total'})
    plot_edges = [{'model_id': model, 'edge_id': edge, 'mean': next((row['mean'] for row in edge_summary_rows if row['model_id'] == model and row['edge_id'] == edge)), 'seed_points': [{'seed': seed, 'q': float(np.mean(all_draw_values[model][seed][edge]))} for seed in anchor_q_reporting_SEEDS]} for model in model_order for edge in EDGE_ORDER]
    return {'schema_version': 1, 'contract': {'seeds': list(anchor_q_reporting_SEEDS), 'edge_order': list(EDGE_ORDER), 'metrics': list(METRICS), 'reference_model': reference, 'bundle_manifest_sha256': bundle_hash, 'tie_tolerance': tolerance, 'draws_per_edge': draw_count, 'full_precision': True, 'seed_summary': 'mean and sample SD (ddof=1) across five seeds', 'mc_uncertainty': 'MC SE from draw scores averaged across seeds before sample SD / sqrt(draw count); independent-edge aggregation', 'paired_uncertainty': 'paired draw differences, averaged across seeds before MC SE; no confidence intervals', 'tables': ['model_seed_edges', 'model_seed_metrics', 'model_edge_seed_summary', 'model_metric_seed_summary', 'global_metric_summary', 'paired_deltas', 'transition_shares', 'component_shares'], 'plot_data': ['edges with mean and five seed points', 'transition shares', 'component shares']}, 'tables': {'model_seed_edges': edge_rows, 'model_seed_metrics': metric_rows, 'model_edge_seed_summary': edge_summary_rows, 'model_metric_seed_summary': metric_summary_rows, 'global_metric_summary': global_summary_rows, 'paired_deltas': paired_rows, 'transition_shares': share_rows, 'component_shares': component_share_rows}, 'plot_data': {'edges': plot_edges, 'transition_shares': share_rows, 'component_shares': component_share_rows}}
```

### DIMSIM scoring and boundary diagnostics


```python
"""Full-range contextual DIMSIM scoring for immutable schema-3 bundles."""
import hashlib
import math
from pathlib import Path
from typing import Any
import numpy as np
MODES = ('dimsim-natural', 'dimsim-balanced')
BIN_LABELS = tuple((f'[{x},{x + 10})' for x in range(0, 120, 10))) + ('[120,+inf)',)
MASKS = tuple((f'{i:03b}' for i in range(8)))

def contextual_dimsim_positive_int(value: Any, name: str) -> int:
    if type(value) is not int or value <= 0:
        raise ValueError(f'{name} must be a positive integer')
    return value

def contextual_dimsim_rho(rows: list[dict]) -> dict:
    return spearman_result([float(r['dimsim_distance']) for r in rows], [float(r['hidden_dissimilarity']) for r in rows])

def score_contextual_dimsim(bundle: dict, occurrence_ids: list[str], hidden: np.ndarray, provenance: dict, bundle_manifest_sha256: str, analysis_config: dict | None=None) -> dict:
    """Score verified DIMSIM draws against raw contextual h.

    Returns both pair tables, full-range and overlap-stratum Spearman statistics,
    distance-bin diagnostics, and per-mode exploratory right-tail boundary curves.
    Invalid DIMSIM metrics remain in ``rows`` but are excluded from statistics.
    """
    cfg = {} if analysis_config is None else analysis_config
    if not isinstance(cfg, dict) or set(cfg) - {'min_unique_pairs_per_bin', 'boundary_min_side_count', 'boundary_thresholds'}:
        raise ValueError('analysis_config contains unsupported settings')
    bin_support = contextual_dimsim_positive_int(cfg.get('min_unique_pairs_per_bin', 100), 'min_unique_pairs_per_bin')
    boundary_support = contextual_dimsim_positive_int(cfg.get('boundary_min_side_count', 500), 'boundary_min_side_count')
    thresholds = cfg.get('boundary_thresholds', list(range(10, 111)))
    if not isinstance(thresholds, list) or not thresholds or any((type(x) is not int or not 10 <= x <= 110 for x in thresholds)) or (len(set(thresholds)) != len(thresholds)):
        raise ValueError('boundary_thresholds must be distinct integers from 10 through 110')
    if thresholds != sorted(thresholds):
        raise ValueError('boundary_thresholds must be sorted ascending')
    if not isinstance(bundle, dict) or type(bundle.get('manifest', {}).get('schema_version')) is not int or bundle['manifest']['schema_version'] != 3:
        raise ValueError('a verified schema-3 bundle is required')
    if not isinstance(bundle_manifest_sha256, str) or len(bundle_manifest_sha256) != 64 or any((c not in '0123456789abcdef' for c in bundle_manifest_sha256)):
        raise ValueError('bundle_manifest_sha256 must be lowercase SHA-256')
    if not isinstance(provenance, dict) or provenance.get('bundle_manifest_sha256') != bundle_manifest_sha256:
        raise ValueError('representation provenance does not match bundle manifest')
    expected = sorted((r['occurrence_id'] for r in bundle['occurrences']))
    ids_are_valid = isinstance(occurrence_ids, list) and all((isinstance(x, str) and x for x in occurrence_ids))
    if not ids_are_valid or len(occurrence_ids) != len(set(occurrence_ids)) or sorted(occurrence_ids) != expected:
        actual = set(occurrence_ids) if ids_are_valid else set()
        missing = sorted(set(expected) - actual)
        extra = sorted(actual - set(expected))
        raise ValueError(f'representation occurrence inventory mismatch; missing IDs={missing}, extra IDs={extra}')
    matrix = np.asarray(hidden)
    if matrix.ndim != 2 or matrix.shape[0] != len(occurrence_ids) or matrix.shape[1] == 0 or (matrix.dtype.kind != 'f') or (not np.isfinite(matrix).all()):
        bad_ids = [occurrence_ids[i] for i in range(min(len(occurrence_ids), matrix.shape[0] if matrix.ndim else 0)) if matrix.ndim == 2 and (not np.isfinite(matrix[i]).all())]
        raise ValueError(f'hidden matrix must be a finite floating matrix; invalid IDs={bad_ids}')
    scales = np.max(np.abs(matrix.astype(np.float64)), axis=1)
    if np.any(scales == 0):
        raise ValueError(f'hidden matrix has zero-norm IDs={[occurrence_ids[i] for i in np.flatnonzero(scales == 0)]}')
    position = {key: i for i, key in enumerate(occurrence_ids)}
    output = {}
    for mode in MODES:
        raw = bundle.get('pairs', {}).get(mode)
        if not isinstance(raw, list):
            raise ValueError(f'bundle lacks {mode} pair rows')
        rows, draw_ids, pair_ids = ([], set(), set())
        for pair in raw:
            draw_id = pair.get('draw_id')
            if not isinstance(draw_id, str) or not draw_id or draw_id in draw_ids:
                raise ValueError(f'missing or duplicate draw_id in {mode}: {draw_id}')
            draw_ids.add(draw_id)
            left_id, right_id = (pair.get('left_id'), pair.get('right_id'))
            if left_id not in position or right_id not in position or left_id == right_id:
                raise ValueError(f'invalid DIMSIM occurrence IDs: {left_id}, {right_id}')
            left, right = (matrix[position[left_id]], matrix[position[right_id]])
            cosine = stable_cosine(left, right, left_id=left_id, right_id=right_id)
            if not math.isfinite(cosine):
                raise ValueError(f'non-finite cosine for IDs: {left_id}, {right_id}')
            distance = pair.get('dimsim_distance')
            status, label = distance_bin(distance)
            metric_status = pair.get('distance_status')
            valid = metric_status == 'ok' and status == 'ok'
            reason = 'ok' if valid else metric_status if metric_status != 'ok' else status
            row = {**pair, 'cosine_similarity': cosine, 'hidden_dissimilarity': 1.0 - cosine, 'analysis_status': reason, 'distance_bin': label}
            rows.append(row)
            pair_ids.add(pair['pair_id'])
        valid = [r for r in rows if r['analysis_status'] == 'ok']
        strata = {mask: contextual_dimsim_rho([r for r in valid if r.get('overlap_mask') == mask]) for mask in MASKS}
        strata['no_zero_initial'] = contextual_dimsim_rho([r for r in valid if r.get('no_zero_initial_diagnostic') is True])
        bins = []
        for label in BIN_LABELS:
            selected = [r for r in valid if r['distance_bin'] == label]
            unique = len({r['pair_id'] for r in selected})
            bins.append({'bin': label, 'draw_count': len(selected), 'unique_pair_count': unique, 'mean_cosine': float(np.mean([r['cosine_similarity'] for r in selected])) if selected else None, 'supported': unique >= bin_support})
        boundary = []
        n = len(valid)
        min_side = max(boundary_support, math.ceil(0.05 * n))
        for threshold in thresholds:
            left = [r for r in valid if float(r['dimsim_distance']) < threshold]
            right = [r for r in valid if float(r['dimsim_distance']) >= threshold]
            enough = len(left) >= min_side and len(right) >= min_side
            tail = contextual_dimsim_rho(right) if enough else {'rho': None, 'valid_pair_count': len(right), 'undefined_reason': 'insufficient_boundary_support'}
            boundary.append({'threshold': threshold, 'left_count': len(left), 'right_count': len(right), 'left_rho': contextual_dimsim_rho(left)['rho'] if enough else None, 'right_rho': tail['rho'], 'status': 'ok' if enough and tail['rho'] is not None else tail['undefined_reason'] or 'insufficient_boundary_support'})
        candidates = [x for x in boundary if x['right_rho'] is not None]
        selected = min(candidates, key=lambda x: (abs(x['right_rho']), x['threshold'])) if candidates else None
        output[mode] = {'rows': rows, 'summary': {'draw_count': len(rows), 'valid_draw_count': len(valid), 'invalid_metric_count': len(rows) - len(valid), 'unique_draw_count': len(draw_ids), 'unique_pair_count': len(pair_ids), 'valid_unique_pair_count': len({r['pair_id'] for r in valid}), 'rho': contextual_dimsim_rho(valid), 'strata': strata, 'bins': bins, 'boundary': {'min_side_count': min_side, 'curve': boundary, 'threshold': selected['threshold'] if selected else None, 'status': 'ok' if selected else 'no_valid_boundary_candidate' if n else 'no_valid_metrics'}}}
    return {'schema_version': 1, 'model_id': provenance.get('model_id'), 'seed': provenance.get('seed'), 'bundle_manifest_sha256': bundle_manifest_sha256, 'contract': {'modes': list(MODES), 'full_range_rho': True, 'bin_width': 10, 'min_unique_pairs_per_bin': bin_support, 'boundary_thresholds': thresholds, 'boundary_min_side_count': boundary_support, 'boundary_min_side_fraction': 0.05, 'boundary_rule': 'exploratory data-selected threshold minimizing absolute right-tail Spearman; ties choose smaller threshold', 'precision': 'full', 'bootstrap': False}, 'modes': output}

def score_contextual_dimsim_paths(bundle_path: str | Path, representation_path: str | Path, analysis_config: dict | None=None) -> dict:
    """Load verified bundle and representations, then call the pure scorer."""
    bundle = load_evaluation_bundle(bundle_path)
    bundle_sha = hashlib.sha256((Path(bundle_path) / 'manifest.json').read_bytes()).hexdigest()
    loaded = load_contextual_representations(representation_path)
    return score_contextual_dimsim(bundle, loaded['occurrence_ids'], loaded['hidden'], loaded['manifest']['provenance'], bundle_sha, analysis_config)
```

### DIMSIM across models and seeds


```python
"""Validated five-seed summaries and plot-ready data for contextual DIMSIM."""
import hashlib
import math
from pathlib import Path
from typing import Any
import numpy as np
dimsim_reporting_SEEDS = (0, 1, 2, 3, 4)
dimsim_reporting_METRIC_FIELDS = ('cosine_similarity', 'hidden_dissimilarity')

def dimsim_reporting_finite(value: Any) -> bool:
    return isinstance(value, (int, float)) and (not isinstance(value, bool)) and math.isfinite(float(value))

def dimsim_reporting_stats(rows: list[dict]) -> dict:
    return spearman_result([float(r['dimsim_distance']) for r in rows], [float(r['hidden_dissimilarity']) for r in rows])

def dimsim_reporting_assert_rho(actual, expected, location):
    if not isinstance(actual, dict) or actual.get('undefined_reason') != expected.get('undefined_reason') or actual.get('valid_pair_count') != expected.get('valid_pair_count'):
        raise ValueError(f'inconsistent rho metadata at {location}')
    value = actual.get('rho')
    if expected['rho'] is None:
        if value is not None:
            raise ValueError(f'unexpected defined rho at {location}')
    elif not dimsim_reporting_finite(value) or not math.isclose(float(value), expected['rho'], rel_tol=1e-12, abs_tol=1e-12):
        raise ValueError(f'rho does not match pair rows at {location}')

def dimsim_reporting_validate_score(score: dict, model: str, seed: int, bundle_hash: str):
    if type(score.get('schema_version')) is not int or score['schema_version'] != 1 or score.get('model_id') != model or (type(score.get('seed')) is not int) or (score['seed'] != seed) or (score.get('bundle_manifest_sha256') != bundle_hash):
        raise ValueError(f'score identity mismatch for {model} seed {seed}')
    contract = score.get('contract')
    required_contract = {'modes', 'full_range_rho', 'bin_width', 'min_unique_pairs_per_bin', 'boundary_thresholds', 'boundary_min_side_count', 'boundary_min_side_fraction', 'boundary_rule', 'precision', 'bootstrap'}
    if not isinstance(contract, dict) or set(contract) != required_contract or contract['modes'] != list(MODES) or (contract['full_range_rho'] is not True) or (type(contract['bin_width']) is not int) or (contract['bin_width'] != 10) or (type(contract['min_unique_pairs_per_bin']) is not int) or (contract['min_unique_pairs_per_bin'] <= 0) or (type(contract['boundary_min_side_count']) is not int) or (contract['boundary_min_side_count'] <= 0) or (contract['boundary_min_side_fraction'] != 0.05) or (contract['precision'] != 'full') or (contract['bootstrap'] is not False):
        raise ValueError(f'invalid analysis contract for {model} seed {seed}')
    thresholds = contract['boundary_thresholds']
    if not isinstance(thresholds, list) or not thresholds or any((type(x) is not int or not 10 <= x <= 110 for x in thresholds)) or (thresholds != sorted(set(thresholds))):
        raise ValueError('boundary thresholds must be unique and sorted')
    modes = score.get('modes')
    if not isinstance(modes, dict) or set(modes) != set(MODES):
        raise ValueError('score mode inventory is invalid')
    identities, mode_counts = ({}, {})
    for mode in MODES:
        value = modes[mode]
        if not isinstance(value, dict) or set(value) != {'rows', 'summary'} or (not isinstance(value['rows'], list)):
            raise ValueError(f'invalid mode rows for {model} seed {seed} {mode}')
        rows = value['rows']
        if not rows:
            raise ValueError('DIMSIM mode cannot have an empty draw table')
        seen_draw, pair_ids, valid, bins = (set(), set(), [], {label: [] for label in BIN_LABELS})
        shared_rows = []
        for index, row in enumerate(rows):
            if not isinstance(row, dict):
                raise ValueError('DIMSIM row must be an object')
            draw_id, pair_id = (row.get('draw_id'), row.get('pair_id'))
            if type(row.get('draw_index')) is not int or row['draw_index'] != index or draw_id != f'{mode}:v1:{index:06d}' or (draw_id in seen_draw) or (not isinstance(pair_id, str)) or (not pair_id):
                raise ValueError(f'invalid draw/pair identity in {model} seed {seed} {mode}')
            seen_draw.add(draw_id)
            pair_ids.add(pair_id)
            left, right, mask = (row.get('left_id'), row.get('right_id'), row.get('overlap_mask'))
            if not all((isinstance(x, str) and x for x in (left, right))) or left >= right or mask not in MASKS:
                raise ValueError('invalid endpoints or overlap mask')
            expected_mode = 'natural' if mode.endswith('natural') else 'balanced'
            if row.get('mode') != expected_mode or pair_id != pair_identity(left, right):
                raise ValueError('pair identity/mode metadata is invalid')
            status, distance = (row.get('distance_status'), row.get('dimsim_distance'))
            bin_status, label = distance_bin(distance)
            is_valid = status == 'ok' and bin_status == 'ok'
            expected_status = 'ok' if is_valid else status if status != 'ok' else bin_status
            if row.get('analysis_status') != expected_status or row.get('distance_bin') != label:
                raise ValueError('DIMSIM metric status/bin does not match source distance')
            if status == 'ok':
                if not dimsim_reporting_finite(distance) or float(distance) < 0:
                    raise ValueError('ok DIMSIM distance must be finite and nonnegative')
            elif distance is not None or not isinstance(status, str) or (not (status == 'not_computed' or status.startswith('invalid:'))):
                raise ValueError('invalid DIMSIM metric status/value')
            cosine, dissimilarity = (row.get(k) for k in dimsim_reporting_METRIC_FIELDS)
            if not dimsim_reporting_finite(cosine) or not dimsim_reporting_finite(dissimilarity) or (not math.isclose(float(dissimilarity), 1.0 - float(cosine), rel_tol=1e-12, abs_tol=1e-12)):
                raise ValueError('cosine/dissimilarity values are invalid or inconsistent')
            if row.get('no_zero_initial_diagnostic') is not True:
                raise ValueError('zero-initial diagnostic flag is invalid')
            common = (draw_id, pair_id, index, left, right, mask, status, distance)
            if expected_mode == 'balanced':
                first, second = (row.get('first_id'), row.get('second_id'))
                if {first, second} != {left, right} or first == second:
                    raise ValueError('balanced draw orientation does not match endpoints')
                common += (first, second, row.get('first_syllable_id'), row.get('second_syllable_id'))
            shared_rows.append(common)
            if is_valid:
                valid.append(row)
                bins[label].append(row)
        summary = value['summary']
        if not isinstance(summary, dict):
            raise ValueError('mode summary is required')
        expected_counts = {'draw_count': len(rows), 'valid_draw_count': len(valid), 'invalid_metric_count': len(rows) - len(valid), 'unique_draw_count': len(seen_draw), 'unique_pair_count': len(pair_ids), 'valid_unique_pair_count': len({r['pair_id'] for r in valid})}
        if any((type(summary.get(k)) is not int or summary[k] != v for k, v in expected_counts.items())):
            raise ValueError(f'mode count diagnostics mismatch for {model} seed {seed} {mode}')
        dimsim_reporting_assert_rho(summary.get('rho'), dimsim_reporting_stats(valid), f'{model}/{seed}/{mode}/full')
        strata = summary.get('strata')
        if not isinstance(strata, dict) or set(strata) != {*MASKS, 'no_zero_initial'}:
            raise ValueError('overlap strata inventory is invalid')
        for key in (*MASKS, 'no_zero_initial'):
            selected = [r for r in valid if (r['overlap_mask'] == key if key in MASKS else r.get('no_zero_initial_diagnostic') is True)]
            dimsim_reporting_assert_rho(strata[key], dimsim_reporting_stats(selected), f'{model}/{seed}/{mode}/{key}')
        bin_rows = summary.get('bins')
        if not isinstance(bin_rows, list) or len(bin_rows) != len(BIN_LABELS) or any((not isinstance(x, dict) for x in bin_rows)) or ([x.get('bin') for x in bin_rows] != list(BIN_LABELS)):
            raise ValueError('distance bins are not canonical')
        for label, actual in zip(BIN_LABELS, bin_rows):
            selected = bins[label]
            unique = len({r['pair_id'] for r in selected})
            mean = float(np.mean([r['cosine_similarity'] for r in selected])) if selected else None
            if type(actual.get('draw_count')) is not int or actual['draw_count'] != len(selected) or type(actual.get('unique_pair_count')) is not int or (actual['unique_pair_count'] != unique) or (actual.get('supported') is not (unique >= contract['min_unique_pairs_per_bin'])):
                raise ValueError('bin count/support diagnostics mismatch')
            av = actual.get('mean_cosine')
            if mean is None:
                if av is not None:
                    raise ValueError('empty bin mean must be null')
            elif not dimsim_reporting_finite(av) or not math.isclose(float(av), mean, rel_tol=1e-12, abs_tol=1e-12):
                raise ValueError('bin cosine mean does not match pair rows')
        boundary = summary.get('boundary')
        if not isinstance(boundary, dict) or type(boundary.get('min_side_count')) is not int:
            raise ValueError('boundary summary is invalid')
        minimum = max(contract['boundary_min_side_count'], math.ceil(0.05 * len(valid)))
        if boundary['min_side_count'] != minimum:
            raise ValueError('boundary side support differs from contract')
        curve = boundary.get('curve')
        if not isinstance(curve, list) or any((not isinstance(r, dict) for r in curve)) or [r.get('threshold') for r in curve] != thresholds:
            raise ValueError('boundary curve threshold grid differs from contract')
        computed = []
        for item in curve:
            t = item['threshold']
            lo = [r for r in valid if r['dimsim_distance'] < t]
            hi = [r for r in valid if r['dimsim_distance'] >= t]
            enough = len(lo) >= minimum and len(hi) >= minimum
            lr, rr = (dimsim_reporting_stats(lo)['rho'] if enough else None, dimsim_reporting_stats(hi)['rho'] if enough else None)
            status = 'ok' if enough and rr is not None else 'insufficient_boundary_support' if not enough else dimsim_reporting_stats(hi)['undefined_reason']
            if item.get('left_count') != len(lo) or item.get('right_count') != len(hi) or item.get('left_rho') != lr or (item.get('right_rho') != rr) or (item.get('status') != status):
                raise ValueError('boundary curve does not match original valid rows')
            if rr is not None:
                computed.append((abs(rr), t))
        chosen = min(computed)[1] if computed else None
        expected_boundary_status = 'ok' if chosen is not None else 'no_valid_boundary_candidate' if valid else 'no_valid_metrics'
        if boundary.get('threshold') != chosen or boundary.get('status') != expected_boundary_status:
            raise ValueError('selected boundary does not match boundary curve')
        identities[mode] = shared_rows
        mode_counts[mode] = {**expected_counts, 'valid_unique_pair_count': expected_counts['valid_unique_pair_count']}
    return (contract, identities, mode_counts)

def dimsim_reporting_summary(values: list[float | None], value_name: str='value', seeds=dimsim_reporting_SEEDS) -> dict:
    labelled = [{'seed': seed, value_name: value} for seed, value in zip(seeds, values)]
    if len(values) != 5 or any((value is None for value in values)):
        return {'seed_values': labelled, 'mean': None, 'sample_sd': None, 'status': 'incomplete_five_valid_values'}
    vals = np.asarray(values, dtype=np.float64)
    if not np.isfinite(vals).all():
        raise ValueError('summary values must be finite')
    return {'seed_values': labelled, 'mean': float(np.mean(vals)), 'sample_sd': float(np.std(vals, ddof=1)), 'status': 'ok'}

def build_dimsim_report(batch_result: dict, reference_model: str='syllable') -> dict:
    """Validate cross-seed pair and analysis identity, then build complete summaries."""
    if not isinstance(batch_result, dict) or type(batch_result.get('schema_version')) is not int or batch_result['schema_version'] != 1:
        raise ValueError('batch schema_version must be integer 1')
    bundle_hash = batch_result.get('bundle_manifest_sha256')
    if not isinstance(bundle_hash, str) or len(bundle_hash) != 64 or any((c not in '0123456789abcdef' for c in bundle_hash)):
        raise ValueError('invalid bundle manifest hash')
    reference = canonical_model_id(reference_model)
    records = batch_result.get('results')
    if not isinstance(records, list) or not records:
        raise ValueError('batch results are required')
    grouped, shared_contract, identity_reference, counts_reference = ({}, None, None, None)
    for rec in records:
        if not isinstance(rec, dict):
            raise ValueError('batch record must be an object')
        model, seed = (rec.get('model_id'), rec.get('seed'))
        if not isinstance(model, str) or canonical_model_id(model) != model or type(seed) is not int or (seed not in dimsim_reporting_SEEDS):
            raise ValueError('invalid model/seed identity')
        if (model, seed) in grouped:
            raise ValueError('duplicate model/seed result')
        score = rec.get('score')
        if not isinstance(score, dict):
            raise ValueError('batch score must be an object')
        contract, identities, mode_counts = dimsim_reporting_validate_score(score, model, seed, bundle_hash)
        if shared_contract is None:
            shared_contract, identity_reference, counts_reference = (contract, identities, mode_counts)
        elif contract != shared_contract or identities != identity_reference or mode_counts != counts_reference:
            raise ValueError('scores differ in analysis contract, pair/draw metadata, counts, or bin support')
        grouped[model, seed] = rec['score']
    models = sorted({m for m, _ in grouped})
    if reference not in models:
        raise ValueError('reference model is missing')
    tables = {'seed_rho': [], 'model_rho': [], 'seed_boundary': [], 'model_boundary': [], 'paired_rho_deltas': [], 'bins': [], 'mode_counts': []}
    plot = {'curves': [], 'boundaries': [], 'mode_counts': []}
    for mode in MODES:
        tables['mode_counts'].append({'mode': mode, **counts_reference[mode]})
        plot['mode_counts'].append({'mode': mode, **counts_reference[mode]})
    for model in models:
        for mode in MODES:
            all5 = [grouped.get((model, seed), {}).get('modes', {}).get(mode) for seed in dimsim_reporting_SEEDS]
            rho_vals = [x['summary']['rho']['rho'] if x else None for x in all5]
            for seed, item, rho in zip(dimsim_reporting_SEEDS, all5, rho_vals):
                s = item['summary'] if item else None
                tables['seed_rho'].append({'model_id': model, 'seed': seed, 'mode': mode, 'rho': rho, 'draw_count': s['draw_count'] if s else None, 'valid_draw_count': s['valid_draw_count'] if s else None, 'invalid_metric_count': s['invalid_metric_count'] if s else None, 'unique_pair_count': s['unique_pair_count'] if s else None, 'undefined_reason': s['rho']['undefined_reason'] if s else 'missing_seed', 'status': 'ok' if rho is not None else 'missing_seed' if s is None else 'undefined_rho'})
            tables['model_rho'].append({'model_id': model, 'mode': mode, **dimsim_reporting_summary(rho_vals, 'rho')})
            curves = []
            for i, label in enumerate(BIN_LABELS):
                bin_seed = [item['summary']['bins'][i] if item else None for item in all5]
                means = [x['mean_cosine'] if x else None for x in bin_seed]
                support = all((x is not None and x['supported'] for x in bin_seed))
                row = {'model_id': model, 'mode': mode, 'bin': label, 'mean_cosine': float(np.mean(means)) if all((x is not None for x in means)) else None, 'seed_points': [{'seed': seed, 'mean_cosine': means[j], 'draw_count': bin_seed[j]['draw_count'] if bin_seed[j] else None, 'unique_pair_count': bin_seed[j]['unique_pair_count'] if bin_seed[j] else None} for j, seed in enumerate(dimsim_reporting_SEEDS)], 'supported': support, 'connect_to_next': False}
                curves.append(row)
            for i in range(len(curves) - 1):
                curves[i]['connect_to_next'] = curves[i]['supported'] and curves[i + 1]['supported']
            tables['bins'].extend(curves)
            plot['curves'].extend(curves)
            thresholds = [item['summary']['boundary']['threshold'] if item else None for item in all5]
            for seed, item, threshold in zip(dimsim_reporting_SEEDS, all5, thresholds):
                tables['seed_boundary'].append({'model_id': model, 'seed': seed, 'mode': mode, 'threshold': threshold, 'status': item['summary']['boundary']['status'] if item else 'missing_seed'})
            bsummary = dimsim_reporting_summary(thresholds, 'threshold')
            median = float(np.median(thresholds)) if bsummary['status'] == 'ok' else None
            tables['model_boundary'].append({'model_id': model, 'mode': mode, **bsummary, 'median': median, 'median_status': 'ok' if median is not None else 'requires_all_five_valid'})
            plot['boundaries'].append({'model_id': model, 'mode': mode, 'median': median, 'seed_points': [{'seed': s, 'threshold': t} for s, t in zip(dimsim_reporting_SEEDS, thresholds)]})
    for model in models:
        if model == reference:
            continue
        for mode in MODES:
            deltas = []
            for seed in dimsim_reporting_SEEDS:
                a_score, b_score = (grouped.get((model, seed)), grouped.get((reference, seed)))
                a = a_score['modes'][mode]['summary']['rho']['rho'] if a_score else None
                b = b_score['modes'][mode]['summary']['rho']['rho'] if b_score else None
                deltas.append(a - b if a is not None and b is not None else None)
            tables['paired_rho_deltas'].append({'model_id': model, 'reference_model': reference, 'mode': mode, **dimsim_reporting_summary(deltas, 'delta_rho')})
    return {'schema_version': 1, 'bundle_manifest_sha256': bundle_hash, 'reference_model': reference, 'contract': {'analysis': shared_contract, 'seed_ids': list(dimsim_reporting_SEEDS), 'mean_and_sample_sd': True, 'summary_requires_five_defined_values': True, 'rho_delta': 'paired by seed vs reference', 'boundary_is_exploratory_data_selected': True, 'boundary_min_side_fraction': 0.05, 'bootstrap': False, 'rho_mcse': False, 'rendering': False}, 'tables': tables, 'plot_data': plot}

def score_contextual_dimsim_batch(bundle_path: str | Path, run_dirs: dict[str, str | Path], representation_dirs: dict[str, dict[int, str | Path]], analysis_config: dict | None=None) -> dict:
    """Score explicit model run directories and explicit model/seed representation paths."""
    if not isinstance(run_dirs, dict) or not run_dirs or (not isinstance(representation_dirs, dict)):
        raise ValueError('explicit run and representation mappings are required')
    runs, reps = ({}, {})
    for source, target in ((run_dirs, runs), (representation_dirs, reps)):
        for key, value in source.items():
            canonical = canonical_model_id(key)
            if canonical in target:
                raise ValueError('model aliases collide after canonicalization')
            target[canonical] = value
    if set(runs) != set(reps):
        raise ValueError('model inventories differ')
    bundle = load_evaluation_bundle(bundle_path)
    bundle_sha = hashlib.sha256((Path(bundle_path) / 'manifest.json').read_bytes()).hexdigest()
    source_hashes = bundle['manifest'].get('source_experiment', {}).get('hashes')
    scored = []
    for model in sorted(runs):
        run = discover_seed_runs(runs[model], model)
        seed_paths = reps[model]
        if not isinstance(seed_paths, dict) or any((type(s) is not int for s in seed_paths)) or set(seed_paths) != set(dimsim_reporting_SEEDS):
            raise ValueError(f'representations must map seeds 0 through 4 for {model}')
        for record in run['seeds']:
            seed = record['seed']
            loaded = load_contextual_representations(seed_paths[seed])
            p = loaded['manifest']['provenance']
            if p['model_id'] != model or p['seed'] != seed or p['checkpoint_sha256'] != record['checkpoint_sha256'] or (p['experiment_hashes'] != run['experiment_hashes']):
                raise ValueError(f'representation/run provenance mismatch for {model} seed {seed}')
            if any((run['experiment_hashes'].get(k) != v for k, v in source_hashes.items())):
                raise ValueError('bundle and training run source hashes differ')
            score = score_contextual_dimsim(bundle, loaded['occurrence_ids'], loaded['hidden'], p, bundle_sha, analysis_config)
            scored.append({'model_id': model, 'seed': seed, 'run_id': run['run_id'], 'run_manifest_sha256': run['manifest_sha256'], 'checkpoint_sha256': record['checkpoint_sha256'], 'score': score})
    return {'schema_version': 1, 'bundle_manifest_sha256': bundle_sha, 'results': scored}
```

```python
"""Matched iid Gaussian occurrence controls for contextual evaluation."""
import hashlib
import math
from pathlib import Path
from typing import Any
import numpy as np
RANDOM_SEEDS = (0, 1, 2, 3, 4)

def random_control_sha(path: Path) -> str:
    return hashlib.sha256(path.read_bytes()).hexdigest()

def resolve_contextual_dimension(model_runs: dict[str, dict[int, str | Path]], *, bundle_manifest_sha256: str, bundle: dict[str, Any]) -> dict[str, Any]:
    """Validate all learned representation runs and pin the shared h dimension."""
    if not isinstance(model_runs, dict) or not model_runs:
        raise ValueError('explicit learned model/seed representation paths are required')
    dimensions, inventory, pins = (set(), None, {})
    expected_inventory = tuple(sorted((row['occurrence_id'] for row in bundle['occurrences'])))
    if bundle.get('manifest', {}).get('schema_version') != 3 or not random_control_is_sha256(bundle_manifest_sha256):
        raise ValueError('dimension resolution requires a verified schema-3 bundle and manifest hash')
    for model_id, seed_runs in sorted(model_runs.items()):
        try:
            if canonical_model_id(model_id) != model_id:
                raise ValueError
        except (TypeError, ValueError) as error:
            raise ValueError(f'learned model ID is not canonical: {model_id}') from error
        if not isinstance(seed_runs, dict) or any((type(seed) is not int for seed in seed_runs)) or set(seed_runs) != set(RANDOM_SEEDS):
            raise ValueError(f'{model_id}: exactly contextual seeds 0..4 are required')
        pins[model_id] = {}
        for seed in RANDOM_SEEDS:
            loaded = load_contextual_representations(seed_runs[seed])
            manifest = loaded['manifest']
            provenance = manifest['provenance']
            if provenance['model_id'] != model_id or provenance['seed'] != seed:
                raise ValueError(f'learned representation identity mismatch for {model_id}/{seed}')
            if provenance['bundle_manifest_sha256'] != bundle_manifest_sha256:
                raise ValueError('learned representations do not share the requested bundle')
            ids = tuple(loaded['occurrence_ids'])
            if inventory is None:
                inventory = ids
            elif ids != inventory:
                raise ValueError('learned representations have mismatched occurrence inventories')
            dimensions.add(int(manifest['dimension']))
            pins[model_id][str(seed)] = {'path': str(Path(seed_runs[seed]).resolve()), 'manifest_sha256': random_control_sha(Path(seed_runs[seed]) / 'manifest.json'), 'checkpoint_sha256': provenance['checkpoint_sha256'], 'dimension': int(manifest['dimension']), 'occurrence_ids': list(ids)}
    if inventory != expected_inventory:
        raise ValueError('learned representations do not match the requested bundle occurrence inventory')
    if len(dimensions) != 1:
        raise ValueError('learned contextual dimensions differ across models or seeds')
    return {'schema_version': 1, 'dimension': dimensions.pop(), 'models': pins, 'bundle_manifest_sha256': bundle_manifest_sha256, 'occurrence_ids': list(expected_inventory)}

def generate_random_control(occurrence_ids: list[str], *, dimension: int, random_seeds: tuple[int, ...] | list[int]=RANDOM_SEEDS) -> dict[str, Any]:
    """Generate five deterministic standard-normal matrices; retain arrays outside reports."""
    if type(dimension) is not int or dimension <= 0:
        raise ValueError('dimension must be a strict positive integer')
    if type(random_seeds) not in (tuple, list) or any((type(seed) is not int for seed in random_seeds)) or tuple(random_seeds) != RANDOM_SEEDS:
        raise ValueError('random seeds are fixed to integer seeds 0..4')
    if not isinstance(occurrence_ids, list) or not occurrence_ids or any((not isinstance(x, str) or not x for x in occurrence_ids)) or (occurrence_ids != sorted(occurrence_ids)) or (len(set(occurrence_ids)) != len(occurrence_ids)):
        raise ValueError('occurrence IDs must be nonempty, unique, and sorted')
    arrays = {}
    for seed in RANDOM_SEEDS:
        rng = np.random.Generator(np.random.PCG64(seed))
        arrays[seed] = validate_matrix(occurrence_ids, rng.standard_normal((len(occurrence_ids), dimension)).astype(np.float32))
    return {'occurrence_ids': list(occurrence_ids), 'matrices': arrays}

def evaluate_random_control(bundle: dict[str, Any], generated: dict[str, Any], *, bundle_manifest_sha256: str, dimension_resolution: dict[str, Any], source_sha256: str, analysis_config: dict[str, Any] | None=None) -> dict[str, Any]:
    """Score the same generated vectors on Q and DIMSIM, returning report and raw arrays separately."""
    ids, matrices = (generated['occurrence_ids'], generated['matrices'])
    if not random_control_is_sha256(source_sha256):
        raise ValueError('explicit build source SHA-256 is required')
    random_control_validate_dimension_resolution(dimension_resolution, bundle, bundle_manifest_sha256, ids)
    if not isinstance(matrices, dict) or any((type(seed) is not int for seed in matrices)) or set(matrices) != set(RANDOM_SEEDS):
        raise ValueError('generated control must contain exactly random seeds 0..4')
    if bundle['manifest'].get('schema_version') != 3:
        raise ValueError('random control requires a verified schema-3 bundle')
    records = []
    for random_seed in RANDOM_SEEDS:
        matrix = np.asarray(matrices[random_seed])
        if matrix.ndim != 2 or matrix.shape != (len(ids), dimension_resolution['dimension']) or matrix.dtype != np.dtype('float32') or (not np.isfinite(matrix).all()) or np.any(np.max(np.abs(matrix.astype(np.float64)), axis=1) == 0):
            raise ValueError(f'random matrix {random_seed} does not match the resolved dimension or finite float32 contract')
        provenance = {'model_id': 'random-control', 'seed': random_seed, 'seed_kind': 'random', 'adapter': 'iid_standard_normal', 'rng': 'PCG64', 'rng_distribution': 'standard_normal', 'numpy_version': np.__version__, 'source_sha256': source_sha256, 'dimension': int(matrix.shape[1]), 'dtype': str(matrix.dtype), 'bundle_manifest_sha256': bundle_manifest_sha256, 'dimension_resolution': dimension_resolution}
        q = anchor_q_score_anchor_q(bundle, ids, matrix, provenance, bundle_manifest_sha256=bundle_manifest_sha256, tie_tolerance=bundle['config']['scoring']['anchor_q']['tie_tolerance'])
        d = score_contextual_dimsim(bundle, ids, matrix, provenance, bundle_manifest_sha256, analysis_config)
        records.append({'random_seed': random_seed, 'seed_kind': 'random', 'provenance': provenance, 'anchor_q': q, 'dimsim': d})
    return {'report': summarize_random_control(records, bundle_manifest_sha256), 'matrices': matrices, 'occurrence_ids': ids}

def random_control_mean_sd(values: list[float]) -> dict[str, Any]:
    vals = np.asarray(values, dtype=np.float64)
    return {'mean': float(vals.mean()), 'sample_sd': float(vals.std(ddof=1)), 'count': int(len(vals))}

def summarize_random_control(records: list[dict[str, Any]], bundle_manifest_sha256: str) -> dict[str, Any]:
    if not random_control_is_sha256(bundle_manifest_sha256) or type(records) is not list or len(records) != 5:
        raise ValueError('random control summary requires five records and a valid bundle hash')
    random_control_validate_summary_records(records, bundle_manifest_sha256)
    edges = [f'{component}:{low}->{high}' for component, low, high in EDGES]
    q_summary = {'edges': {}, 'components': {}, 'overall': {}}
    per_seed_edge_means: dict[str, list[float]] = {}
    for edge in edges:
        seed_values = random_control_seed_edge_means(records, edge)
        per_seed_edge_means[edge] = seed_values
        vals = random_control_matched_q_values(records, edge)
        q_summary['edges'][edge] = {**random_control_mean_sd(seed_values), **random_control_mcse(vals)}
    for component in ('initial', 'final', 'tone'):
        component_edges = [e for e in edges if e.split(':', 1)[0] == component]
        edge_mc = [q_summary['edges'][e]['mc_se_seed_mean'] for e in component_edges]
        seed_values = [float(np.mean([per_seed_edge_means[e][i] for e in component_edges])) for i in range(5)]
        mcse = random_control_combine_edge_mcse(edge_mc)
        q_summary['components'][component] = {**random_control_mean_sd(seed_values), 'mc_se_seed_mean': mcse, 'mc_se_status': 'ok' if mcse is not None else 'not_available_single_draw'}
    edge_mc = [q_summary['edges'][e]['mc_se_seed_mean'] for e in edges]
    seed_values = [float(np.mean([per_seed_edge_means[e][i] for e in edges])) for i in range(5)]
    mcse = random_control_combine_edge_mcse(edge_mc)
    q_summary['overall'] = {**random_control_mean_sd(seed_values), 'mc_se_seed_mean': mcse, 'mc_se_status': 'ok' if mcse is not None else 'not_available_single_draw'}
    dim_summary = {}
    for mode in records[0]['dimsim']['modes']:
        rhos = [random_control_validated_rho(r['dimsim']['modes'][mode]['summary']['rho']) for r in records]
        defined = [x for x in rhos if x is not None]
        complete = len(defined) == 5
        dim_summary[mode] = {'rho_mean': float(np.mean(defined)) if complete else None, 'rho_sample_sd': float(np.std(defined, ddof=1)) if complete else None, 'defined_seed_count': len(defined), 'status': 'ok' if complete else 'requires_all_five_defined', 'random_seeds': list(RANDOM_SEEDS), 'per_seed_rho': [{'random_seed': r['random_seed'], 'rho': r['dimsim']['modes'][mode]['summary']['rho']['rho'], 'undefined_reason': r['dimsim']['modes'][mode]['summary']['rho'].get('undefined_reason')} for r in records]}
    return {'schema_version': 1, 'model_id': 'random-control', 'adapter': 'iid_standard_normal', 'seed_kind': 'random', 'random_seeds': list(RANDOM_SEEDS), 'bundle_manifest_sha256': bundle_manifest_sha256, 'interpretation': 'Variation across these controls reflects random-vector variation, distinct from training-seed variation.', 'anchor_q': q_summary, 'dimsim': dim_summary, 'per_seed': records}

def random_control_is_sha256(value: Any) -> bool:
    return isinstance(value, str) and len(value) == 64 and all((c in '0123456789abcdef' for c in value))

def random_control_validate_dimension_resolution(resolution: Any, bundle: dict[str, Any], bundle_hash: str, occurrence_ids: list[str]) -> None:
    if not isinstance(resolution, dict) or resolution.get('schema_version') != 1 or type(resolution.get('dimension')) is not int or (resolution['dimension'] <= 0) or (resolution.get('bundle_manifest_sha256') != bundle_hash) or (resolution.get('occurrence_ids') != occurrence_ids):
        raise ValueError('validated dimension resolution does not match bundle and occurrence inventory')
    models = resolution.get('models')
    if not isinstance(models, dict) or not models:
        raise ValueError('dimension resolution lacks learned model pins')
    for model, seeds in models.items():
        try:
            if canonical_model_id(model) != model:
                raise ValueError
        except (TypeError, ValueError) as error:
            raise ValueError('dimension resolution contains noncanonical model identity') from error
        if not isinstance(seeds, dict) or set(seeds) != {str(x) for x in RANDOM_SEEDS}:
            raise ValueError('dimension resolution must pin exact learned seeds 0..4')
        for seed, pin in seeds.items():
            if not isinstance(pin, dict) or not random_control_is_sha256(pin.get('manifest_sha256')) or (not random_control_is_sha256(pin.get('checkpoint_sha256'))) or (pin.get('dimension') != resolution['dimension']) or (pin.get('occurrence_ids') != occurrence_ids):
                raise ValueError(f'invalid learned dimension pin {model}/{seed}')

def random_control_validate_summary_records(records: list[dict[str, Any]], bundle_hash: str) -> None:
    reference_provenance = None
    for expected, record in enumerate(records):
        if not isinstance(record, dict) or type(record.get('random_seed')) is not int or record['random_seed'] != expected or (record.get('seed_kind') != 'random'):
            raise ValueError('random control summary requires exact ordered integer seeds 0..4')
        q, d = (record.get('anchor_q'), record.get('dimsim'))
        if not isinstance(q, dict) or not isinstance(d, dict) or q.get('bundle_manifest_sha256') != bundle_hash or (d.get('bundle_manifest_sha256') != bundle_hash) or (q.get('model_id') != 'random-control') or (q.get('seed') != expected) or (d.get('model_id') != 'random-control') or (d.get('seed') != expected):
            raise ValueError('random control score identity or bundle hash mismatch')
        provenance = record.get('provenance')
        required = {'adapter': 'iid_standard_normal', 'rng': 'PCG64', 'rng_distribution': 'standard_normal', 'seed_kind': 'random', 'model_id': 'random-control', 'seed': expected, 'bundle_manifest_sha256': bundle_hash}
        if not isinstance(provenance, dict) or any((provenance.get(key) != value for key, value in required.items())) or (not random_control_is_sha256(provenance.get('source_sha256'))):
            raise ValueError('random control provenance or RNG contract is invalid')
        invariant = {key: value for key, value in provenance.items() if key != 'seed'}
        if reference_provenance is None:
            reference_provenance = invariant
        elif invariant != reference_provenance:
            raise ValueError('random control provenance differs across seeds')
    base_q = records[0]['anchor_q']['draw_scores']
    edges = [f'{component}:{low}->{high}' for component, low, high in EDGES]
    if not isinstance(records[0]['anchor_q'].get('edges'), dict) or set(records[0]['anchor_q']['edges']) != set(edges):
        raise ValueError('random control Q edge inventory/order mismatch')
    base_triplets = [(r['draw_id'], r['triplet_id'], r['edge_id']) for r in base_q]
    random_control_validate_q_draws(base_q, edges)
    base_dims = records[0]['dimsim']
    base_dim_rows = {}
    for mode, result in base_dims['modes'].items():
        base_dim_rows[mode] = random_control_dimsim_contract_rows(result['rows'])
    for record in records[1:]:
        q, d = (record['anchor_q'], record['dimsim'])
        if not isinstance(q.get('edges'), dict) or set(q['edges']) != set(edges):
            raise ValueError('random control Q edge inventory/order mismatch')
        if [(r['draw_id'], r['triplet_id'], r['edge_id']) for r in q['draw_scores']] != base_triplets:
            raise ValueError('random control Q triplet identity mismatch')
        random_control_validate_q_draws(q['draw_scores'], edges)
        if q['tie_tolerance'] != records[0]['anchor_q']['tie_tolerance']:
            raise ValueError('random control Q scoring contract mismatch')
        for mode in base_dims['modes']:
            left, right = (base_dims['modes'][mode], d['modes'].get(mode))
            if right is None or left['summary']['draw_count'] != right['summary']['draw_count'] or [(x['draw_id'], x['pair_id'], x['left_id'], x['right_id']) for x in left['rows']] != [(x['draw_id'], x['pair_id'], x['left_id'], x['right_id']) for x in right['rows']] or (random_control_dimsim_contract_rows(right['rows']) != base_dim_rows[mode]) or (d['contract'] != base_dims['contract']):
                raise ValueError('random control DIMSIM pair identity or analysis contract mismatch')

def random_control_validate_q_draws(rows: list[dict[str, Any]], edges: list[str]) -> None:
    if not isinstance(rows, list) or not rows or any((not isinstance(row, dict) for row in rows)):
        raise ValueError('random control Q draw table is invalid')
    edge_count = len(rows) // len(edges)
    if edge_count <= 0 or len(rows) != edge_count * len(edges):
        raise ValueError('random control Q draw coverage is incomplete')
    for index, row in enumerate(rows):
        edge_index, draw_index = divmod(index, edge_count)
        score = row.get('score')
        if row.get('edge_id') != edges[edge_index] or type(score) not in (int, float) or (not math.isfinite(score)) or (score not in (0, 0.5, 1)) or (not isinstance(row.get('draw_id'), str)) or (not row['draw_id']) or (not isinstance(row.get('triplet_id'), str)) or (not row['triplet_id']):
            raise ValueError('random control Q draw values or canonical edge order are invalid')
        if row['draw_id'] != f'anchor-q:v1:{edges[edge_index]}:{draw_index:06d}':
            raise ValueError('random control Q draw ID/index order is invalid')

def random_control_matched_q_values(records: list[dict[str, Any]], edge: str) -> list[float]:
    by_seed = [{row['draw_id']: float(row['score']) for row in record['anchor_q']['draw_scores'] if row['edge_id'] == edge} for record in records]
    if any((set(values) != set(by_seed[0]) for values in by_seed[1:])):
        raise ValueError('matched Q draw identities differ across random seeds')
    ids = list(by_seed[0])
    return [float(np.mean([values[draw_id] for values in by_seed])) for draw_id in ids]

def random_control_seed_edge_means(records: list[dict[str, Any]], edge: str) -> list[float]:
    result = []
    for record in records:
        scores = [float(row['score']) for row in record['anchor_q']['draw_scores'] if row['edge_id'] == edge]
        if not scores:
            raise ValueError(f'random control Q edge has no draws: {edge}')
        result.append(float(np.mean(scores)))
    return result

def random_control_validated_rho(value: Any) -> float | None:
    if not isinstance(value, dict):
        raise ValueError('random control DIMSIM rho record is invalid')
    rho = value.get('rho')
    if rho is None:
        reason = value.get('undefined_reason')
        if not isinstance(reason, str) or not reason:
            raise ValueError('undefined random control rho requires a reason')
        return None
    if type(rho) not in (int, float) or not math.isfinite(rho) or (not -1 <= rho <= 1):
        raise ValueError('random control DIMSIM rho must be finite and within [-1,1]')
    return float(rho)

def random_control_dimsim_contract_rows(rows: list[dict[str, Any]]) -> list[tuple[Any, ...]]:
    if not isinstance(rows, list):
        raise ValueError('DIMSIM rows must be a list')
    result = []
    for row in rows:
        distance = row.get('dimsim_distance')
        if distance is not None and (type(distance) not in (int, float) or not math.isfinite(distance)):
            raise ValueError('DIMSIM distance metadata must be finite or null')
        for key in ('draw_id', 'pair_id', 'left_id', 'right_id', 'analysis_status', 'distance_bin'):
            if not isinstance(row.get(key), str):
                raise ValueError(f'DIMSIM row lacks valid {key}')
        if row.get('distance_status') != 'ok' and distance is not None:
            raise ValueError('invalid DIMSIM metric must not carry a numeric distance')
        result.append((row['draw_id'], row['pair_id'], row['left_id'], row['right_id'], distance, row.get('distance_status'), row.get('overlap_mask'), row.get('no_zero_initial_diagnostic'), row['analysis_status'], row['distance_bin']))
    return result

def random_control_mcse(values: list[float]) -> dict[str, Any]:
    if len(values) <= 1:
        return {'mc_se_seed_mean': None, 'mc_se_status': 'not_available_single_draw', 'draw_count': len(values)}
    return {'mc_se_seed_mean': float(np.std(np.asarray(values, dtype=np.float64), ddof=1) / math.sqrt(len(values))), 'mc_se_status': 'ok', 'draw_count': len(values)}

def random_control_combine_edge_mcse(values: list[float | None]) -> float | None:
    if not values or any((value is None for value in values)):
        return None
    return float(math.sqrt(sum((float(value) ** 2 for value in values))) / len(values))
```

### Report structure


```python
"""Pure coordinator for the independent contextual Anchor-Q and DIMSIM reports."""
import json
import re
from typing import Any
report_contract_SEEDS = (0, 1, 2, 3, 4)
report_contract_SHA = re.compile('^[0-9a-f]{64}$')
report_contract_EXPERIMENT_HASH_NAMES = {'manifest.json', 'config.json', 'vocab.json', 'samples.jsonl', 'splits.json', 'acoustic_targets.npy'}
report_contract_FIGURES = (('Anchor-Q transitions', 'contextual-anchor-q', {'mean_and_seed_points': 'plot_data.edges'}), ('Transition shares', 'contextual-anchor-q', {'shares': 'plot_data.transition_shares'}), ('Component shares', 'contextual-anchor-q', {'shares': 'plot_data.component_shares'}), ('DIMSIM geometry', 'contextual-dimsim', {'curves': 'plot_data.curves', 'boundaries': 'plot_data.boundaries', 'counts': 'plot_data.mode_counts'}), ('DIMSIM correlation', 'contextual-dimsim', {'model_rho': 'tables.model_rho', 'seed_rho': 'tables.seed_rho'}))

def report_contract_batch_identity(batch: dict[str, Any], label: str) -> tuple[str, tuple[str, ...], dict[tuple[str, int], dict]]:
    if not isinstance(batch, dict) or type(batch.get('schema_version')) is not int or batch['schema_version'] != 1:
        raise ValueError(f'{label} batch schema_version must be integer 1')
    bundle_hash = batch.get('bundle_manifest_sha256')
    if not isinstance(bundle_hash, str) or not report_contract_SHA.fullmatch(bundle_hash):
        raise ValueError(f'{label} batch bundle hash is invalid')
    records = batch.get('results')
    if not isinstance(records, list) or not records:
        raise ValueError(f'{label} batch results are required')
    grouped = {}
    for row in records:
        if not isinstance(row, dict):
            raise ValueError(f'{label} record must be an object')
        model, seed = (row.get('model_id'), row.get('seed'))
        try:
            canonical = canonical_model_id(model)
        except (TypeError, ValueError) as exc:
            raise ValueError(f'{label} model identity is invalid') from exc
        if model != canonical or type(seed) is not int or seed not in report_contract_SEEDS:
            raise ValueError(f'{label} model/seed identity is invalid')
        key = (model, seed)
        if key in grouped:
            raise ValueError(f'{label} has duplicate model/seed')
        for field in ('run_id', 'run_manifest_sha256', 'checkpoint_sha256'):
            if not isinstance(row.get(field), str) or not row[field].strip():
                raise ValueError(f'{label} {field} is required for {model} seed {seed}')
        for field in ('run_manifest_sha256', 'checkpoint_sha256'):
            if not report_contract_SHA.fullmatch(row[field]):
                raise ValueError(f'{label} {field} is invalid')
        hashes = row.get('experiment_hashes')
        if hashes is not None and (not isinstance(hashes, dict) or set(hashes) != report_contract_EXPERIMENT_HASH_NAMES or any((not isinstance(v, str) or not report_contract_SHA.fullmatch(v) for v in hashes.values()))):
            raise ValueError(f'{label} experiment_hashes inventory is invalid')
        grouped[key] = row
    models = tuple(sorted({model for model, _ in grouped}))
    expected = {(model, seed) for model in models for seed in report_contract_SEEDS}
    if set(grouped) != expected:
        raise ValueError(f'{label} must contain all five seeds for every model')
    for model in models:
        model_rows = [grouped[model, seed] for seed in report_contract_SEEDS]
        for field in ('run_id', 'run_manifest_sha256', 'experiment_hashes'):
            available = [row.get(field) for row in model_rows if row.get(field) is not None]
            if available and any((value != available[0] for value in available[1:])):
                raise ValueError(f'{label} {field} differs across seeds for {model}')
    return (bundle_hash, models, grouped)

def report_contract_semantic(report: dict, path: str) -> Any:
    value: Any = report
    for key in path.split('.'):
        if not isinstance(value, dict) or key not in value:
            raise ValueError(f'figure data mapping does not resolve: {path}')
        value = value[key]
    return value

def report_contract_json_finite(value: Any, label: str) -> None:
    try:
        json.dumps(value, allow_nan=False, sort_keys=True)
    except (TypeError, ValueError) as exc:
        raise ValueError(f'{label} must be finite JSON-compatible data') from exc

def build_evaluation_report(q_batch: dict, dimsim_batch: dict, reference_model: str='syllable', run_ids: dict[str, str] | None=None) -> dict:
    """Build a validated JSON contract and artifact plan without writing files."""
    q_bundle, q_models, q_rows = report_contract_batch_identity(q_batch, 'Anchor-Q')
    d_bundle, d_models, d_rows = report_contract_batch_identity(dimsim_batch, 'DIMSIM')
    if q_bundle != d_bundle:
        raise ValueError('Anchor-Q and DIMSIM bundle hashes differ')
    if q_models != d_models:
        raise ValueError('Anchor-Q and DIMSIM canonical model inventories differ')
    try:
        reference = canonical_model_id(reference_model)
    except (TypeError, ValueError) as exc:
        raise ValueError('reference model is invalid') from exc
    if reference != reference_model:
        raise ValueError('reference model must be canonical')
    if run_ids is not None and (not isinstance(run_ids, dict) or set(run_ids) != {'contextual-anchor-q', 'contextual-dimsim'}):
        raise ValueError('run_ids must map both analysis names to explicit safe IDs')
    pins = []
    for model in q_models:
        for seed in report_contract_SEEDS:
            q, d = (q_rows[model, seed], d_rows[model, seed])
            for field in ('run_id', 'run_manifest_sha256', 'checkpoint_sha256'):
                if q[field] != d[field]:
                    raise ValueError(f'Anchor-Q/DIMSIM {field} mismatch for {model} seed {seed}')
            qh, dh = (q.get('experiment_hashes'), d.get('experiment_hashes'))
            if qh is not None and dh is not None and (qh != dh):
                raise ValueError(f'Anchor-Q/DIMSIM experiment_hashes mismatch for {model} seed {seed}')
            q_prov = q['score'].get('provenance')
            if not isinstance(q_prov, dict):
                raise ValueError(f'Anchor-Q score provenance missing for {model} seed {seed}')
            for field, expected in (('model_id', model), ('seed', seed), ('bundle_manifest_sha256', q_bundle), ('checkpoint_sha256', q['checkpoint_sha256'])):
                if q_prov.get(field) != expected:
                    raise ValueError(f'Anchor-Q outer/nested {field} mismatch for {model} seed {seed}')
            nested_hashes = q_prov.get('experiment_hashes')
            if not isinstance(nested_hashes, dict) or set(nested_hashes) != report_contract_EXPERIMENT_HASH_NAMES or any((not isinstance(value, str) or not report_contract_SHA.fullmatch(value) for value in nested_hashes.values())):
                raise ValueError(f'Anchor-Q nested experiment_hashes inventory is invalid for {model} seed {seed}')
            hashes = qh if qh is not None else dh
            if hashes is not None and nested_hashes != hashes:
                raise ValueError(f'Anchor-Q outer/nested experiment_hashes mismatch for {model} seed {seed}')
            hashes = hashes if hashes is not None else nested_hashes
            pin = {'model_id': model, 'seed': seed, 'run_id': q['run_id'], 'run_manifest_sha256': q['run_manifest_sha256'], 'checkpoint_sha256': q['checkpoint_sha256'], 'bundle_manifest_sha256': q_bundle, 'experiment_hashes': dict(sorted(hashes.items()))}
            pins.append(pin)
    for model in q_models:
        model_pins = [pin for pin in pins if pin['model_id'] == model]
        if any((pin['experiment_hashes'] != model_pins[0]['experiment_hashes'] for pin in model_pins[1:])):
            raise ValueError(f'resolved experiment_hashes differ across seeds for {model}')
    q_report = build_anchor_q_report(q_batch, reference_model=reference)
    d_report = build_dimsim_report(dimsim_batch, reference_model=reference)
    for name, report in (('Anchor-Q report', q_report), ('DIMSIM report', d_report)):
        report_contract_json_finite(report, name)
    report_contract_json_finite(q_batch, 'Anchor-Q raw batch')
    report_contract_json_finite(dimsim_batch, 'DIMSIM raw batch')
    figures = []
    for name, analysis, sources in report_contract_FIGURES:
        report = q_report if analysis == 'contextual-anchor-q' else d_report
        data = {semantic_name: report_contract_semantic(report, key) for semantic_name, key in sources.items()}
        geometry = name == 'DIMSIM geometry'
        figures.append({'name': name, 'analysis': analysis, 'data_keys': dict(sources), 'data': data, 'axis_right_counts': geometry, 'ticks_labels_every_panel_both_columns': geometry, 'export_format': 'unresolved', 'config_metadata': {'source_paths': dict(sources), 'analysis': analysis}})
    artifact_plan = []
    for analysis in ('contextual-anchor-q', 'contextual-dimsim'):
        report_name = 'anchor_q_report.json' if analysis.endswith('anchor-q') else 'dimsim_report.json'
        batch_name = 'anchor_q_batch.json' if analysis.endswith('anchor-q') else 'dimsim_batch.json'
        run_id = safe_run_id(run_ids[analysis]) if run_ids is not None else None
        if run_id is None:
            filenames, directories = (None, None)
        else:
            base = f'outputs/{analysis}/{run_id}'
            filenames = {'config': f'{base}/config.json', 'manifest': f'{base}/manifest.json', 'input_batch': f'{base}/data/{batch_name}', 'report': f'{base}/data/{report_name}', 'tables': f'{base}/tables/tables.json', 'plot_data': f'{base}/figures/plot_data.json'}
            directories = [f'{base}/data', f'{base}/tables', f'{base}/figures', f'{base}/diagnostics']
        artifact_plan.append({'analysis': analysis, 'run_id': run_id, 'path_template': f'outputs/{analysis}/{{run_id}}', 'directories': directories, 'filenames': filenames})
    result = {'contract_version': '1.0', 'reference_model': reference, 'scientific_input_pins': {'bundle_manifest_sha256': q_bundle, 'model_ids': list(q_models), 'seeds': list(report_contract_SEEDS), 'model_seed_provenance': pins}, 'validation_limits': {'dimsim_nested_provenance': 'DIMSIM score schema carries model, seed and bundle identity but not checkpoint or experiment hashes; those pins are checked across the outer batches and retained explicitly.'}, 'raw_batches': {'contextual-anchor-q': q_batch, 'contextual-dimsim': dimsim_batch}, 'reports': {'contextual-anchor-q': q_report, 'contextual-dimsim': d_report}, 'figures': figures, 'artifact_plan': artifact_plan, 'readiness': {'numeric_contract_valid': True, 'five_seed_complete': True, 'output_writer': 'pending', 'notebook_wiring': 'pending', 'extraction_wiring': 'pending', 'random_baseline_integration': 'pending', 'model_runtime': 'pending', 'figure_exports': 'pending', 'end_to_end_ready': False}}
    report_contract_json_finite(result, 'evaluation report contract')
    return result
```

### Output paths and manifests


```python
"""Reusable atomic research-run artifact management."""
from contextlib import contextmanager
from hashlib import sha256
import json, os, platform, re, subprocess
from importlib.metadata import PackageNotFoundError, version
from pathlib import Path

def file_sha256(path, chunk=1024 * 1024):
    h = sha256()
    with Path(path).open('rb') as f:
        while (data := f.read(chunk)):
            h.update(data)
    return h.hexdigest()

def safe_run_id(value):
    if not isinstance(value, str) or not re.fullmatch('[A-Za-z0-9][A-Za-z0-9._-]{0,127}', value) or value in {'.', '..'}:
        raise ValueError('run_id must be a non-empty safe filename component')
    return value

def safe_analysis(value):
    if not isinstance(value, str) or not re.fullmatch('[a-z0-9][a-z0-9-]{0,63}', value):
        raise ValueError('analysis must be a safe lowercase name')
    return value

def ensure_output_outside_inputs(output, input_roots):
    output = Path(output).resolve()
    for root in map(lambda p: Path(p).resolve(), input_roots):
        if output == root or root in output.parents:
            raise ValueError(f'Output must not be inside input: {root}')

def write_json_new(path, value):
    with Path(path).open('x', encoding='utf-8') as f:
        json.dump(value, f, ensure_ascii=False, indent=2, sort_keys=True, allow_nan=False)

def git_sha(root):
    try:
        return subprocess.run(['git', 'rev-parse', 'HEAD'], cwd=root, text=True, capture_output=True, check=True).stdout.strip()
    except Exception:
        return None

def output_artifacts_expected_paths(values):
    if not values:
        raise ValueError('expected_files must be a non-empty collection')
    result = set()
    for value in values:
        path = Path(value)
        if path.is_absolute() or '..' in path.parts or path.as_posix() in {'.', ''}:
            raise ValueError(f'Unsafe expected artifact path: {value}')
        result.add(path.as_posix())
    if 'manifest.json' in result:
        raise ValueError('manifest.json cannot self-hash')
    return result

@contextmanager
def research_run(project_root, analysis, run_id, input_roots, config, source_paths, expected_files):
    expected = output_artifacts_expected_paths(expected_files)
    run_id = safe_run_id(run_id)
    analysis = safe_analysis(analysis)
    root = Path(project_root).resolve() / 'outputs' / analysis / run_id
    ensure_output_outside_inputs(root, input_roots)
    root.mkdir(parents=True, exist_ok=False)
    manifest = {'schema_version': 1, 'analysis': analysis, 'run_id': run_id, 'status': 'running', 'required_artifacts': sorted(expected), 'artifacts': []}
    write_json_new(root / 'manifest.json', manifest)
    try:
        for name in ('data', 'tables', 'figures', 'diagnostics'):
            (root / name).mkdir()
        dependencies = {}
        for name in ('numpy', 'pandas', 'matplotlib', 'scipy'):
            try:
                dependencies[name] = version(name)
            except PackageNotFoundError:
                dependencies[name] = None
        inputs = []
        for p in input_roots:
            p = Path(p).resolve()
            identity = p if p.is_file() else p / 'manifest.json'
            if not identity.is_file():
                raise FileNotFoundError(f'Input identity file is missing: {identity}')
            inputs.append({'path': str(p), 'identity_path': str(identity), 'sha256': file_sha256(identity)})
        write_json_new(root / 'config.json', config)
        manifest.update({'git_sha': git_sha(project_root), 'python': platform.python_version(), 'dependencies': dependencies, 'source_hashes': {str(Path(p).resolve().relative_to(Path(project_root).resolve())): file_sha256(p) for p in source_paths}, 'inputs': inputs})
        yield (root, manifest)
        actual = {p.relative_to(root).as_posix() for p in root.rglob('*') if p.is_file() and p.name != 'manifest.json'}
        missing = expected - actual
        unexpected = actual - expected
        if missing or unexpected:
            raise RuntimeError(f'Artifact contract mismatch; missing={sorted(missing)}, unexpected={sorted(unexpected)}')
        empty = [name for name in expected if (root / name).stat().st_size <= 0]
        if empty:
            raise RuntimeError(f'Required artifacts are empty: {sorted(empty)}')
        artifacts = []
        for p in sorted(root.rglob('*')):
            if p.is_file() and p.name != 'manifest.json':
                artifacts.append({'path': p.relative_to(root).as_posix(), 'sha256': file_sha256(p), 'bytes': p.stat().st_size})
        manifest['artifacts'] = artifacts
        manifest['status'] = 'complete'
    except Exception as exc:
        manifest['status'] = 'failed'
        manifest['error'] = {'type': type(exc).__name__, 'message': str(exc)}
        raise
    finally:
        temp = root / 'manifest.json.tmp'
        temp.write_text(json.dumps(manifest, ensure_ascii=False, indent=2, sort_keys=True, allow_nan=False), encoding='utf-8')
        os.replace(temp, root / 'manifest.json')
```

### Extraction plan


```python
"""Pinned, read-only planning and execution for contextual extraction batches."""
import hashlib
import json
import re
from contextlib import nullcontext
from pathlib import Path
from typing import Any, Callable
extraction_batch_HEX = re.compile('^[0-9a-f]{64}$')
extraction_batch_ID = re.compile('^[A-Za-z0-9](?:[A-Za-z0-9._-]*[A-Za-z0-9])?$')
extraction_batch_EXP_FILES = {'manifest.json', 'config.json', 'vocab.json', 'samples.jsonl', 'splits.json', 'acoustic_targets.npy'}

def extraction_batch_hash(path: Path) -> str:
    h = hashlib.sha256()
    with path.open('rb') as stream:
        for block in iter(lambda: stream.read(1024 * 1024), b''):
            h.update(block)
    return h.hexdigest()

def extraction_batch_digest(value: Any) -> bool:
    return isinstance(value, str) and extraction_batch_HEX.fullmatch(value) is not None

def extraction_batch_exp_hashes(root: Path, manifest_hash: str) -> dict[str, str]:
    manifest = json.loads((root / 'manifest.json').read_text(encoding='utf-8'))
    artifacts = manifest.get('artifacts')
    if not isinstance(artifacts, dict) or not extraction_batch_digest(artifacts.get('acoustic_targets.npy')):
        raise ValueError(f'experiment manifest lacks acoustic_targets.npy hash: {root}')
    result = {name: manifest_hash if name == 'manifest.json' else artifacts.get(name) for name in extraction_batch_EXP_FILES}
    for name, digest in result.items():
        path = root / name
        if not extraction_batch_digest(digest) or not path.is_file() or extraction_batch_hash(path) != digest:
            raise ValueError(f'experiment source hash mismatch: {root / name}')
    return result

def resolve_encoder_source_hash(build_metadata: dict[str, Any]) -> str:
    """Resolve the encoder digest from an explicit verified build record."""
    hashes = build_metadata.get('source_sha256') if isinstance(build_metadata, dict) else None
    value = hashes.get('evaluation/encoders.py') if isinstance(hashes, dict) else None
    if not extraction_batch_digest(value):
        raise ValueError('build metadata lacks evaluation/encoders.py SHA-256')
    return value

def extraction_batch_safe_id(value: Any) -> bool:
    reserved = {'CON', 'PRN', 'AUX', 'NUL', *(f'COM{i}' for i in range(1, 10)), *(f'LPT{i}' for i in range(1, 10))}
    return isinstance(value, str) and extraction_batch_ID.fullmatch(value) is not None and ('..' not in value) and (not value.endswith(('.', ' '))) and (value.upper().split('.')[0] not in reserved)

def plan_contextual_extraction(*, bundle_path: str | Path, bundle_manifest_sha256: str, experiment_path: str | Path, experiment_manifest_sha256: str, training_runs: dict[str, str | Path], run_ids: dict[str, dict[int, str]], output_root: str | Path, device: str='cpu', batch_size: int=32, expected_seeds: tuple[int, ...]=(0, 1, 2, 3, 4), source_sha256: str) -> dict[str, Any]:
    """Validate all 25+ selected runs and return a deterministic JSON-compatible plan."""
    if not extraction_batch_digest(bundle_manifest_sha256) or not extraction_batch_digest(experiment_manifest_sha256) or (not extraction_batch_digest(source_sha256)):
        raise ValueError('manifest and source pins must be lowercase SHA-256 digests')
    if tuple(expected_seeds) != (0, 1, 2, 3, 4) or any((type(x) is not int for x in expected_seeds)):
        raise ValueError('expected_seeds must be exactly integer seeds 0 through 4')
    if not isinstance(device, str) or not device.strip() or type(batch_size) is not int or (batch_size <= 0):
        raise ValueError('device and positive integer batch_size are required')
    if not isinstance(training_runs, dict) or not training_runs or (not isinstance(run_ids, dict)):
        raise ValueError('training_runs and run_ids mappings are required')
    canonical = {}
    for key, path in training_runs.items():
        model = canonical_model_id(key)
        if key != model or model in canonical:
            raise ValueError('training_runs keys must be unique canonical model IDs')
        canonical[model] = path
    models = sorted(canonical)
    if set(run_ids) != set(models):
        raise ValueError('run_ids must contain exactly the canonical training models')
    bundle_root, exp_root = (Path(bundle_path).resolve(strict=True), Path(experiment_path).resolve(strict=True))
    if extraction_batch_hash(bundle_root / 'manifest.json') != bundle_manifest_sha256:
        raise ValueError('bundle manifest pin mismatch')
    bundle = load_evaluation_bundle(bundle_root)
    if bundle.get('manifest', {}).get('schema_version') != 3:
        raise ValueError('contextual extraction requires a schema-3 evaluation bundle')
    if extraction_batch_hash(exp_root / 'manifest.json') != experiment_manifest_sha256:
        raise ValueError('experiment manifest pin mismatch')
    six_hashes = extraction_batch_exp_hashes(exp_root, experiment_manifest_sha256)
    if bundle['manifest'].get('source_experiment', {}).get('hashes') != {k: v for k, v in six_hashes.items() if k != 'acoustic_targets.npy'}:
        raise ValueError('bundle and explicit experiment source inventory disagree')
    output = Path(output_root).resolve()
    entries, targets, input_roots = ([], set(), {bundle_root, exp_root})
    for model in models:
        model_ids = run_ids[model]
        if not isinstance(model_ids, dict) or set(model_ids) != {0, 1, 2, 3, 4} or any((type(s) is not int for s in model_ids)):
            raise ValueError(f'run_ids must use integer seed keys 0..4 for {model}')
        run_root = Path(canonical[model]).resolve(strict=True)
        input_roots.add(run_root)
        found = discover_seed_runs(run_root, model, (0, 1, 2, 3, 4))
        if found['experiment_hashes'] != six_hashes:
            raise ValueError(f'training source hashes mismatch for {model}: {run_root}')
        for rec in found['seeds']:
            seed, out_id = (rec['seed'], model_ids[rec['seed']])
            if not extraction_batch_safe_id(out_id):
                raise ValueError(f'unsafe representation output ID for {model} seed {seed}: {out_id!r}')
            run_dir = output / 'contextual-representations' / out_id
            data_dir = run_dir / 'data'
            if run_dir in targets:
                raise ValueError(f'output ID collision for {model} seed {seed}: {run_dir}')
            targets.add(run_dir)
            entries.append({'model_id': model, 'seed': seed, 'checkpoint_path': rec['checkpoint_path'], 'checkpoint_sha256': rec['checkpoint_sha256'], 'training_run_path': str(run_root), 'training_run_id': found['run_id'], 'training_manifest_sha256': found['manifest_sha256'], 'output_run_id': out_id, 'output_run_dir': str(run_dir), 'output_dir': str(data_dir)})
    if any((p.exists() for p in targets)):
        raise FileExistsError('one or more planned output run directories already exist')
    if any((t == i or t in i.parents or i in t.parents for t in targets for i in input_roots)):
        raise ValueError('planned output overlaps bundle, experiment, or training run input')
    return {'schema_version': 1, 'mode': 'extract', 'bundle_path': str(bundle_root), 'bundle_manifest_sha256': bundle_manifest_sha256, 'experiment_path': str(exp_root), 'experiment_manifest_sha256': experiment_manifest_sha256, 'experiment_hashes': six_hashes, 'training_runs': {m: str(canonical[m]) for m in models}, 'source_sha256': source_sha256, 'output_root': str(output), 'device': device, 'batch_size': batch_size, 'expected_seeds': [0, 1, 2, 3, 4], 'inventory': entries}

def extraction_batch_revalidate_plan(plan: dict[str, Any]) -> dict[str, Any]:
    if not isinstance(plan, dict) or plan.get('schema_version') != 1 or plan.get('mode') != 'extract':
        raise ValueError('unsupported or malformed contextual extraction plan')
    ids: dict[str, dict[int, str]] = {}
    inventory = plan.get('inventory')
    if not isinstance(inventory, list) or not inventory:
        raise ValueError('plan inventory is empty')
    for item in inventory:
        if not isinstance(item, dict) or type(item.get('seed')) is not int or item['seed'] not in range(5):
            raise ValueError('plan entry has invalid seed or shape')
        model = item.get('model_id')
        if not isinstance(model, str) or canonical_model_id(model) != model:
            raise ValueError('plan entry model ID is not canonical')
        ids.setdefault(model, {})[item['seed']] = item.get('output_run_id')
    regenerated = plan_contextual_extraction(bundle_path=plan['bundle_path'], bundle_manifest_sha256=plan['bundle_manifest_sha256'], experiment_path=plan['experiment_path'], experiment_manifest_sha256=plan['experiment_manifest_sha256'], training_runs=plan['training_runs'], run_ids=ids, output_root=plan['output_root'], device=plan['device'], batch_size=plan['batch_size'], expected_seeds=(0, 1, 2, 3, 4), source_sha256=plan['source_sha256'])
    if regenerated != plan:
        raise ValueError('extraction plan was modified or is stale; regenerate it')
    return regenerated

def execute_contextual_extraction(plan: dict[str, Any], *, extractor: Callable[..., Any] | None=None, context_factory: Callable[[dict[str, Any]], Any] | None=None) -> dict[str, Any]:
    """Execute only a revalidated full plan and verify all immutable contextual exports."""
    plan = extraction_batch_revalidate_plan(plan)
    if extractor is None:
        extractor = extract_bilstm
    bundle = load_evaluation_bundle(plan['bundle_path'])
    expected_ids = sorted((row['occurrence_id'] for row in bundle['occurrences']))
    result: dict[str, dict[int, str]] = {}
    exports = []
    for item in plan['inventory']:
        try:
            lifecycle = context_factory(item) if context_factory is not None else nullcontext()
            with lifecycle:
                extractor(bundle_path=plan['bundle_path'], experiment_path=plan['experiment_path'], checkpoint_path=item['checkpoint_path'], output_dir=item['output_dir'], device=plan['device'], batch_size=plan['batch_size'], contextual_only=True, expected_model_id=item['model_id'], expected_seed=item['seed'], source_sha256=plan['source_sha256'])
                loaded = load_contextual_representations(item['output_dir'])
                prov = loaded['manifest']['provenance']
                if prov['model_id'] != item['model_id'] or prov['seed'] != item['seed'] or prov['checkpoint_sha256'] != item['checkpoint_sha256'] or (prov['bundle_manifest_sha256'] != plan['bundle_manifest_sha256']) or (prov['source_sha256'] != plan['source_sha256']) or (prov['experiment_hashes'] != plan['experiment_hashes']) or (prov['config_sha256'] != plan['experiment_hashes']['config.json']) or (prov['vocab_sha256'] != plan['experiment_hashes']['vocab.json']) or (loaded['occurrence_ids'] != expected_ids):
                    raise ValueError('export provenance or occurrence inventory mismatch')
                digest = extraction_batch_hash(Path(item['output_dir']) / 'manifest.json')
            result.setdefault(item['model_id'], {})[item['seed']] = item['output_dir']
            exports.append({'model_id': item['model_id'], 'seed': item['seed'], 'representation_dir': item['output_dir'], 'manifest_sha256': digest, 'manifest_identity': prov})
        except Exception as error:
            raise ValueError(f"extraction failed for {item['model_id']} seed {item['seed']} at {item['output_dir']}: {error}") from error
    return {'representation_dirs': result, 'exports': exports}
```

### Numerical output files


```python
"""Immutable, hash-verified persistence for contextual evaluation outputs."""
from contextlib import contextmanager
import json
import os
from pathlib import Path
import platform
import tempfile
from typing import Any
from importlib.metadata import PackageNotFoundError, version

def evaluation_outputs_json(value: Any) -> bytes:
    return (json.dumps(value, ensure_ascii=False, indent=2, sort_keys=True, allow_nan=False) + '\n').encode('utf-8')

def evaluation_outputs_write(path: Path, value: Any) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open('xb') as stream:
        stream.write(evaluation_outputs_json(value))

def evaluation_outputs_tree_files(root: Path) -> set[str]:
    return {p.relative_to(root).as_posix() for p in root.rglob('*') if p.is_file()}

def evaluation_outputs_finite_json(path: Path):

    def bad(token):
        raise ValueError(f'non-finite JSON constant: {token}')
    value = json.loads(path.read_text(encoding='utf-8'), parse_constant=bad)
    evaluation_outputs_json(value)
    return value

def evaluation_outputs_pins(inputs, scientific_pins):
    bundle_pin = scientific_pins.get('bundle_manifest_sha256')
    if bundle_pin and bundle_pin not in {x['sha256'] for x in inputs}:
        raise ValueError('scientific bundle manifest pin does not match an input identity')
    exp_pin = scientific_pins.get('experiment_manifest_sha256')
    if exp_pin and exp_pin not in {x['sha256'] for x in inputs}:
        raise ValueError('scientific experiment manifest pin does not match an input identity')

def evaluation_outputs_verify_inventory(root: Path, analysis: str, run_id: str, expected: set[str], *, profile: str='numerical_outputs', figure_exports: str='pending'):
    safe_analysis(analysis)
    safe_run_id(run_id)
    manifest = evaluation_outputs_finite_json(root / 'manifest.json')
    if manifest.get('schema_version') != 1 or manifest.get('status') != 'complete' or manifest.get('analysis') != analysis or (manifest.get('run_id') != run_id) or (manifest.get('scope') != profile) or (manifest.get('figure_exports') != figure_exports) or (set(manifest.get('required_artifacts', [])) != expected):
        raise ValueError('output manifest identity/profile mismatch')
    if not isinstance(manifest.get('python'), str) or not manifest['python'] or (not isinstance(manifest.get('dependencies'), dict)) or ('numpy' not in manifest['dependencies']) or (not isinstance(manifest.get('inputs'), list)) or (not isinstance(manifest.get('source_hashes'), dict)) or (not manifest['source_hashes']):
        raise ValueError('output provenance metadata is incomplete')
    for row in manifest['inputs']:
        if not isinstance(row, dict) or not isinstance(row.get('path'), str) or (not isinstance(row.get('identity_path'), str)) or (not evaluation_outputs_is_sha(row.get('sha256'))):
            raise ValueError('output input identity metadata is invalid')
    if any((not evaluation_outputs_is_sha(x) for x in manifest['source_hashes'].values())):
        raise ValueError('output source hash metadata is invalid')
    actual, inventory = (evaluation_outputs_tree_files(root) - {'manifest.json'}, manifest.get('artifacts'))
    declared = {x.get('path') for x in inventory if isinstance(x, dict)} if isinstance(inventory, list) else set()
    if actual != expected or declared != expected or len(inventory or []) != len(expected):
        raise ValueError('output artifact inventory mismatch')
    for row in inventory:
        if not isinstance(row, dict) or set(row) != {'path', 'sha256', 'bytes'}:
            raise ValueError('malformed artifact inventory row')
        path = root / row['path']
        if not path.is_file() or path.stat().st_size <= 0 or path.stat().st_size != row['bytes'] or (file_sha256(path) != row['sha256']):
            raise ValueError(f"artifact hash or size mismatch: {row.get('path')}")
    return manifest

def evaluation_outputs_is_sha(value):
    return isinstance(value, str) and len(value) == 64 and all((c in '0123456789abcdef' for c in value))

@contextmanager
def evaluation_outputs_evaluation_run(project_root: str | Path, analysis: str, run_id: str, *, input_roots: list[str | Path], config: dict, source_hashes: dict, expected_files: set[str], scientific_pins: dict, create_data: bool=True, output_path: str | Path | None=None, profile: str='numerical_outputs', figure_exports_on_complete: str='pending'):
    """Write one run, gating every file except its top-level manifest."""
    analysis, run_id = (safe_analysis(analysis), safe_run_id(run_id))
    if figure_exports_on_complete not in {'pending', 'complete'}:
        raise ValueError('unsupported figure export completion status')
    root = Path(output_path).resolve() if output_path is not None else Path(project_root).resolve() / 'outputs' / analysis / run_id
    ensure_output_outside_inputs(root, input_roots)
    if root.exists():
        raise FileExistsError(f'refusing to overwrite existing output directory: {root}')
    if not isinstance(config, dict) or not isinstance(source_hashes, dict) or (not source_hashes):
        raise ValueError('explicit config and source hash map are required')
    if any((not isinstance(v, str) or len(v) != 64 or any((c not in '0123456789abcdef' for c in v)) for v in source_hashes.values())):
        raise ValueError('source hashes must be lowercase SHA-256 digests')
    evaluation_outputs_json({'config': config, 'source_hashes': source_hashes, 'scientific_pins': scientific_pins})
    inputs = []
    for raw in input_roots:
        p = Path(raw).resolve(strict=True)
        identity = p if p.is_file() else p / 'manifest.json'
        if not identity.is_file():
            raise FileNotFoundError(f'input manifest identity missing: {identity}')
        inputs.append({'path': str(p), 'identity_path': str(identity), 'sha256': file_sha256(identity)})
    evaluation_outputs_pins(inputs, scientific_pins)
    dependencies = {}
    for package in ('numpy', 'scipy', 'pandas', 'matplotlib'):
        try:
            dependencies[package] = version(package)
        except PackageNotFoundError:
            dependencies[package] = None
    root.mkdir(parents=True, exist_ok=False)
    manifest = {'schema_version': 1, 'analysis': analysis, 'run_id': run_id, 'status': 'running', 'scope': profile, 'figure_exports': 'pending', 'required_artifacts': sorted(expected_files), 'artifacts': [], 'scientific_input_pins': scientific_pins, 'source_hashes': source_hashes, 'inputs': inputs, 'python': platform.python_version(), 'dependencies': dependencies}
    evaluation_outputs_write(root / 'manifest.json', manifest)
    try:
        names = (('data',) if create_data else ()) + ('tables', 'figures', 'diagnostics')
        for name in names:
            (root / name).mkdir()
        evaluation_outputs_write(root / 'config.json', config)
        yield root
        actual = evaluation_outputs_tree_files(root) - {'manifest.json'}
        if actual != expected_files:
            raise RuntimeError(f'artifact inventory mismatch; missing={sorted(expected_files - actual)}, unexpected={sorted(actual - expected_files)}')
        artifacts = []
        for relative in sorted(actual):
            path = root / relative
            if path.stat().st_size <= 0:
                raise RuntimeError(f'required artifact is empty: {relative}')
            artifacts.append({'path': relative, 'sha256': file_sha256(path), 'bytes': path.stat().st_size})
        manifest.update(status='complete', artifacts=artifacts, figure_exports=figure_exports_on_complete)
    except BaseException as exc:
        manifest.update(status='failed', error={'type': type(exc).__name__, 'message': str(exc)})
        raise
    finally:
        fd, name = tempfile.mkstemp(prefix='manifest-', suffix='.tmp', dir=root)
        try:
            with os.fdopen(fd, 'wb') as stream:
                stream.write(evaluation_outputs_json(manifest))
            os.replace(name, root / 'manifest.json')
        finally:
            if os.path.exists(name):
                os.unlink(name)

def write_evaluation_outputs(project_root: str | Path, contract: dict, *, run_ids: dict[str, str], input_roots: list[str | Path], source_hashes: dict[str, str], resolved_config: dict) -> dict[str, Path]:
    """Persist both report-contract analyses to their explicit immutable run IDs."""
    if not isinstance(contract, dict) or not isinstance(run_ids, dict) or set(run_ids) != {'contextual-anchor-q', 'contextual-dimsim'}:
        raise ValueError('contract and explicit IDs for both analyses are required')
    rebuilt = build_evaluation_report(contract['raw_batches']['contextual-anchor-q'], contract['raw_batches']['contextual-dimsim'], contract['reference_model'], run_ids)
    for key in ('scientific_input_pins', 'reports', 'figures'):
        if contract.get(key) != rebuilt.get(key):
            raise ValueError(f'evaluation contract {key} is stale or inconsistent')
    if not isinstance(source_hashes, dict) or not source_hashes:
        raise ValueError('explicit source code hash map is required')
    if any((not isinstance(v, str) or len(v) != 64 or any((c not in '0123456789abcdef' for c in v)) for v in source_hashes.values())):
        raise ValueError('source hashes must be lowercase SHA-256 digests')
    if not isinstance(resolved_config, dict):
        raise ValueError('resolved config is required')
    project = Path(project_root).resolve()
    target_roots = []
    for name in ('contextual-anchor-q', 'contextual-dimsim'):
        target = project / 'outputs' / safe_analysis(name) / safe_run_id(run_ids[name])
        ensure_output_outside_inputs(target, input_roots)
        if target.exists():
            raise FileExistsError(f'refusing to overwrite existing output directory: {target}')
        if any((target == prior or target in prior.parents or prior in target.parents for prior in target_roots)):
            raise ValueError('evaluation output targets overlap')
        target_roots.append(target)
    basepins = contract['scientific_input_pins']
    results = {}
    for analysis in ('contextual-anchor-q', 'contextual-dimsim'):
        run_id = safe_run_id(run_ids[analysis])
        q = analysis.endswith('anchor-q')
        batch_name = 'anchor_q_batch.json' if q else 'dimsim_batch.json'
        report_name = 'anchor_q_report.json' if q else 'dimsim_report.json'
        report = rebuilt['reports'][analysis]
        batch = rebuilt['raw_batches'][analysis]
        tables, plot = (report['tables'], report['plot_data'])
        expected = {'config.json', f'data/{batch_name}', f'data/{report_name}', 'tables/tables.json', 'figures/plot_data.json', 'diagnostics/summary.json'}
        pins = {**basepins, 'analysis': analysis, 'run_id': run_id}
        with evaluation_outputs_evaluation_run(project_root, analysis, run_id, input_roots=input_roots, config=resolved_config, source_hashes=source_hashes, expected_files=expected, scientific_pins=pins) as root:
            evaluation_outputs_write(root / f'data/{batch_name}', batch)
            evaluation_outputs_write(root / f'data/{report_name}', report)
            evaluation_outputs_write(root / 'tables/tables.json', tables)
            evaluation_outputs_write(root / 'figures/plot_data.json', plot)
            evaluation_outputs_write(root / 'diagnostics/summary.json', {'scope': 'numerical_outputs', 'figure_exports': 'pending', 'model_count': len(basepins['model_ids']), 'seed_ids': basepins['seeds'], 'table_row_counts': {name: len(rows) for name, rows in tables.items()}, 'plot_data_keys': sorted(plot)})
        results[analysis] = root
    return results

def load_evaluation_outputs(root: str | Path) -> dict:
    """Load a complete numerical run after checking exact inventory and hashes."""
    root = Path(root).resolve(strict=True)
    manifest_path = root / 'manifest.json'
    analysis = root.parent.name
    if analysis not in {'contextual-anchor-q', 'contextual-dimsim'}:
        raise ValueError('unknown evaluation output analysis')
    q = analysis.endswith('anchor-q')
    bn, rn = ('anchor_q_batch.json', 'anchor_q_report.json') if q else ('dimsim_batch.json', 'dimsim_report.json')
    expected = {'config.json', f'data/{bn}', f'data/{rn}', 'tables/tables.json', 'figures/plot_data.json', 'diagnostics/summary.json'}
    manifest = evaluation_outputs_verify_inventory(root, analysis, root.name, expected)
    evaluation_outputs_pins(manifest.get('inputs', []), manifest.get('scientific_input_pins', {}))
    if set(manifest['scientific_input_pins']) != {'bundle_manifest_sha256', 'model_ids', 'seeds', 'model_seed_provenance', 'analysis', 'run_id'}:
        raise ValueError('numerical output pins are malformed')
    batch, report = (evaluation_outputs_finite_json(root / f'data/{bn}'), evaluation_outputs_finite_json(root / f'data/{rn}'))
    tables, plot = (evaluation_outputs_finite_json(root / 'tables/tables.json'), evaluation_outputs_finite_json(root / 'figures/plot_data.json'))
    config = evaluation_outputs_finite_json(root / 'config.json')
    evaluation_outputs_finite_json(root / 'diagnostics/summary.json')
    if tables != report.get('tables') or plot != report.get('plot_data'):
        raise ValueError('stored tables or plot data differ from stored report')
    if batch.get('bundle_manifest_sha256') != manifest['scientific_input_pins']['bundle_manifest_sha256']:
        raise ValueError('stored batch bundle pin mismatch')
    results = batch.get('results')
    pins = manifest['scientific_input_pins']
    if not isinstance(results, list) or sorted({r.get('model_id') for r in results if isinstance(r, dict)}) != pins['model_ids']:
        raise ValueError('stored batch model inventory differs from manifest pins')
    if any((not isinstance(r, dict) or type(r.get('seed')) is not int or r.get('seed') not in pins['seeds'] for r in results)):
        raise ValueError('stored batch seed inventory is invalid')
    identity_pins = {(p.get('model_id'), p.get('seed')): p for p in pins['model_seed_provenance'] if isinstance(p, dict)}
    expected_pairs = {(model, seed) for model in pins['model_ids'] for seed in pins['seeds']}
    if len(identity_pins) != len(pins['model_seed_provenance']) or set(identity_pins) != expected_pairs:
        raise ValueError('manifest model/seed provenance inventory is invalid')
    for row in results:
        pin = identity_pins[row['model_id'], row['seed']]
        if any((row.get(k) != pin.get(k) for k in ('run_id', 'run_manifest_sha256', 'checkpoint_sha256'))):
            raise ValueError('stored batch provenance differs from manifest pins')
        if row.get('experiment_hashes') is not None and row['experiment_hashes'] != pin.get('experiment_hashes'):
            raise ValueError('stored batch experiment hashes differ from manifest pins')
    if config.get('reference_model') is not None and report.get('reference_model') is not None and (report.get('reference_model') != config['reference_model']):
        raise ValueError('stored report reference metadata differs from config')
    return {'root': root, 'manifest': manifest, 'config': config, 'batch': batch, 'report': report, 'tables': tables, 'plot_data': plot}

def write_random_control_outputs(project_root: str | Path, run_id: str, control: dict, *, input_roots: list[str | Path], source_hashes: dict[str, str], resolved_config: dict) -> Path:
    """Persist an already evaluated random control without loading learned representations."""
    import io
    import numpy as np
    if not isinstance(control, dict) or not isinstance(control.get('report'), dict):
        raise ValueError('evaluated random-control result is required')
    ids, matrices, report = (control.get('occurrence_ids'), control.get('matrices'), control['report'])
    if not isinstance(ids, list) or not ids or ids != sorted(ids) or (len(ids) != len(set(ids))) or any((not isinstance(x, str) or not x for x in ids)) or (not isinstance(matrices, dict)) or any((type(s) is not int for s in matrices)) or (set(matrices) != set(RANDOM_SEEDS)):
        raise ValueError('random-control IDs and five matrices are required')
    dimension = None
    normalized_matrices = {}
    for seed in RANDOM_SEEDS:
        matrix = np.asarray(matrices[seed])
        if matrix.dtype != np.dtype('float32') or matrix.ndim != 2 or matrix.shape[0] != len(ids) or (not np.isfinite(matrix).all()):
            raise ValueError(f'random matrix {seed} violates finite float32 shape contract')
        if dimension is None:
            dimension = matrix.shape[1]
        if matrix.shape[1] != dimension or np.any(np.linalg.norm(matrix.astype(np.float64), axis=1) == 0):
            raise ValueError('random matrix dimensions or norms are invalid')
        normalized_matrices[seed] = matrix
    if report.get('seed_kind') != 'random' or report.get('model_id') != 'random-control' or report.get('random_seeds') != list(RANDOM_SEEDS):
        raise ValueError('control report identity or seed-kind is invalid')
    if not evaluation_outputs_is_sha(report.get('bundle_manifest_sha256')) or not isinstance(report.get('per_seed'), list) or len(report['per_seed']) != 5:
        raise ValueError('random-control report bundle pin and five seed records are required')
    regenerated = generate_random_control(ids, dimension=int(dimension))
    for seed in RANDOM_SEEDS:
        if normalized_matrices[seed].tobytes(order='C') != regenerated['matrices'][seed].tobytes(order='C'):
            raise ValueError(f'random matrix {seed} does not match its declared PCG64 realization')
        rows = report['per_seed']
        if not isinstance(rows[seed], dict) or rows[seed].get('random_seed') != seed or rows[seed].get('provenance', {}).get('dimension') != dimension or (rows[seed].get('provenance', {}).get('bundle_manifest_sha256') != report.get('bundle_manifest_sha256')):
            raise ValueError('random report seed, dimension, or bundle metadata mismatch')
    try:
        summary_matches = evaluation_outputs_json(summarize_random_control(report['per_seed'], report['bundle_manifest_sha256'])) == evaluation_outputs_json(report)
    except (KeyError, TypeError, ValueError) as exc:
        raise ValueError('random-control per-seed report is invalid') from exc
    if not summary_matches:
        raise ValueError('random-control report is stale relative to its per-seed records')
    target = Path(project_root).resolve() / 'outputs' / 'contextual-random-control' / safe_run_id(run_id)
    ensure_output_outside_inputs(target, input_roots)
    if target.exists():
        raise FileExistsError(f'refusing to overwrite existing output directory: {target}')
    if not isinstance(resolved_config, dict):
        raise ValueError('resolved config is required')
    expected = {'config.json', 'data/occurrence_ids.json', 'data/control_report.json', 'tables/tables.json', 'figures/plot_data.json', 'diagnostics/summary.json', *(f'data/random_seed_{s}.npy' for s in RANDOM_SEEDS)}
    pins = {'bundle_manifest_sha256': report.get('bundle_manifest_sha256'), 'seed_kind': 'random', 'random_seeds': list(RANDOM_SEEDS), 'dimension': int(dimension)}
    with evaluation_outputs_evaluation_run(project_root, 'contextual-random-control', run_id, input_roots=input_roots, config=resolved_config, source_hashes=source_hashes, expected_files=expected, scientific_pins=pins) as root:
        evaluation_outputs_write(root / 'data/occurrence_ids.json', ids)
        evaluation_outputs_write(root / 'data/control_report.json', report)
        for seed in RANDOM_SEEDS:
            buf = io.BytesIO()
            np.save(buf, matrices[seed], allow_pickle=False)
            (root / f'data/random_seed_{seed}.npy').write_bytes(buf.getvalue())
        q, d = (report.get('anchor_q', {}), report.get('dimsim', {}))
        tables = {'anchor_q': q, 'dimsim': d}
        plot = {'anchor_q': q, 'dimsim': d}
        evaluation_outputs_write(root / 'tables/tables.json', tables)
        evaluation_outputs_write(root / 'figures/plot_data.json', plot)
        evaluation_outputs_write(root / 'diagnostics/summary.json', {'scope': 'numerical_outputs', 'figure_exports': 'pending', 'occurrence_count': len(ids), 'dimension': int(dimension), 'random_seeds': list(RANDOM_SEEDS)})
    return target

def load_random_control_summary(root: str | Path) -> dict:
    """Read hash-verified random-control summaries without loading vector matrices."""
    root = Path(root).resolve(strict=True)
    expected = {'config.json', 'data/occurrence_ids.json', 'data/control_report.json', 'tables/tables.json', 'figures/plot_data.json', 'diagnostics/summary.json', *(f'data/random_seed_{seed}.npy' for seed in range(5))}
    manifest = evaluation_outputs_verify_inventory(root, 'contextual-random-control', root.name, expected)
    pins = manifest.get('scientific_input_pins', {})
    evaluation_outputs_pins(manifest.get('inputs', []), pins)
    if set(pins) != {'bundle_manifest_sha256', 'seed_kind', 'random_seeds', 'dimension'} or pins['seed_kind'] != 'random' or pins['random_seeds'] != list(range(5)) or (type(pins['dimension']) is not int) or (pins['dimension'] <= 0):
        raise ValueError('random-control scientific pins malformed')
    report = evaluation_outputs_finite_json(root / 'data/control_report.json')
    if report.get('schema_version') != 1 or report.get('model_id') != 'random-control' or report.get('seed_kind') != 'random' or (report.get('random_seeds') != list(range(5))) or (report.get('bundle_manifest_sha256') != pins['bundle_manifest_sha256']):
        raise ValueError('stored random-control identity mismatch')
    tables = evaluation_outputs_finite_json(root / 'tables/tables.json')
    plot = evaluation_outputs_finite_json(root / 'figures/plot_data.json')
    if tables != {'anchor_q': report.get('anchor_q'), 'dimsim': report.get('dimsim')} or plot != tables:
        raise ValueError('random-control table or plot payload mismatch')
    return {'root': root, 'manifest': manifest, 'config': evaluation_outputs_finite_json(root / 'config.json'), 'report': {key: value for key, value in report.items() if key != 'per_seed'}, 'tables': tables, 'plot_data': plot}

def load_random_control_outputs(root: str | Path) -> dict:
    """Recover persisted Gaussian controls directly; this never invokes learned loaders."""
    import numpy as np
    root = Path(root).resolve(strict=True)
    expected = {'config.json', 'data/occurrence_ids.json', 'data/control_report.json', 'tables/tables.json', 'figures/plot_data.json', 'diagnostics/summary.json', *(f'data/random_seed_{s}.npy' for s in range(5))}
    manifest = evaluation_outputs_verify_inventory(root, 'contextual-random-control', root.name, expected)
    evaluation_outputs_pins(manifest.get('inputs', []), manifest.get('scientific_input_pins', {}))
    if set(manifest['scientific_input_pins']) != {'bundle_manifest_sha256', 'seed_kind', 'random_seeds', 'dimension'}:
        raise ValueError('random-control scientific pins malformed')
    ids = evaluation_outputs_finite_json(root / 'data/occurrence_ids.json')
    matrices = {seed: np.load(root / f'data/random_seed_{seed}.npy', allow_pickle=False) for seed in range(5)}
    report = evaluation_outputs_finite_json(root / 'data/control_report.json')
    if not isinstance(ids, list) or not ids or any((not isinstance(x, str) or not x for x in ids)) or (ids != sorted(ids)) or (len(ids) != len(set(ids))) or (report.get('model_id') != 'random-control') or (report.get('seed_kind') != 'random') or (report.get('bundle_manifest_sha256') != manifest['scientific_input_pins']['bundle_manifest_sha256']) or (report.get('random_seeds') != [0, 1, 2, 3, 4]):
        raise ValueError('stored random-control identity mismatch')
    dimension = manifest['scientific_input_pins']['dimension']
    expected_matrices = generate_random_control(ids, dimension=dimension)['matrices']
    for seed, matrix in matrices.items():
        if matrix.dtype != np.dtype('float32') or matrix.shape != (len(ids), dimension) or (not np.isfinite(matrix).all()) or np.any(np.linalg.norm(matrix.astype(np.float64), axis=1) == 0) or (matrix.tobytes(order='C') != expected_matrices[seed].tobytes(order='C')):
            raise ValueError(f'stored random matrix {seed} violates declared realization')
    if evaluation_outputs_json(summarize_random_control(report['per_seed'], report['bundle_manifest_sha256'])) != evaluation_outputs_json(report):
        raise ValueError('stored random-control report summary is stale')
    tables, plot = (evaluation_outputs_finite_json(root / 'tables/tables.json'), evaluation_outputs_finite_json(root / 'figures/plot_data.json'))
    if tables != {'anchor_q': report['anchor_q'], 'dimsim': report['dimsim']} or plot != tables:
        raise ValueError('random-control table or plot payload mismatch')
    evaluation_outputs_finite_json(root / 'config.json')
    evaluation_outputs_finite_json(root / 'diagnostics/summary.json')
    return {'manifest': manifest, 'config': evaluation_outputs_finite_json(root / 'config.json'), 'occurrence_ids': ids, 'matrices': matrices, 'report': report, 'tables': tables, 'plot_data': plot}

def run_contextual_extraction_outputs(plan: dict, *, extractor=None) -> dict:
    """Execute an explicit extraction plan and gate each raw export after validation."""
    from contextlib import contextmanager
    plan = extraction_batch_revalidate_plan(plan)
    inputs = [plan['bundle_path'], plan['experiment_path'], *plan['training_runs'].values()]
    source_hashes = {'evaluation/encoders.py': plan['source_sha256']}
    contexts = {}
    for item in plan['inventory']:
        run_id = safe_run_id(item['output_run_id'])
        root = Path(item['output_run_dir'])
        pins = {'bundle_manifest_sha256': plan['bundle_manifest_sha256'], 'experiment_manifest_sha256': plan['experiment_manifest_sha256'], 'experiment_hashes': plan['experiment_hashes'], 'model_id': item['model_id'], 'seed': item['seed'], 'checkpoint_sha256': item['checkpoint_sha256'], 'training_manifest_sha256': item['training_manifest_sha256']}
        expected = {'config.json', 'data/manifest.json', 'data/occurrence_ids.json', 'data/occurrence_hidden.npy', 'diagnostics/summary.json'}
        config = {'mode': 'contextual_only', 'device': plan['device'], 'batch_size': plan['batch_size'], 'model_id': item['model_id'], 'seed': item['seed'], 'checkpoint_sha256': item['checkpoint_sha256']}

        @contextmanager
        def lifecycle(item=item, root=root, run_id=run_id, pins=pins, expected=expected, config=config):
            with evaluation_outputs_evaluation_run(plan['output_root'], 'contextual-representations', run_id, input_roots=inputs, config=config, source_hashes=source_hashes, expected_files=expected, scientific_pins=pins, create_data=False, output_path=root, profile='contextual_representation_export') as runroot:
                yield
                evaluation_outputs_write(runroot / 'diagnostics/summary.json', {'model_id': item['model_id'], 'seed': item['seed'], 'representation': 'contextual_hidden_prehead', 'validated_after_export': True, 'nested_manifest_sha256': file_sha256(root / 'data/manifest.json')})
        contexts[item['output_run_id']] = lifecycle
    metadata = execute_contextual_extraction(plan, extractor=extractor, context_factory=lambda item: contexts[item['output_run_id']]())
    return metadata

def load_contextual_extraction_outputs(root: str | Path) -> dict:
    """Recover a validated raw contextual export including its nested manifest pin."""
    import numpy as np
    root = Path(root).resolve(strict=True)
    expected = {'config.json', 'data/manifest.json', 'data/occurrence_ids.json', 'data/occurrence_hidden.npy', 'diagnostics/summary.json'}
    manifest = evaluation_outputs_verify_inventory(root, 'contextual-representations', root.name, expected, profile='contextual_representation_export')
    nested_export = load_contextual_representations(root / 'data')
    nested = nested_export['manifest']
    diagnostic = evaluation_outputs_finite_json(root / 'diagnostics/summary.json')
    pins = manifest.get('scientific_input_pins', {})
    provenance = nested.get('provenance', {})
    if diagnostic.get('nested_manifest_sha256') != file_sha256(root / 'data/manifest.json') or provenance.get('model_id') != pins.get('model_id') or provenance.get('seed') != pins.get('seed') or (provenance.get('checkpoint_sha256') != pins.get('checkpoint_sha256')) or (provenance.get('bundle_manifest_sha256') != pins.get('bundle_manifest_sha256')) or (provenance.get('experiment_hashes') != pins.get('experiment_hashes')) or (provenance.get('source_sha256') != manifest.get('source_hashes', {}).get('evaluation/encoders.py')):
        raise ValueError('nested contextual manifest pin mismatch')
    evaluation_outputs_pins(manifest.get('inputs', []), pins)
    return {'manifest': manifest, 'config': evaluation_outputs_finite_json(root / 'config.json'), 'occurrence_ids': json.loads((root / 'data/occurrence_ids.json').read_text(encoding='utf-8')), 'hidden': nested_export['hidden'], 'representation_manifest': nested, 'diagnostics': diagnostic}
```

### Evaluation pipeline


```python
"""Thin orchestration helpers for the generated five-model evaluation notebooks."""
import hashlib
import json
from pathlib import Path
from typing import Any
MODEL_ORDER = ('syllable', 'components', 'acoustic', 'components-acoustic-lambda-1', 'components-acoustic-lambda-10')
notebook_runtime_SEEDS = (0, 1, 2, 3, 4)

def notebook_runtime_sha(path: Path) -> str:
    return hashlib.sha256(path.read_bytes()).hexdigest()

def notebook_runtime_build_source_digest(build_metadata: dict) -> str:
    hashes = build_metadata.get('source_sha256') if isinstance(build_metadata, dict) else None
    if not isinstance(hashes, dict) or not hashes or any((not isinstance(v, str) or len(v) != 64 or any((c not in '0123456789abcdef' for c in v)) for v in hashes.values())):
        raise ValueError('BUILD_METADATA requires the complete source SHA-256 mapping')
    payload = json.dumps(hashes, sort_keys=True, separators=(',', ':'), ensure_ascii=False).encode('utf-8')
    return hashlib.sha256(payload).hexdigest()

def notebook_runtime_reject_immutable_ancestors(target: Path) -> None:
    for ancestor in (target, *target.parents):
        manifest = ancestor / 'manifest.json'
        if not manifest.is_file():
            continue
        try:
            record = json.loads(manifest.read_text(encoding='utf-8'))
        except (OSError, ValueError):
            continue
        if str(record.get('analysis', '')).startswith('contextual-') and record.get('status') in {'complete', 'failed'}:
            raise ValueError(f'Refusing output beneath immutable evaluation run: {ancestor}')

def preflight_evaluation(config: dict, build_metadata: dict) -> dict[str, Any]:
    """Validate explicit inputs and every 25-seed/output target without creating files."""
    required = ('bundle_path', 'bundle_manifest_sha256', 'experiment_path', 'experiment_manifest_sha256', 'model_run_paths', 'extraction_mode', 'extraction_prefix', 'existing_representation_runs', 'project_root', 'reference_model', 'run_ids', 'device', 'batch_size', 'dimsim_boundary_start', 'dimsim_boundary_stop', 'dimsim_boundary_step', 'dimsim_min_side_count', 'dimsim_min_unique_pairs_per_bin')
    if not isinstance(config, dict) or set(required) - set(config):
        raise ValueError('The single parameter cell must define every evaluation input and output')

    def path_value(key):
        value = config[key]
        if not isinstance(value, (str, Path)) or not str(value).strip() or 'EDIT_ME' in str(value).upper():
            raise ValueError(f'Set explicit {key} in the parameter cell')
        return Path(value).resolve(strict=True)
    bundle_path = path_value('bundle_path')
    experiment_path = path_value('experiment_path')
    project_root = Path(config['project_root']).expanduser().resolve()
    if not str(config['project_root']).strip() or 'EDIT_ME' in str(config['project_root']).upper():
        raise ValueError('Set explicit project_root in the parameter cell')
    bundle_manifest = bundle_path / 'manifest.json'
    experiment_manifest = experiment_path / 'manifest.json'
    if notebook_runtime_sha(bundle_manifest) != config['bundle_manifest_sha256']:
        raise ValueError('Evaluation bundle manifest SHA-256 mismatch')
    if notebook_runtime_sha(experiment_manifest) != config['experiment_manifest_sha256']:
        raise ValueError('Experiment manifest SHA-256 mismatch')
    bundle = load_evaluation_bundle(bundle_path)
    if bundle['manifest'].get('schema_version') != 3:
        raise ValueError('A schema-3 contextual evaluation bundle is required')
    model_paths = config['model_run_paths']
    if not isinstance(model_paths, dict) or set(model_paths) != set(MODEL_ORDER):
        raise ValueError('Provide exactly the five canonical model run paths')
    if config['reference_model'] != 'syllable' or not isinstance(config['device'], str) or (not config['device'].strip()) or (type(config['batch_size']) is not int) or (config['batch_size'] <= 0):
        raise ValueError('Set the syllable reference, a nonempty device, and a positive integer batch size')
    if any((type(config[k]) is not int for k in ('dimsim_boundary_start', 'dimsim_boundary_stop', 'dimsim_boundary_step', 'dimsim_min_side_count', 'dimsim_min_unique_pairs_per_bin'))):
        raise ValueError('DIMSIM thresholds and support counts must be integers')
    if config['dimsim_boundary_start'] < 10 or config['dimsim_boundary_stop'] > 110 or config['dimsim_boundary_stop'] < config['dimsim_boundary_start'] or (config['dimsim_boundary_step'] <= 0) or (config['dimsim_min_side_count'] < 1) or (config['dimsim_min_unique_pairs_per_bin'] < 1):
        raise ValueError('DIMSIM thresholds must lie in 10..110 with positive step and support counts')
    if config['reference_model'] != 'syllable':
        raise ValueError('reference_model must be syllable')
    runs = {}
    rep_ids = {}
    discovered = {}
    encoder_sha = resolve_encoder_source_hash(build_metadata)
    experiment_hashes = extraction_batch_exp_hashes(experiment_path, config['experiment_manifest_sha256'])
    if bundle['manifest'].get('source_experiment', {}).get('hashes') != {k: v for k, v in experiment_hashes.items() if k != 'acoustic_targets.npy'}:
        raise ValueError("Selected experiment artifacts do not match the bundle's complete source inventory")
    if config['extraction_mode'] not in {'extract', 'reuse'}:
        raise ValueError('extraction_mode must be extract or reuse')
    if config['extraction_mode'] == 'extract':
        if config['existing_representation_runs'] is not None:
            raise ValueError('extract mode requires existing_representation_runs=None')
        prefix = safe_run_id(config['extraction_prefix'])
    else:
        prefix = None
        existing = config['existing_representation_runs']
        if not isinstance(existing, dict) or set(existing) != set(MODEL_ORDER):
            raise ValueError('reuse mode requires exact model -> seed representation pins')
        for model in MODEL_ORDER:
            if not isinstance(existing[model], dict) or set(existing[model]) != set(notebook_runtime_SEEDS):
                raise ValueError(f'reuse mode requires exactly seeds 0..4 for {model}')
            for seed, item in existing[model].items():
                if not isinstance(item, dict) or set(item) != {'path', 'manifest_sha256'}:
                    raise ValueError(f'invalid representation pin for {model} seed {seed}')
    for key in ('bundle_manifest_sha256', 'experiment_manifest_sha256'):
        if not isinstance(config[key], str) or len(config[key]) != 64 or any((c not in '0123456789abcdef' for c in config[key])):
            raise ValueError(f'Set explicit lowercase SHA-256 for {key}')
    for model in MODEL_ORDER:
        run_path = Path(model_paths[model]).resolve(strict=True)
        found = discover_seed_runs(run_path, model, notebook_runtime_SEEDS)
        runs[model] = run_path
        discovered[model] = found
        if found['experiment_hashes'] != experiment_hashes:
            raise ValueError(f'Training source inventory does not match selected bundle: {model}')
        seed_records = {row['seed']: row for row in found['seeds']}
        found['seeds_by_id'] = seed_records
        rep_ids[model] = {seed: safe_run_id(f'{prefix}-{model}-seed-{seed}') if prefix else None for seed in notebook_runtime_SEEDS}
    ids = [rep_ids[m][s] for m in MODEL_ORDER for s in notebook_runtime_SEEDS]
    if prefix and len(set(ids)) != 25:
        raise ValueError('Expanded extraction IDs collide')
    run_ids = config['run_ids']
    expected_run_names = {'contextual-anchor-q', 'contextual-dimsim', 'contextual-random-control'}
    if not isinstance(run_ids, dict) or set(run_ids) != expected_run_names:
        raise ValueError('Explicit Q, DIMSIM, and random run IDs are required')
    for value in run_ids.values():
        safe_run_id(value)
    numerical_targets = [project_root / 'outputs' / name / run_ids[name] for name in expected_run_names]
    targets = list(numerical_targets)
    if config['extraction_mode'] == 'extract':
        targets.extend((project_root / 'outputs' / 'contextual-representations' / run_id for run_id in ids))
    for target in targets:
        ensure_output_outside_inputs(target, [bundle_path, experiment_path, *runs.values()])
        notebook_runtime_reject_immutable_ancestors(target)
        if target.exists():
            raise FileExistsError(f'Evaluation output target already exists: {target}')
    reps = {m: {} for m in MODEL_ORDER}
    if config['extraction_mode'] == 'reuse':
        for model in MODEL_ORDER:
            for seed in notebook_runtime_SEEDS:
                item = config['existing_representation_runs'][model][seed]
                rep_root = Path(item['path']).resolve(strict=True)
                if not isinstance(item.get('manifest_sha256'), str) or notebook_runtime_sha(rep_root / 'manifest.json') != item['manifest_sha256']:
                    raise ValueError(f'Explicit representation manifest pin mismatch: {model} seed {seed}')
                safe_run_id(rep_root.name)
                export = load_contextual_extraction_outputs(rep_root)
                pin = export['manifest']['scientific_input_pins']
                record = discovered[model]['seeds_by_id'][seed]
                expected = {'bundle_manifest_sha256': config['bundle_manifest_sha256'], 'experiment_manifest_sha256': config['experiment_manifest_sha256'], 'model_id': model, 'seed': seed, 'checkpoint_sha256': record['checkpoint_sha256'], 'training_manifest_sha256': discovered[model]['manifest_sha256']}
                if any((pin.get(k) != v for k, v in expected.items())) or export['manifest']['source_hashes'].get('evaluation/encoders.py') != encoder_sha:
                    raise ValueError(f'Reused contextual extraction pins mismatch for {model} seed {seed}')
                rep_ids[model][seed] = export['manifest']['run_id']
                reps[model][seed] = rep_root / 'data'
        ids = [rep_ids[m][s] for m in MODEL_ORDER for s in notebook_runtime_SEEDS]
        if len(set(ids)) != 25:
            raise ValueError('Reused representation run IDs must be distinct')
    if config['extraction_mode'] == 'extract':
        context_plan = plan_contextual_extraction(bundle_path=bundle_path, bundle_manifest_sha256=config['bundle_manifest_sha256'], experiment_path=experiment_path, experiment_manifest_sha256=config['experiment_manifest_sha256'], training_runs=runs, run_ids=rep_ids, output_root=project_root / 'outputs', device=config['device'], batch_size=config['batch_size'], source_sha256=encoder_sha)
    else:
        context_plan = None
    return {'config': config, 'build_metadata': build_metadata, 'bundle': bundle, 'bundle_path': bundle_path, 'experiment_path': experiment_path, 'project_root': project_root, 'model_runs': runs, 'discovered': discovered, 'representation_ids': rep_ids, 'representation_dirs': reps, 'source_hashes': dict(build_metadata['source_sha256']), 'source_digest': notebook_runtime_build_source_digest(build_metadata), 'encoder_sha256': encoder_sha, 'extraction_plan': context_plan}

def extract_or_reuse(context: dict, *, extractor=None) -> dict:
    """Load all 25 pinned reusable exports, or execute one complete explicit extraction plan."""
    if context['config']['extraction_mode'] == 'reuse':
        return context['representation_dirs']
    c = context['config']
    run_ids = context['representation_ids']
    plan = context['extraction_plan']
    result = run_contextual_extraction_outputs(plan, extractor=extractor)
    context['extraction_result'] = result
    return result['representation_dirs']

def notebook_runtime_score_anchor_q(context: dict, representation_dirs: dict) -> dict:
    return score_anchor_q_batch(context['bundle_path'], context['model_runs'], representation_dirs)

def score_dimsim(context: dict, representation_dirs: dict) -> dict:
    c = context['config']
    analysis = {'boundary_thresholds': list(range(c['dimsim_boundary_start'], c['dimsim_boundary_stop'] + 1, c['dimsim_boundary_step'])), 'boundary_min_side_count': c['dimsim_min_side_count'], 'min_unique_pairs_per_bin': c['dimsim_min_unique_pairs_per_bin']}
    return score_contextual_dimsim_batch(context['bundle_path'], context['model_runs'], representation_dirs, analysis)

def score_random_control(context: dict, representation_dirs: dict) -> dict:
    resolution = resolve_contextual_dimension(representation_dirs, bundle_manifest_sha256=context['config']['bundle_manifest_sha256'], bundle=context['bundle'])
    generated = generate_random_control(sorted((row['occurrence_id'] for row in context['bundle']['occurrences'])), dimension=resolution['dimension'])
    c = context['config']
    analysis = {'boundary_thresholds': list(range(c['dimsim_boundary_start'], c['dimsim_boundary_stop'] + 1, c['dimsim_boundary_step'])), 'boundary_min_side_count': c['dimsim_min_side_count'], 'min_unique_pairs_per_bin': c['dimsim_min_unique_pairs_per_bin']}
    random_sha = context['source_hashes'].get('evaluation/random_control.py')
    if not isinstance(random_sha, str):
        raise ValueError('build metadata lacks random-control source SHA-256')
    return evaluate_random_control(context['bundle'], generated, bundle_manifest_sha256=c['bundle_manifest_sha256'], dimension_resolution=resolution, source_sha256=random_sha, analysis_config=analysis)

def persist_results(context: dict, q_batch: dict, dimsim_batch: dict, random_result: dict) -> dict:
    """Save numerical results only, with no plotting or presentation dependencies."""
    c = context['config']
    run_ids = c['run_ids']
    analyses = ('contextual-anchor-q', 'contextual-dimsim', 'contextual-random-control')
    if not isinstance(run_ids, dict) or set(run_ids) != set(analyses):
        raise ValueError('Explicit Q, DIMSIM, and random run IDs are required')
    inputs = [context['bundle_path'], context['experiment_path'], *context['model_runs'].values()]
    targets = {name: Path(context['project_root']).resolve() / 'outputs' / name / safe_run_id(run_ids[name]) for name in analyses}
    for target in targets.values():
        ensure_output_outside_inputs(target, inputs)
        notebook_runtime_reject_immutable_ancestors(target)
        if target.exists():
            raise FileExistsError(f'Evaluation output target already exists: {target}')
    numerical_ids = {name: run_ids[name] for name in analyses[:2]}
    contract = build_evaluation_report(q_batch, dimsim_batch, c['reference_model'], numerical_ids)

    def jsonable(value):
        if isinstance(value, Path):
            return str(value)
        if isinstance(value, dict):
            return {str(k): jsonable(v) for k, v in value.items()}
        if isinstance(value, (list, tuple)):
            return [jsonable(v) for v in value]
        return value
    data_config = {k: v for k, v in c.items() if k not in {'presentation_id', 'export_formats', 'export_dpi'}}
    resolved = jsonable({**data_config, 'expanded_representation_ids': context['representation_ids'], 'model_ids': list(MODEL_ORDER), 'seeds': list(notebook_runtime_SEEDS), 'source_digest': context['source_digest'], 'analysis_config': {'dimsim_boundary_thresholds': list(range(c['dimsim_boundary_start'], c['dimsim_boundary_stop'] + 1, c['dimsim_boundary_step']))}})
    numerical = write_evaluation_outputs(context['project_root'], contract, run_ids=numerical_ids, input_roots=inputs, source_hashes=context['source_hashes'], resolved_config=resolved)
    random_path = write_random_control_outputs(context['project_root'], run_ids['contextual-random-control'], random_result, input_roots=inputs, source_hashes=context['source_hashes'], resolved_config=resolved)
    return {'numerical': numerical, 'random_control': random_path}
```
