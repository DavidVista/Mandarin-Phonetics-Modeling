# Prepare Mandarin sound sources

Pack or unpack the Chinese Lessons audio archive and build the audited sound manifest used by the acoustic embedding stage. Each operation is controlled by its own explicit flag.


## Project imports

Locate the local project and import the archive and manifest helpers from the shared source preparation API.


```python
from pathlib import Path
import sys

def find_project_root(start=Path.cwd()):
    for candidate in (start.resolve(), *start.resolve().parents):
        if (candidate / 'datasets').is_dir() and (candidate / 'src').is_dir():
            return candidate
    raise RuntimeError('Could not find a project root containing datasets/ and src/')

PROJECT_ROOT = find_project_root()
sys.path.insert(0, str(PROJECT_ROOT / 'src'))
from phonetics.source_preparation import build_sound_manifest, pack_sounds, unpack_sounds
```

## Configuration

The archive SHA pins the unpacked source. Filename aliases normalize the seven known `ü` spellings before canonical pinyin validation; unsupported labels are recorded by the manifest builder.


```python
RUN_PACK = False
RUN_UNPACK = True
RUN_BUILD_MANIFEST = True
SOURCE_DIR = PROJECT_ROOT / 'datasets' / 'source-data' / 'sounds'
ARCHIVE_PATH = PROJECT_ROOT / 'datasets' / 'chinese-lessons.zip'
UNPACK_DESTINATION = SOURCE_DIR
EXPECTED_ARCHIVE_SHA256 = '9952aa628eab590618c944b37487d37e9f50ccc5fb1be1df9f7fb7b8430a6d67'
MANIFEST_SOUND_DIR = UNPACK_DESTINATION
MANIFEST_OUTPUT_DIR = PROJECT_ROOT / 'datasets' / 'source-data' / 'sounds-manifest' / 'chinese-sounds-v1'
SOUND_ALIASES = {
    'lue4.mp3': 'lve4', 'nue4.mp3': 'nve4',
    'luu2.mp3': 'lv2', 'luu3.mp3': 'lv3', 'luu4.mp3': 'lv4',
    'nuu3.mp3': 'nv3', 'nuu4.mp3': 'nv4',
}
```

## Build local sound artifacts

Review paths and the pinned archive hash before enabling an operation. The helpers refuse to overwrite existing outputs.


```python
if RUN_PACK:
    assert 'REPLACE_WITH' not in str(ARCHIVE_PATH)
    print(pack_sounds(SOURCE_DIR, ARCHIVE_PATH))
if RUN_UNPACK:
    assert 'REPLACE_WITH' not in repr((UNPACK_DESTINATION, EXPECTED_ARCHIVE_SHA256))
    print(unpack_sounds(ARCHIVE_PATH, UNPACK_DESTINATION, expected_sha256=EXPECTED_ARCHIVE_SHA256))
if RUN_BUILD_MANIFEST:
    assert 'REPLACE_WITH' not in repr((MANIFEST_SOUND_DIR, MANIFEST_OUTPUT_DIR))
    print(build_sound_manifest(MANIFEST_SOUND_DIR, MANIFEST_OUTPUT_DIR,
                               aliases=SOUND_ALIASES, tone_convention='neutral_unmarked'))
if not RUN_PACK and not RUN_UNPACK and not RUN_BUILD_MANIFEST:
    print('Archive and manifest operations disabled; set one explicit RUN flag after reviewing paths and hash.')
```
